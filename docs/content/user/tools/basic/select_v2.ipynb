## The Essentials

The `select()` function is the primary way to address subsets of a system. It translates human-readable queries or numeric lists into lists of 0-based indices. 

```{note}
This tutorial focuses on how to use the function. For the full grammar rules (logic, proximity, keywords), see the [**Selection Language section**](../../foundations/04_language/selection_grammar.ipynb) of the Foundations guide.
```

````python
import molsysmt as msm
from molsysmt import systems

lysozyme = systems['T4 lysozyme L99A']['181l.bcif.gz']
````

### 1. Attribute-based selection
You can select elements by their names, types, or IDs.

````python
# Select all Carbon Alphas of Isoleucines
indices = msm.select(lysozyme, selection='atom_name=="CA" and group_name=="ILE"')
print(f"Selected atom indices: {indices}")
````

### 2. Spatial selection
You can find atoms based on their physical distance to other elements.

````python
# Select all water molecules within 0.5 nm of the first protein chain
waters_near_prot = msm.select(lysozyme, selection='molecule_type=="water" within 0.5 nm of chain_index==0')
print(f"Found {len(waters_near_prot)} near-protein water molecules.")
````

### 3. Selection at different levels
By default, `select()` returns atom indices. You can change the target level using the `element` argument.

````python
# Select molecules of type protein
prot_mols = msm.select(lysozyme, element='molecule', selection='molecule_type=="protein"')
print(f"Protein molecule indices: {prot_mols}")
````

# View

*Visualizing a molecular system.*

Molecular visualization engines allow us to inspect and interact with molecular systems directly in Jupyter notebooks. MolSysMT provides the function {func}`molsysmt.basic.view` with MolSysViewer as the default interactive 3D visualization engine, while also supporting third-party viewers such as NGLView.

:::{hint}
:class: dropdown

Visit the section {ref}`user-foundations-support-viewers` for an overview of supported visualization backends.
:::

:::{versionadded} 1.0.0
:::

:::{admonition} API documentation
:class: dropdown

Follow this link for a detailed description of the input arguments, raised errors, and returned objects of this function: {func}`molsysmt.basic.view`.
:::


## Basic usage

Let's explore how {func}`molsysmt.basic.view` visualizes a molecular system (`181L`) using MolSysViewer as the default visualization engine:


```python
import molsysmt as msm
```

```python
molsysviewer_htmlfile = '_static/views/tools_basic_view_1.html'
```

```python
msm.view('181L')
```

:::{tip}
:class: dropdown

All methods defined in the {ref}`molsysmt.basic <API basic>` module can also be invoked from the main level of the library. Hence, {func}`molsysmt.view` is the same function as {func}`molsysmt.basic.view`.
:::


As with many other MolSysMT functions, {func}`molsysmt.basic.view` operates on any supported molecular system form and accepts arguments such as `selection` or `structure_indices`:


```python
molsysviewer_htmlfile = '_static/views/tools_basic_view_2.html'
```

```python
msm.view('181L', selection='molecule_name=="BENZENE"')
```

Let's now see another example using a different molecular system form, and applying a spatial selection constraint:


```python
molecular_system = msm.convert('181L', selection='molecule_type=="protein" or molecule_name=="BENZENE"', to_form='openmm.Modeller')
```

```python
molsysviewer_htmlfile = '_static/views/tools_basic_view_3.html'
```

```python
msm.view(molecular_system, selection='all within 7.0 angstroms of molecule_type=="small molecule"')
```

## Working with NGLView

In addition to MolSysViewer, MolSysMT also supports NGLView as a third-party visualization backend. Specifying `viewer='NGLView'` returns an `nglview.NGLWidget` object, which is recognized by MolSysMT as a valid molecular system form.


```python
nglview_htmlfile = None
```

```python
view = msm.view('181L', viewer='NGLView')
```

```python
msm.get_form(view)
```

As such, `nglview.NGLWidget` is recognized as a valid form in MolSysMT and can be passed to any function that operates on molecular systems:


```python
msm.info(view)
```

```python
msm.get(view, box=True)
```

```python
msm.get_form(msm.convert(view, to_form='openmm.Modeller'))
```

```python
msm.structure.get_distances(view, selection='atom_index==0', selection_2='atom_index==20')
```

:class: dropdown

There are more examples on how MolSysMT works with `nglview.NGLWidget` objects in the third-party NGLView guide.

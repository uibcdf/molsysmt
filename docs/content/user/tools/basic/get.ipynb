# Get

*Retrieving attribute values from a molecular system.*

A molecular system is defined by its attributes and the attributes of its elements — such as the number of atoms, group names, bonded atoms, coordinates, box dimensions, and more. However, not all attributes are available in every representation (form) of a system. Molecular systems can take different forms, such as string identifiers, file paths, or Python objects. Each form includes a specific set of attributes and omits others. Regardless of the form, retrieving attribute values is a key operation in any workflow involving molecular models.

:::{hint}
:class: dropdown

Visit the section {ref}`user-foundations-molecular-system-definition` in case you are not familiar with the concepts of "form" or "attribute" in MolSysMT.
:::

The list of attributes defined in MolSysMT for a molecular system can be checked in the section {ref}`Introduction_Attributes`. And the function {func}`molsysmt.basic.get` is the tool to get the attribute values of a molecular system in {ref}`any supported form <Introduction_Forms>`.

:::{versionadded} 1.0.0
:::

:::{admonition} API documentation
:class: dropdown

Follow this link for a detailed description of the input arguments, raised errors, and returned objects of this function: {func}`molsysmt.basic.get`.
:::

Native `molsys.get(...)` shares the public query contract and uses known native
providers directly for explicit index requests. Element aliases, scalar selections and attribute flags are validated and
normalized by default. For example, `molsys.get(element='atoms', selection=1,
atom_index=True)` queries position 1. Invalid output selectors and unknown
attribute keywords fail at the public boundary. Leave `skip_digestion=False`
for user input; a shortcut does not declare that input trusted.


## Basic usage

Let's show how this function works using a PDB system (`1TCD`):

A valid native topology can have atoms and bonds but zero groups, for example an RDKit molecule without residue metadata. In that case, atom-level `group_id`, `group_name`, and `group_type` queries return `None`; `n_groups` remains zero. MolSysMT does not invent residue groups or a ligand label. Check `msm.has_attribute(molsys, "group_name")` before requiring that hierarchy.


```python
import molsysmt as msm
```

```python
molsys = msm.convert('1TCD', to_form='molsysmt.MolSys')
```

```python
msm.info(molsys)
```

Let's retrieve the names of atoms with indices 32, 33, and 34 (0-based). The function {func}`molsysmt.basic.get()` uses the element argument to specify the type of element being queried (e.g., 'atom', 'group', 'molecule', etc.). To filter which elements to include, use the selection argument, which accepts a list of indices or a query string.

```python
names = msm.get(molsys, element='atom', selection=[32,33,34], name=True)
print('Atom names:',names)
```

:::{tip}
:class: dropdown
All methods defined in the {ref}`molsysmt.basic <API basic>` module can be invoked also from the main level of the library. Hence, {func}`molsysmt.get` is the same method as {func}`molsysmt.basic.get`.
:::

We are not limited to retrieving just one attribute. Multiple attributes can be requested in a single call:

```python
names, group_indices, group_names = msm.get(molsys, element='atom', selection=[32,33,34],
                                            name=True, group_index=True, group_name=True)
print('Atom names:', names)
print('Group indices:', group_indices)
print('Group names:', group_names)
```

Let's illustrate now how things can work with other elements:

```python
n_atoms = msm.get(molsys, element='group', selection=[10,11,12], n_atoms=True)
print('Number of atoms:', n_atoms)
```

```python
msm.get(molsys, element='group', selection=[10,11,12], n_atoms=True)
```

```python
names, atom_indices, atom_names = msm.get(molsys, element='group', selection=[10,11,12],
                                            name=True, atom_index=True, atom_name=True)
print('Group names:', names)
print('Atom indices:', atom_indices)
print('Atom names:', atom_names)
```

```python
indices, component_indices = msm.get(molsys, element='component', selection=[55, 56, 57],
                                    index=True, group_index=True)
print('Indices:', indices)
print('Component indices:', component_indices)
```

If no `selection` is specified, the function applies to all elements of the chosen type by default. See for example:

```python
msm.get(molsys, element='atom', n_atoms=True)
```

The function {func}`msm.basic.get()` also supports selection strings using logical or spatial queries. If you're not familiar with selection syntax, visit [User guide > Tools > Basic > Select](select.ipynb). Let's see how ``selection`` works beyond indices lists with some examples:

```python
msm.get(molsys, element='atom', selection='group_index==20', name=True, index=True)
```

```python
msm.get(molsys, element='atom', selection='molecule_type=="protein"', n_groups=True)
```

```python
msm.get(molsys, element='molecule', selection='molecule_type=="water"', n_molecules=True)
```

```python
msm.get(molsys, element='atom', selection='all within 7.0 angstroms of atom_index==0', n_atoms=True)
```

Finally, not all molecular system forms include all attributes. For example, an `openmm.Topology` object has no atom coordinates, and a `file:inpcrd` system may not contain chain names. When an attribute is unavailable, the function returns `None`.

```python
inpcrd_file = msm.systems['pentalanine']['pentalanine.inpcrd']
msm.get(inpcrd_file, n_groups=True) is None
```

## Examples with structural attributes

### Inspecting alternate sites

For forms exposing sparse alternate-site mappings, request `alternate_location=True`. A query such as `msm.get(molsys, selection=[4, 2], structure_indices=[2, 0, 2], alternate_location=True)` returns three dictionaries in the requested structure order, including the repeated structure. Dictionary keys are integer **source atom indices**, so filtering atoms does not renumber keys to `0` and `1`. Each entry's `atom_id` values are string **labels**, independently of those indices. Join a site to the source atom with its key, not its atom ID.

Structures without alternate sites return empty dictionaries, and an absent attribute returns `None`. This query filters evidence; it does not choose an alternate conformation or move coordinates. The doctest in the API reference provides a complete three-atom example.

Extraction or conversion with an atom selection creates a new atom domain: retain its correspondence separately and use the extracted object's indices when querying that object. The source-index rule applies to the system supplied to the query.

## Querying empty connectivity

A geometric molecular system can validly have an empty bond collection. For a
native four-atom MolSys or Topology, `msm.get(molsys, element='atom', n_bonds=True)`
returns `[0, 0, 0, 0]`; `bond_index=True` returns `[[], [], [], []]`.
Explicit atom subsets keep one count or collection per selected atom, and an
empty atom selection returns an empty list. Internal bond indices also remain
empty. With `element='system'` (the default), `n_bonds=True` returns the total `0`.

This differs from an attribute unavailable in the input form, which can return
`None`. Zero observed/stored bonds do not prove complete chemistry and must not
be used to certify chemical typing or a force-field-ready molecular graph.

With `element='group'`, `n_bonds=True` returns one count of distinct incident
bonds per selected group, including bonds to atoms outside that group. An
internal bond contributes once; a bond between two groups contributes once
to each group. Empty connectivity gives zero per selected group and an empty
group selection gives `[]`. Native MolSys and Topology use the selected
`chemical_state`; an explicit state index does not change the stored reference.


## Querying partial hierarchy

A source can contain atoms without group membership. After combining a protein
with an SDF ligand, native atom-level group ID/name/type queries keep real protein
values and return `None` for ungrouped ligand atoms. Group-based molecule/entity
queries propagate missing parents. Missing membership never means group index 0.
A wholly absent group table retains the existing group-identity `None` result;
empty selections on a populated hierarchy return an empty list.

A supported single-record SDF has one structure: `msm.get(source,
n_structures=True)` returns 1, `[0]` selects that structure, and an empty frame
selection returns 0. Other frame indices raise. Use indices independently of IDs.


## Selecting structure metadata

B factors have length-squared units and shape `(n_selected_structures, n_atoms)`. A scalar structure index retains one row; explicit lists retain order and repetitions, and an empty list returns zero rows. With `element="atom"`, you can also select atoms. This independent control uses different values in every structure.


```python
import molsysmt as msm
import numpy as np
from molsysmt import pyunitwizard as puw
molsys = msm.convert(msm.systems['T4 lysozyme L99A']['181l.pdb'], to_form='molsysmt.MolSys')
molsys = msm.extract(molsys, selection=[0, 1])
second = msm.copy(molsys)
msm.append_structures(molsys, second)
msm.append_structures(molsys, second)
msm.set(molsys, b_factor=puw.quantity([[1, 2], [11, 12], [21, 22]], 'angstrom**2'))
factors = msm.get(molsys, structure_indices=[2, 0, 2], b_factor=True)
print(puw.get_value(factors, to_unit='angstrom**2'))
np.testing.assert_allclose(puw.get_value(factors, to_unit='angstrom**2'), [[21, 22], [1, 2], [21, 22]])
assert puw.get_value(msm.get(molsys, structure_indices=[], b_factor=True)).shape == (0, 2)
```

## Querying native domains

You can use `get()` and `info()` on `MolSys`, `Topology`, `Structures`,
`ChemicalStates`, `Interactions` and `MolSysBuilder`. They share the basic-tool
attribute vocabulary, validation and result conventions. `get()` returns a value,
an ordered list of values or an attribute dictionary; `info()` offers a Styler,
a DataFrame or dictionary records. Each object exposes its available domain;
missing attributes retain the basic-tool behavior.

Use named arguments when moving between the free functions and instance methods.
`MolSys`'s existing positional arguments are preserved: native `get()` accepts
`chemical_state` by keyword, and native `info()` accepts `structure_indices`,
`output_type` and `mask` by keyword. For an instance call, omit the free
function's `molecular_system` argument. Leave `skip_digestion=False` for user input.

Explicit index queries use known native providers. Expressions and nested
selections retain the existing selector. Structures quantities follow the
session unit policy and selected axes preserve order and repetitions. Existing
`get_coordinates()`, `get_box()` and `get_time()` remain available. Interaction
occurrences continue to use `query()`, `between_selections()` and `to_page()`;
the shared molecular `get()` does not replace that sparse-query API.

Dictionary forms and file handlers keep the form-agnostic `msm.get()`/`msm.info()`
route. A plain Python dictionary's `.get()` keeps its ordinary Python meaning.

For `TopologyDict` and `MolSysDict`, `msm.get()` reads the atom count,
indices and already stored string ID/name/type fields directly. Missing fields
keep their normal defaults and element inference. Hierarchy counts and complete
`info()` summaries can still require native conversion. For repeated queries
across many attributes, convert once to a resident native object. A dictionary
inventory query does not validate unrelated serialized domains.

See the {ref}`native query API <API Native Queries>` for the actual method signatures.


```python
import numpy as np
import molsysmt as msm
from molsysmt.native import MolSys, Structures, MolSysBuilder
molsys = MolSys(n_atoms=3)
molsys.structures = Structures(
    coordinates=msm.pyunitwizard.quantity(np.arange(18).reshape(2, 3, 3), 'nm'))
assert molsys.get(n_atoms=True) == 3
assert molsys.topology.get(element='atom', selection=[2, 0], atom_index=True) == [2, 0]
selected = molsys.structures.get(element='atom', selection=[2, 0],
    structure_indices=[1, 0, 1], coordinates=True)
assert selected.shape == (3, 2, 3)
np.testing.assert_array_equal(msm.pyunitwizard.get_value(selected, to_unit='nm')[0],
                             [[15, 16, 17], [9, 10, 11]])
assert molsys.chemical_states.get(n_chemical_states=True) == 1
assert MolSysBuilder(molsys).get(n_atoms=True) == 3
summary = molsys.structures.info(output_type='dictionary')
assert summary[0]['n_atoms'] == 3 and summary[0]['n_structures'] == 2
print(molsys.get(n_atoms=True, n_structures=True, output_type='dictionary'))
print(selected.shape)

# Read selected stored atom labels through the same public query API.
atom_data = {'atoms': [
    {'atom_id': '10', 'atom_name': 'N', 'atom_type': 'N'},
    {'atom_id': '20', 'atom_name': 'O', 'atom_type': 'O'}]}
molsys_dict = msm.TopologyDict(data=atom_data)
assert msm.get(molsys_dict, n_atoms=True) == 2
assert msm.get(molsys_dict, element='atom', selection=[1, 0, 1],
               atom_id=True, atom_name=True) == [
                   ['20', '10', '20'], ['O', 'N', 'O']]
```

## Reading typed chemical dictionaries

`ChemicalStatesDict` supports both `msm.get(payload, ...)` and `payload.get(...)`
without decoding the complete collection. Select the chemical state explicitly
when you need its own bond/component axes. Repeated atom positions stay repeated;
unknown atom properties remain None. These reads inspect requested stored columns,
not unrelated states or preparation history. For edits, convert explicitly back to
`ChemicalStates`. See the
{ref}`ChemicalStatesDict foundation <user-foundations-native-world-classes-molsysmt-chemicalstatesdict>`.

```python
chemical_states = msm.ChemicalStates(n_atoms=3)
chemical_states.append_state()
chemical_states.append_state()
chemical_states.add_bonds([[0, 1]], chemical_state=1)
payload = msm.convert(chemical_states, to_form='molsysmt.ChemicalStatesDict')
print(msm.get(payload, chemical_state=1, n_bonds=True, n_components=True))
print(payload.get(element='atom', selection=[2, 0, 2],
    chemical_state=1, component_index=True))
assert payload.get(element='atom', chemical_state=1, formal_charge=True) is None
assert payload.get(reference_chemical_state_index=True) == 0
```

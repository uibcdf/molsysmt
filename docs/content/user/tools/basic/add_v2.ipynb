*Adding elements from one molecular system into another.*

## The Essentials

The `add()` function allows you to harvest elements (atoms, groups, or molecules) from a source system and inject them into a target system. By default, this operation is performed **in-place** on the target system.

````python
import molsysmt as msm
from molsysmt import systems

# Target: An alanine dipeptide
molsys_A = msm.convert(systems['alanine dipeptide']['alanine_dipeptide.h5msm'])

# Source: A valine dipeptide
molsys_B = msm.convert(systems['valine dipeptide']['valine_dipeptide.h5msm'])

print(f"Initial molecules in A: {msm.get(molsys_A, n_molecules=True)}")

# Add B into A
msm.add(molsys_A, molsys_B)

print(f"Final molecules in A: {msm.get(molsys_A, n_molecules=True)}")
````

## Data Integrity

When adding elements, MolSysMT ensures that the **Topological Integrity** is maintained. If the source system has coordinates, they will be appended to the target system structures. 

```{note}
Both systems must be compatible in terms of structure count. If not, you must specify which frames to use via the `structure_indices` argument.
```

# Get form

*Identifying the form of a molecular system.*

A single molecular system can be represented in different forms, such as a file, a string ID, or a Python object. These forms can be converted between one another depending on the attributes needed and the steps of a workflow. The form of a molecular system can be obtained with the function {func}`molsysmt.basic.get_form`.

:::{hint}
:class: dropdown

Visit the section {ref}`Introduction_Forms` if you are not familiar with the concept of "form" in MolSysMT.
:::

:::{versionadded} 1.0.0
:::

:::{admonition} API documentation
:class: dropdown

Follow this link for a detailed description of the input arguments, raised errors, and returned objects of this function: {func}`molsysmt.basic.get_form`.
:::


## Basic usage

Let's show how this function works using different representations of molecular systems:


```python
import molsysmt as msm
```

```python
molsys = '181L'
```

```python
msm.get_form(molsys)
```

:::{tip}
:class: dropdown

All methods defined in the {ref}`molsysmt.basic <API basic>` module can also be invoked from the main level of the library. Hence, {func}`molsysmt.get_form` is the same function as {func}`molsysmt.basic.get_form`.
:::

Let's try with other forms of this and other molecular systems:

```python
molsys_B = msm.convert(molsys, to_form='openmm.Modeller')
msm.get_form(molsys_B)
```

```python
msm.get_form('AceAlaValTyrNme')
```

```python
molsys = msm.systems['pentalanine']['pentalanine.inpcrd']
msm.get_form(molsys)
```

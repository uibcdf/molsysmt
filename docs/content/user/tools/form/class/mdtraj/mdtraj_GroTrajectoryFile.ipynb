## Overview

This optional MDTraj reader exposes coordinates, available time and box, and atom-axis positions. Its registered native MolSys route embeds Structures, not a reconstructed full hierarchy. Attribute queries use an independent read cursor; the original cursor is untouched. This is distinct from native GRO handler parsing and chemistry reconstruction.


## Supported attributes

The adapter declares **11 attributes**:

| Category | Subcategory / Level | Supported Attributes |
|---|---|---|
| **Stable inventory** | Atom level | `atom_index` |
| **Stable inventory** | Counts | `n_atoms` |
| **Structural** | Structures and observables | `structure_index`, `time`, `coordinates`, `n_structures` |
| **Structural** | Box quantities | `box`, `box_shape`, `box_angles`, `box_lengths`, `box_volume` |

A declaration describes adapter delivery, not a guarantee that every source instance has a value. Optional series may be absent. Physical values carry units; request a target unit when extracting numeric magnitudes. `atom_type` denotes an element symbol, not a force-field type.

Coordinate, box, time and structure-count queries use independent read
handles, preserving the caller's original stream position. Atom counts use
provider metadata without reading coordinates. Complete-handler copying opens
a new read cursor at the start of the same backing file with independent
mutable topology; it does not clone the original cursor or duplicate disk
bytes. Subset extraction as a GRO handler remains unsupported; convert to
native Structures when you need selected coordinate arrays. Generic
`msm.get(reader, coordinates=True)` delivers the advertised coordinate quantity.

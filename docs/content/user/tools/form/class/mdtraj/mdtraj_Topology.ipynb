## Overview

An MDTraj Topology has atom/group/chain inventory and declared chemical metadata, without coordinate structures. MolSys conversion separates stable inventory and available ChemicalStates; a native Topology projection contains stable inventory only. No missing bond, charge or force-field assignment is inferred by the declaration.

When converting native group-free or partially grouped systems, external-only
UNL/fallback parents satisfy MDTraj's hierarchy requirement. Source chemistry
and hierarchy remain unchanged. See {ref}`Cookbook_Group_Free_Hierarchy` for the
metadata-selector limits and executed examples.

When you extract an atom subset from an existing `mdtraj.Topology`, only
parents containing retained atoms are included, with dense local atom, group
and chain indices. Chain IDs, group sequence labels (including `resSeq=0`) and
segment IDs, atom serials, known or unknown formal charges, and retained bond
types and orders keep their source values. The source stays unchanged. This
topology form has no coordinates or structure axis.

| **Chemical states** | Connectivity | `bond_index`, `bond_type`, `bond_order`, `bond_is_aromatic`, `bonded_atoms`, `bonded_atom_pairs`, `n_bonds` |
| **Stable inventory** | Counts | `n_atoms`, `n_groups`, `n_chains`, `n_molecules`, `n_entities`, `n_amino_acids`, `n_ions`, `n_waters`, `n_small_molecules`, `n_peptides`, `n_proteins`, `n_dnas`, `n_rnas`, `n_lipids`, `n_polysaccharides`, `n_saccharides` |
| **Chemical states** | Atom assignments | `formal_charge` |

A declaration describes adapter delivery, not a guarantee that every source instance has a value. Optional series may be absent. Physical values carry units; request a target unit when extracting numeric magnitudes. `atom_type` denotes an element symbol, not a force-field type.

Declared connectivity, components and chemical assignments belong to ChemicalStates in native MolSys. Native Topology alone preserves stable inventory, not those chemical fields.

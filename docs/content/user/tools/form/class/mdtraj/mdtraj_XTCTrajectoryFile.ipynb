## Overview

This optional MDTraj reader supplies nm coordinates, available box/time and the stored step label without molecular topology. The only registered direct targets are its own reader form and native Structures; a direct MolSys route is not declared here.

| **Stable inventory** | Counts | `n_atoms` |
| **Structural** | Structures and observables | `structure_id`, `time`, `coordinates`, `n_structures` |
| **Structural** | Box quantities | `box` |

A declaration describes adapter delivery, not a guarantee that every source instance has a value. Optional series may be absent. Physical values carry units; request a target unit when extracting numeric magnitudes. `atom_type` denotes an element symbol, not a force-field type.

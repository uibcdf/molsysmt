## Overview

An MDTraj Trajectory stores coordinates, time, optional periodic boxes and topology in memory. Geometry crosses this provider boundary in nm and ps, normally with float32 coordinate storage. This resident object is not a file-streaming interface. Native MolSys conversion separates available stable inventory, chemistry and geometry.

Conversion to `parmed.Structure` transfers all selected coordinates and available cells. Atom subsets retain source atom order, while explicit structure lists retain their supplied order. ParmEd stores lengths in angstroms and cell angles in degrees; changing the active PyUnitWizard standard units does not change the transferred geometry. This conversion does not infer force-field parameters or retain trajectory times, which ParmEd does not store. See {ref}`Cookbook_Form_Teleportation`.

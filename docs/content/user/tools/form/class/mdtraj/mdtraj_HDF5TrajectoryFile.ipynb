## Overview

This interoperability reader can carry embedded MDTraj topology, coordinates and optional box, time, velocities, temperature and energies. It is not the native H5MSM format. The MolSys route keeps available chemistry in ChemicalStates alongside stable Topology; a Topology-only projection deliberately excludes chemical connectivity.

## Overview

`pytraj.Trajectory` is a supported representation form in MolSysMT. Form adapter and capabilities for pytraj.Trajectory.

Install this optional provider through Conda-forge AmberTools, for example
`mamba install -c conda-forge ambertools`. The standalone pip PyTraj project
is obsolete for current Python. A version label alone does not identify the
extension ABI: MolSysMT checks chain identifiers and rejects an obsolete
`Frame.__del__` build on Python 3.13 and later before constructing a trajectory.
Use a compatible AmberTools bundle; MolSysMT never replaces your environment
automatically. Python-side and native-engine checks do not certify every
AmberTools workflow.

A resident PyTraj coordinate ensemble and topology. Geometry and available time/cell data cross explicit quantity boundaries; stored chemical declarations belong to ChemicalStates in native MolSys. Native conversion is an in-memory operation, not a bounded file-streaming reader. Runtime/provider compatibility is checked separately from the existence of a registered route.

## Overview

`pytraj.Topology` is a supported representation form in MolSysMT. Form adapter and capabilities for pytraj.Topology.

Install this optional provider through Conda-forge AmberTools, for example
`mamba install -c conda-forge ambertools`. Its bundle supplies PyTraj; the
standalone PyPI and AmberMD PyTraj routes are obsolete for current Python.
MolSysMT declares the Conda route explicitly and does not install or replace
packages while converting a system. Supported bundles can expose different
chain-identifier ABIs, which the adapter detects.

A PyTraj atom/group inventory with declared connectivity and available assignments, without coordinate structures. Native MolSys separates stable Topology from ChemicalStates. The registered PyTraj Trajectory route requires explicit unit-bearing coordinates; it does not create a conformation from topology alone.

| **Stable inventory** | Molecule level | `molecule_index`, `molecule_name`, `molecule_id`, `molecule_type` |
| **Chemical states** | Connectivity | `bond_index`, `inner_bonded_atoms` |

A declaration is form-level capability metadata, not evidence that every instance has a stored value. Check actual availability before requiring optional data. Physical quantities retain units. Native atom_type denotes an element symbol; force-field labels and partial charges belong to MolecularMechanics. Stored connectivity and state-dependent assignments belong to ChemicalStates, independently of native Topology.

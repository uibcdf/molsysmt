## Overview

`rdkit.Mol` represents a chemical graph and any stored coordinate conformers.

RDKit support is optional and requires **RDKit 2024.03.4 or newer**. This
baseline supplies the coordinate API needed for native coordinate export and
CIP assignment from coordinates. Follow each tool's documented chemical scope.

Native MolSys conversion separates stable atom inventory, declared ChemicalStates, conformer Structures and available mechanical partial charges. A native Topology-only projection excludes chemistry. A graph without group metadata does not gain invented groups. Reading does not assign a partial-charge model or place hydrogens.

Conformer selection uses zero-based positions in the stored conformer sequence, independently of conformer IDs. Coordinates originate in angstroms and are standardized through PyUnitWizard. Same-form extraction can retain requested atom order and selected conformers; native inventory conversion has its canonical source-order contract. Atom subsets are sanitized on a copy and may be chemically invalid. Whole same-form extraction may reuse the original with copy_if_all=False. SMILES/SMI output carries a chemical graph, not conformer coordinates. Native construction assigns available stereochemical labels on a copy.

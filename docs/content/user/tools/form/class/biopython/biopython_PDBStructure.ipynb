## Overview

This adapter recognizes `Bio.PDB.Structure.Structure`. Native inventory is constructed from the first Bio.PDB model; native structures collect the requested models as the structure axis. Atom coordinates originate in angstroms and are standardized through PyUnitWizard. This route assumes the same atoms in the same order in every selected model; it does not establish correspondence between heterogeneous models.

MolSysMT's optional Biopython baseline is `biopython>=1.80`; this shared
integration baseline includes the alignment-row API used by sequence tools.
The form conversion does not imply that every Bio.PDB field has a native
round-trip representation.

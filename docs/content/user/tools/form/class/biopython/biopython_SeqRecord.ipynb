## Overview

A Biopython `SeqRecord` stores a sequence together with labels, annotations and features. Its group queries address sequence positions. It has no atom inventory or structures, and record metadata is not a chemical-state assignment.

The optional Biopython integration uses `biopython>=1.80`, including the
alignment-row API used by sequence alignment. Native sequence parsing and
one-/three-letter conversion remain independent of this optional provider.


## Supported attributes

The adapter declares **2 attributes**:

| Category | Subcategory / Level | Supported Attributes |
|---|---|---|
| **Topological** | Group Level | `group_index`, `group_name` |

Query these attributes with `msm.get(molsys, element="group", group_index=True, group_name=True)`.
Indices refer to source sequence positions. Names are sequence characters, preserving case and alphabet.
A selection such as `[2, 0, 2]` preserves order and repetitions without renumbering.
A defined empty sequence returns empty lists. Undefined or partially defined sequences return `None` for names.
A record without a sequence returns `None` for both attributes. Record IDs and annotations remain unchanged.

Declarations do not guarantee a value for every instance. `has_attribute` can distinguish undefined or missing sequence content; it does not infer atoms, bonds or coordinates.

## Overview

`cupy_ndarray` is a supported representation form in MolSysMT. Form adapter and capabilities for cupy_ndarray.

This adapter recognizes CuPy-shaped coordinate containers. Physical conversion needs valid length units as well as shape. The CPU XYZ/Structures routes transfer device values to host storage and preserve the quantity unit. Recognition does not establish a working CUDA runtime or validate an arbitrary bare device array as physical molecular coordinates.


## Supported attributes

The adapter declares **5 attributes**:

| Category | Subcategory / Level | Supported Attributes |
|---|---|---|
| **Stable inventory** | Atom level | `atom_index` |
| **Stable inventory** | Counts | `n_atoms` |
| **Structures** | Coordinates and metadata | `structure_index`, `coordinates`, `n_structures` |

A declaration is form-level capability metadata, not evidence that every instance has a stored value. Check actual availability before requiring optional data. Physical quantities retain units. Native atom_type denotes an element symbol; force-field labels and partial charges belong to MolecularMechanics. Stored connectivity and state-dependent assignments belong to ChemicalStates, independently of native Topology.

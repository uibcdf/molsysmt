## Overview

`XYZ` is a supported representation form in MolSysMT. Form adapter and capabilities for XYZ.


XYZ carries a length quantity with shape `(3,)`, `(n_atoms, 3)` or `(n_structures, n_atoms, 3)`. Query coordinates with `msm.get(molsys, element='atom', coordinates=True)`; the result has rank three. The default system-level XYZ coordinate query is unsupported and raises `NotWithThisFormError`. To deliver an existing quantity to a mutable system, use `msm.set(molsys, structure_indices=0, coordinates=quantity)` directly and preserve its declared atom correspondence. See {ref}`cookbook-delivering-coordinates` for both public routes, selected axes and units.

XYZ is a length-bearing coordinate quantity, not an XYZ text file. Supported ranks represent one atom, one structure, or an ensemble; rank-three normalization uses (n_structures, n_atoms, 3). Coordinates do not supply chemical identity, connectivity or mechanical parameters. Native inventory projection can declare unassigned atom rows without inventing chemistry.

The adapter declares **5 attributes**:

| **Stable inventory** | Atom level | `atom_index` |
| **Stable inventory** | Counts | `n_atoms` |
| **Structures** | Coordinates and metadata | `structure_index`, `coordinates`, `n_structures` |

A declaration is form-level capability metadata, not evidence that every instance has a stored value. Check actual availability before requiring optional data. Physical quantities retain units. Native atom_type denotes an element symbol; force-field labels and partial charges belong to MolecularMechanics. Stored connectivity and state-dependent assignments belong to ChemicalStates, independently of native Topology.

## Overview

`pdbfixer.PDBFixer` is a supported representation form in MolSysMT. Form adapter and capabilities for pdbfixer.PDBFixer.

PDBFixer is optional and loaded when you convert to `pdbfixer.PDBFixer`.
If it is unavailable, converting a PDB file or PDB text to this form raises
`LibraryNotFoundError` with installation hints. Converting those inputs to
native MolSysMT forms does not require PDBFixer. If an installed PDBFixer
cannot import one of its dependencies, the original import error propagates.
See {ref}`third-party bridges <user-foundations-support-third-party-bridges>`
for optional-provider availability.

The optional provider requires `pdbfixer>=1.11`. Version 1.11 fixes ACE and
NME cap hydrogen completion in the legacy pH route of
{func}`molsysmt.build.add_missing_hydrogens`. Fixed-state hydrogen addition
(`mode='fixed_chemical_state'`) requires explicit `engine='RDKit'` and
`pH=None`.

An OpenMM-based topology and single coordinate pose held by PDBFixer. Reading or converting this object is separate from chemical preparation. Whole copying can preserve the provider object; selected reconstruction follows supported molecular data and does not promise to preserve pending provider-specific repair plans.

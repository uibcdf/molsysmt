## Overview

`networkx.Graph` is a supported representation form in MolSysMT. Form adapter and capabilities for networkx.Graph.

A graph representation whose native exporter uses canonical atom-index node keys and stores available atom/chemical fields as node/edge metadata. Atom IDs remain string labels in attributes, distinct from node keys. Whole operations preserve this graph; selected molecular extraction is unavailable until its index and chemical-reference remapping contract is defined.

## Overview

`nglview.NGLWidget` is a supported representation form in MolSysMT. Form adapter and capabilities for nglview.NGLWidget.

The optional integration requires `nglview>=3.0.4`, whose HTML exporter supports
the ipywidgets 8 widget-state API. Python-side conversion and serialization
can be checked without a browser; rendering still requires a compatible
notebook/browser frontend. MolSysViewer remains the default documentation viewer.

Use {func}`molsysmt.third_party.nglview.write_html` to export this view with its
required widget resources. Independent molecular widgets in the same kernel
are excluded from that export.

An NGL widget can expose supported molecular data from a single represented component. MolSysMT-created trajectories carry independent topology/chemical sidecars; other inputs use supported structure-string/coordinate routes with their format limits. Multiple graphical components do not by themselves declare one molecular index space.

Coordinates are materialized; camera/representations/browser sessions are not cloned. Copy does not infer missing bonds. Selected extraction follows native canonical atom order and selected structure order. Native sidecars preserve supported declared chemistry; fallback PDB metadata has its format/axis limits.

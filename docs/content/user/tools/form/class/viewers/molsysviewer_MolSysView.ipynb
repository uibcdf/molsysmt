## Overview

`molsysviewer.MolSysView` is a supported representation form in MolSysMT. MolSysViewer visualization native object.

A MolSysViewer view with an accessible native molecular payload can supply molecular data through that payload. An iframe display object alone does not establish resident molecular data. Molecular conversion, copying and extraction do not preserve a complete graphical scene or saved session; those belong to MolSysViewer.

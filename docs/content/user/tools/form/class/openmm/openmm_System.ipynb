## Overview

An OpenMM System stores particles, forces, constraints and default box vectors. It has no positions or atom/group names and is not a molecular inventory. Default box vectors alone do not establish a periodic force policy. A Context conversion needs its own integrator/platform and coordinate choices.


## Supported attributes

The adapter declares **3 attributes**:

| Category | Subcategory / Level | Supported Attributes |
|---|---|---|
| **Stable inventory** | Atom level | `atom_index` |
| **Stable inventory** | Counts | `n_atoms` |
| **Structural** | Box quantities | `box` |

A declaration is not a guarantee that every source instance contains a value. Missing quantities, chemical assignments or engine settings are not invented. Query physical values as quantities and request an explicit target unit when extracting numeric arrays. `atom_type` denotes an element symbol, not a force-field atom type.

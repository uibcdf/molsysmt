## Overview

This OpenMM PDB reader retains its loaded models on the structure axis. Direct coordinate queries preserve requested atom/model order and repetitions. Same-form and MolSys extraction retain source-sorted unique inventory with matching coordinates. The one stored topology box is repeated over selected models; no independent per-model box or heterogeneous-model correspondence is inferred.

The adapter declares **29 attributes**:

| Category | Subcategory / Level | Supported Attributes |
|---|---|---|

| **Chemical states** | Components | `component_index`, `component_name`, `component_id`, `component_type` |
| **Stable inventory** | Chain level | `chain_index`, `chain_name`, `chain_id`, `chain_type` |
| **Stable inventory** | Molecule level | `molecule_index`, `molecule_name`, `molecule_id`, `molecule_type` |
| **Stable inventory** | Entity level | `entity_index`, `entity_name`, `entity_id`, `entity_type` |
| **Structural** | Structures and coordinates | `structure_id`, `time`, `coordinates`, `n_structures` |
| **Structural** | Box quantities | `box` |

A declaration is not a guarantee that every source instance contains a value. Missing quantities, chemical assignments or engine settings are not invented. Query physical values as quantities and request an explicit target unit when extracting numeric arrays. `atom_type` denotes an element symbol, not a force-field atom type.

Connectivity, components and chemical assignments belong to native ChemicalStates when carried into MolSys. A native Topology target contains stable inventory only; those chemical fields must not be assumed to accompany that projection.

## Overview

A Simulation combines an engine System, integrator, Context and topology. Its structural queries capture the current pose, not trajectory history. Conversion or coordinate access does not establish an independent executable copy; runtime settings and available source assignments remain distinct from native stable inventory.

A declaration is not a guarantee that every source instance contains a value. Missing quantities, chemical assignments or engine settings are not invented. Query physical values as quantities and request an explicit target unit when extracting numeric arrays. `atom_type` denotes an element symbol, not a force-field atom type.

Connectivity, components and chemical assignments belong to native ChemicalStates when carried into MolSys. A native Topology target contains stable inventory only; those chemical fields must not be assumed to accompany that projection.

The PDBFixer conversion owns a managed intermediate PDB through eager reading.
Success and preparation/read failure retire that scratch, while the returned
fixer remains in memory. Cleanup errors propagate. A file:pdb conversion
requested as its own output remains a caller-owned result.

Simulation PDB export writes the current context pose and box on a copied
topology. The one-structure box is explicitly converted to OpenMM units;
PDB lengths remain in angstroms independently of the session length policy.
The original topology box is preserved.

Registered routes may require additional optional providers, complete source information or explicit preparation choices. They are not a certification of every possible input, metadata field, chemical interpretation or engine profile. A conversion is not an independent live-runtime copy.

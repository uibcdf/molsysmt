## Overview

An OpenMM Modeller holds topology and positions for one structure. Its MolSys projection separates stable inventory, available chemistry and geometry into native domains. Extracted atoms remain source-sorted and unique, with matching positions; this object is not a trajectory container.

A Modeller stores one structure. When converting from `parmed.Structure`, MolSysMT uses the first selected structure, including its corresponding periodic cell. Atom subsets retain source atom order and the input object remains unchanged. Use an ensemble-capable form such as `mdtraj.Trajectory` or `molsysmt.Structures` to retain multiple structures.

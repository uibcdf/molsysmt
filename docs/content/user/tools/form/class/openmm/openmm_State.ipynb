## Overview

An OpenMM State stores only data requested when it was captured. Missing positions or velocities are not supplied by the form declaration. It has no independent topology and no coordinate mutator. Public and direct assignment refuse with `NotImplementedMethodError`; edit a Context or native Structures instead.


## Supported attributes

The adapter declares **8 attributes**:

| Category | Subcategory / Level | Supported Attributes |
|---|---|---|
| **Stable inventory** | Atom level | `atom_index` |
| **Stable inventory** | Counts | `n_atoms` |
| **Structural** | Structures and coordinates | `structure_id`, `time`, `coordinates`, `velocities`, `n_structures` |
| **Structural** | Box quantities | `box` |

A declaration is not a guarantee that every source instance contains a value. Missing quantities, chemical assignments or engine settings are not invented. Query physical values as quantities and request an explicit target unit when extracting numeric arrays. `atom_type` denotes an element symbol, not a force-field atom type.

These atom indices address the current engine/capture order. Supply an independently declared topology correspondence when attaching a separate molecular hierarchy.

## Overview

`openmm.Topology` stores atoms, groups, chains, declared bonds and an optional
periodic box. It contains no coordinates and has no structure axis.

Use {func}`molsysmt.basic.copy` for an independent topology, or
{func}`molsysmt.basic.extract` to keep selected atoms. Copies and subsets
preserve atom elements, ID labels, group insertion codes, known or
unknown formal charges, retained bond types and orders, and the periodic
box with its units. They do not infer missing chemistry or change the source.

Subsets keep atoms in source order and assign dense local indices. For
example, `selection=[3, 1, 1]` keeps source atoms 1 and 3 once each, together
with their retained groups, chains and internal bonds. Full copies also
preserve empty groups and chains; subsets contain only selected parents.
The shared `structure_indices` argument does not change atom selection.
Extraction and same-form conversion copy all atoms by default;
`copy_if_all=False` returns the original object for `selection="all"`.

| **Chemical states** | Connectivity | `bond_index`, `bond_type`, `bond_order`, `bond_is_aromatic`, `bonded_atoms`, `bonded_atom_pairs`, `inner_bonded_atoms`, `inner_bonded_atom_pairs`, `inner_bond_index`, `n_bonds`, `n_inner_bonds` |
| **Stable inventory** | Counts | `n_atoms`, `n_groups`, `n_chains`, `n_molecules`, `n_entities`, `n_amino_acids`, `n_ions`, `n_waters`, `n_small_molecules`, `n_peptides`, `n_proteins`, `n_dnas`, `n_rnas`, `n_lipids`, `n_polysaccharides`, `n_saccharides` |
| **Structural** | Box quantities | `box`, `box_shape`, `box_angles`, `box_lengths`, `box_volume` |
| **Chemical states** | Atom assignments | `formal_charge` |

A declaration is not a guarantee that every source instance contains a value. Missing quantities, chemical assignments or engine settings are not invented. Query physical values as quantities and request an explicit target unit when extracting numeric arrays. `atom_type` denotes an element symbol, not a force-field atom type.

Connectivity, components and chemical assignments belong to native ChemicalStates when carried into MolSys. A native Topology target contains stable inventory only; those chemical fields must not be assumed to accompany that projection.

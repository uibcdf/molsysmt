## Overview

This OpenMM PSF reader carries topology and an optional declared unit cell, not coordinate structures. Native MolSys conversion retains available declared chemistry alongside stable inventory. This adapter does not register a parameterized System conversion or infer missing force-field assignments.


## Supported attributes

The adapter declares **9 attributes**:

| Category | Subcategory / Level | Supported Attributes |
|---|---|---|
| **Stable inventory** | Atom level | `atom_index`, `atom_name`, `atom_id`, `atom_type` |
| **Stable inventory** | Group level | `group_index`, `group_name`, `group_id`, `group_type` |
| **Structural** | Box quantities | `box` |

A declaration is not a guarantee that every source instance contains a value. Missing quantities, chemical assignments or engine settings are not invented. Query physical values as quantities and request an explicit target unit when extracting numeric arrays. `atom_type` denotes an element symbol, not a force-field atom type.

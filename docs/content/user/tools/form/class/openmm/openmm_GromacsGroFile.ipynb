## Overview

This OpenMM GRO reader supplies format-limited atom/group labels, coordinate models and boxes. Its native MolSys conversion assembles inventory and selected structures before one coherent atom extraction. It does not invent partial charges, complete chemistry or force-field parameters absent from the coordinate source.

A declaration is not a guarantee that every source instance contains a value. Missing quantities, chemical assignments or engine settings are not invented. Query physical values as quantities and request an explicit target unit when extracting numeric arrays. `atom_type` denotes an element symbol, not a force-field atom type.

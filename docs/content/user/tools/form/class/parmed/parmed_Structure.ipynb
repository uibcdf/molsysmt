## Overview

A ParmEd molecular inventory with optional coordinates, cells and force-field assignments. Native import keeps chemical assignments in ChemicalStates and available force-field labels/partial charges in MolecularMechanics. Native atom_type denotes the element, not the provider force-field type. Conversion does not parameterize the system.

Conversions to MDTraj keep atom names, coordinates and periodic cells aligned. Atom subsets retain source atom order; a structure list such as `[2, 0]` also reorders the cells. One stored cell can be shared by all structures. Native `MolSys` conversion keeps partial charges in `molecular_mechanics`. OpenMM Modeller represents only the first selected structure and its corresponding cell. These bridges do not assign a force field. See {ref}`Cookbook_Form_Teleportation` for an executable MDTraj–ParmEd–OpenMM example.

ParmEd remains optional (`parmed>=4.3.1`) and is loaded when a requested
conversion needs it. ParmEd 4.3.0 fixes coordinate and box assignment with
NumPy 2; 4.3.1 also removes the obsolete `numpy.compat` import absent from
NumPy 2.3 and newer. See {ref}`dependency management
<user-foundations-governance-dependency-management>` for availability checks.

Native `MolSys` export retains selected coordinate structures and their aligned
canonical periodic cells, including requested structure order and repetitions.
ParmEd stores coordinates and cell lengths in angstroms and cell angles in
degrees. Explicit ChemicalStates supply nullable formal charges, atom
aromaticity, numeric bond orders and independent qualitative aromatic/dative
types. Available MolecularMechanics partial charges and force-field type labels
are retained without assigning force-field parameters. Representable integer
atom/group ID labels and chain labels are preserved. Atom subsets keep source
atom order.

A native `Topology` exports stable inventory only, even when attached to a
`MolSys`; it does not recover that system's chemistry or coordinates.
Arbitrary atom/group labels, unknown numeric bond orders, partial-charge arrays
with unknown entries, and rotated or left-handed cells require an explicit
limitation report or rejection. Export does not rotate coordinates to change
the cell convention. Use `return_report=True` to inspect audited losses and
`strict=True` to reject detected loss before construction; report scopes still
identify unassessed semantics. See {ref}`Tutorial_Convert`.

## Overview

The native class `molsysmt.native.PDBFileHandler` loads text and parsed records
into memory. This is a resident reader, not an implemented streaming writer.
Its raw record coordinates use angstroms; public quantity queries follow
PyUnitWizard. See {ref}`user-foundations-native-world-file-handlers-molsysmt-pdbfilehandler`.

Native MolSys/Topology conversion accepts `from_naming_scheme` before classifying fresh groups or running native inference. The caller-owned handler stays open, and its parsed records are not rewritten: independent conversions may declare different conventions without leaking one into the next. Semantic selections use interpreted inventory. MolSys reader history retains the original source declaration; pure Topology output carries inventory and naming metadata only. Unsupported contextual OpenMM interpretation retains the explicit-refusal versus optional-warning policy. See {ref}`user-foundations-group-naming-schemes` and {ref}`Tutorial_Infer_Covalent_Bonds`.

## Overview

MolecularMechanics stores per-atom parameters, system settings and available
assignment reports. It is a parameter container with experimental scope, not a
coordinate trajectory or a molecular-dynamics engine. Molecular identity and
chemical-state assignments remain in their separate native domains.


## Supported attributes

The adapter exposes **21 declared attributes**. Check
instance availability with {func}`molsysmt.form.has_attribute` before requesting
a field.

| Category | Subcategory / Level | Supported Attributes |
|---|---|---|
| **Atom axis** | Positions | `atom_index` |
| **Legacy assignment** | Chemical state | `formal_charge` |
| **Topological** | Element Counts | `n_atoms` |
| **Structural** | Other Attributes | `atom_index`, `n_atoms` |
| **Mechanical** | Force Field Parameters | `partial_charge`, `atom_ff_type`, `forcefield`, `water_model`, `implicit_solvent` |
| **Mechanical** | Non-Bonded & Solvation | `non_bonded_method`, `cutoff_distance`, `switch_distance`, `dispersion_correction`, `ewald_error_tolerance`, `solute_dielectric`, `solvent_dielectric`, `salt_concentration`, `kappa` |
| **Mechanical** | Constraints & Dynamics | `hydrogen_mass`, `constraints`, `flexible_constraints`, `rigid_water` |

An atom axis comes from stored per-atom parameter rows, not coordinates. The
container has no structures. Its legacy formal-charge interface does not move
chemical-state ownership out of ChemicalStates. Mechanical parameters and their
assignment evidence do not establish a fully validated force-field model. This
container retains its declared experimental scope before 1.0.

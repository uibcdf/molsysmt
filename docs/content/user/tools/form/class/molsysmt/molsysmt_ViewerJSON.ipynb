## Overview

`molsysmt.ViewerJSON` is a supported representation form in MolSysMT. Form adapter and capabilities for molsysmt.ViewerJSON.

Native topology and MolSys exports contain ordinary JSON scalars, including nulls
for missing table values. Bond endpoints are integer atom indices. You can
serialize the returned payload with `payload.dumps()` or standard
`json.dumps(payload.to_dict())`; no NumPy-specific encoder is needed.

`msm.extract(payload, selection=[2, 1], structure_indices=[0])` returns an
independent graphics subset in that atom order. Only bonds whose two endpoints
are retained survive; atom pairs, endpoint aliases and aligned bond columns
are remapped together. Atom indices must be unique. Structure indices may repeat
and retain their requested order. Empty selections are supported.


## Supported conversions

The adapter has **2 registered direct conversion routes**:

| Target Form | Form Type | API Documentation |
|---|---|---|

| `molsysmt.ViewerJSON` | `class` | [to_molsysmt_ViewerJSON()](https://github.com/uibcdf/molsysmt/blob/ebded7917735c6e326e250831f1750fc5ed6cf9f/molsysmt/form/molsysmt_ViewerJSON/to_molsysmt_ViewerJSON.py#L5) |

Native recovery keeps IDs as strings and uses optional per-atom `group_index`,
`chain_index` and `entity_index` columns for identity. Native exports supply
these columns, so repeated labels remain distinct. Legacy payloads group by
chain/entity labels and by group label within a chain; they cannot recover
identities absent from those labels. Invalid or contradictory declarations
raise a format error. The graphics schema omits full molecule metadata; use
H5MSM for complete-domain recovery. See
{ref}`ViewerJSON <user-foundations-native-world-classes-molsysmt-viewerjson>`.

Bond queries and native recovery accept canonical pairs or legacy `indexA`/`indexB`
columns, including a wrapper containing one bond definition. Empty collections
have zero bonds. Multiple sets, conflicting endpoint encodings and misaligned
bond columns raise a format error. To choose a set, copy the payload and assign
`chosen.data["bonds"] = chosen.data["bonds"]["sets"][set_index]` before conversion.
Graphics extraction still keeps separate sets; they are not inferred chemical states.

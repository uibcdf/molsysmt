## Overview

The native handler eagerly parses the first GRO entry into `handler.entry`.
It retains the resident text buffer and provides native conversion/read routes;
it does not stream a concatenated coordinate series or write GRO records.
The registered form name is `molsysmt.GROFileHandler`; construct the Python class
from `molsysmt.native`.

Declarations are query capabilities, not a guarantee that every GRO file stores
every value. GRO supplies labels, coordinates, optional velocities and a box;
connectivity is inferred by supported native conversion policy, and higher-level
values may be derived. A declared `time` attribute does not make the native
handler a coordinate-series reader. Attribute availability does not imply an
editing route. See the Foundations guide for naming and inference limits.

Whole-handler `copy`, `extract` and default identity conversion now return
independent resident buffers and parsed entries. `copy_if_all=False` explicitly
permits aliasing. Partial handler extraction and writing remain unsupported.
Access parsed fields through `handler.entry`, for example `entry.header` and
`entry.n_atoms`; the handler reads one GRO entry. See
{ref}`user-foundations-native-world-file-handlers-molsysmt-grofilehandler`.

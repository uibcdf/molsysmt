## Overview

MolSysBuilder holds editable native inventory, chemical states and structures
for constructing a MolSys. Use its builder methods to add molecular inventory
and its supported setters to update existing fields. Conversion builds or
serializes the available domains; it does not calculate missing chemistry or
force-field parameters.

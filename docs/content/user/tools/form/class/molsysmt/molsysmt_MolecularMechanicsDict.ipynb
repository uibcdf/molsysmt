## Overview

`molsysmt.MolecularMechanicsDict` is an ordinary Python dictionary of accepted
mechanical parameters. It can contain per-atom arrays or system parameters
without coordinates. Values may include PyUnitWizard quantities; use an
explicit supported serialization route rather than assuming a JSON payload.


## Supported attributes

The adapter exposes **21 declared attributes**. Declaration is a form-level
capability inventory; use {func}`molsysmt.form.has_attribute` to check a particular
payload before requesting its data.

| Category | Subcategory / Level | Supported Attributes |
|---|---|---|
| **Atom axis** | Positions | `atom_index` |
| **Legacy assignment** | Chemical state | `formal_charge` |
| **Topological** | Element Counts | `n_atoms` |
| **Structural** | Other Attributes | `atom_index`, `n_atoms` |
| **Mechanical** | Force Field Parameters | `partial_charge`, `atom_ff_type`, `forcefield`, `water_model`, `implicit_solvent` |
| **Mechanical** | Non-Bonded & Solvation | `non_bonded_method`, `cutoff_distance`, `switch_distance`, `dispersion_correction`, `ewald_error_tolerance`, `solute_dielectric`, `solvent_dielectric`, `salt_concentration`, `kappa` |
| **Mechanical** | Constraints & Dynamics | `hydrogen_mass`, `constraints`, `flexible_constraints`, `rigid_water` |

Per-atom parameter arrays establish atom positions and counts; system-only
parameters do not supply an atom axis. This form has no structures. Its legacy
`formal_charge` key does not move chemical-state ownership out of
ChemicalStates. Distinguish that assignment from a force-field partial charge.

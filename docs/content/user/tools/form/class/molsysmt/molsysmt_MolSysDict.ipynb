## Overview

`molsysmt.MolSysDict` is a native payload wrapper, not an arbitrary Python
dictionary. Its schema records molecular domains for conversion to native
objects or YAML. Version 0.2 separates inventory from chemical states and
structures; the 0.1 reader retains legacy declarations. Query availability
depends on the payload, its schema and the requested domain.

Schema 0.2 stores stable inventory, chemical states and structures in separate
domains; an absent domain supplies no data. Schema 0.1 remains readable as a
legacy declaration. The table combines these adapter declarations; it does not
promise every field in every schema or payload. Stored charges, H assignments,
bonds and components belong to ChemicalStates, even where historical getter
module names include topological. This adapter does not advertise general
mechanical-parameter queries.

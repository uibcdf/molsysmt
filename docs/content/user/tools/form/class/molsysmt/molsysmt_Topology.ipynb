## Overview

`molsysmt.Topology` is the registered form for stable atoms, groups, molecules,
entities and chains. Construct the Python class from `molsysmt.native`.
Coordinates and structural observations belong to `Structures`; bonds,
components and chemical assignments belong to `ChemicalStates`. Attaching a
Topology to MolSys does not restore a chemical or structural facade. Retain the
complete MolSys when those independent domains are needed.

Declarations do not guarantee populated labels or complete hierarchy. IDs are
string labels; indices are positional. Stable molecule membership does not
change with later state-local bond edits. Formal charges, aromatic assignments,
bonds, chemical components and chemical-state indices are not Topology
capabilities. Query the appropriate ChemicalStates domain instead.

For resident objects, `topology.set_atom_molecules(molecule_indices,
atom_indices='all')` declares stable ownership for ungrouped atoms. Use `None`
for unknown membership; grouped positions must agree with their group's
molecule. A partial assignment accepts unique atom indices. This native method
is separate from form adapter operations. See {ref}`API Native Molecule Membership`
and the executed {ref}`Cookbook_Group_Free_Hierarchy` recipe for complete,
partial and cleared assignments, preservation through bond edits and invalidation
of attached analyses.

Native `get()`/`info()` provide the common query experience without universal form
discovery. Query keywords such as `atom_index=True` are Boolean flags; label
strings belong to selections or assignments. For example:

```python
import molsysmt as msm
from molsysmt.native import Topology

molsys = Topology(n_atoms=3)
assert molsys.get(element='atom', selection=[2, 0], index=True) == [2, 0]
chunks = [row['atom_index'] for row in msm.Iterator(
    molsys, element='atom', atom_index=True, chunk=2, output_type='dictionary')]
assert chunks == [[0, 1], [2]]
```

The form-specific inventory iterator queries selected row values once and keeps
those arrays in memory. Explicit indices preserve order and repetitions;
start/stop/step address that selected sequence, and chunk groups its results.
It yields fresh result mappings/lists. Aggregate counts and unavailable values
are refused as row fields. Public `msm.Iterator` retains its existing outer
result-container conventions; consume or copy each returned mapping as needed.
No chemical state, coordinates or optional engine is constructed by this loop.

Use the native ChemicalStates methods for bond edits. The former Topology
add_bonds/remove_bonds adapters are retired, not alternative chemical stores.

A registration does not certify complete chemical or geometric fidelity for
every input. Optional targets require their declared providers; geometry-bearing
writers and viewers need explicit coordinates. Converting a pure Topology to
MolSys preserves inventory without inventing chemistry or coordinates. Its
TopologyDict projection is stable inventory schema 0.2/0.3, distinct from the
legacy 0.1 reader that can carry chemistry. Use a complete source MolSys when
chemical assignments must accompany conversion. Inspect
{func}`molsysmt.basic.get_conversion_report` and each target's prerequisites;
partial audit coverage is not a universal losslessness or execution guarantee.

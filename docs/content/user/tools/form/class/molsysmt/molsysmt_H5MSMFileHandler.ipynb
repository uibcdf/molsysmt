## Overview

The registered form identifies a live legacy H5MSM 0.3/0.4 handle. Construct
it with `from molsysmt.native import H5MSMFileHandler`; opening legacy input
emits its deprecation warning. The class rejects modular 0.5/0.6/1.0 roots.
Public default 1.0 files instead use the resident read/layer APIs described in
{ref}`user-tools-form-h5msm-10`. Close handles explicitly or use a context manager.
The class exposes `file` and `format_version`; HDF5 groups are reached through
`handler.file`, not native Topology/Structures properties.

## Overview

`molsysmt.StructuresDict` is an ordinary Python dictionary of recognized
structural fields, including NumPy arrays and PyUnitWizard quantities. It is
an in-memory representation, not the versioned MolSysDict schema or a promise
of direct JSON serialization. It supplies coordinates and available
observables without declaring atom names, groups or chemical connectivity.


## Supported attributes

The adapter exposes **20 declared attributes**. Declaration is a form-level
capability inventory; use {func}`molsysmt.form.has_attribute` to check a particular
payload before requesting its data.

| Category | Subcategory / Level | Supported Attributes |
|---|---|---|
| **Topological** | Atom Level | `atom_index` |
| **Topological** | Element Counts | `n_atoms` |
| **Structural** | Coordinates & Conformations | `structure_index`, `structure_id`, `coordinates`, `velocities`, `n_structures` |
| **Structural** | Periodic Box | `box`, `box_shape`, `box_angles`, `box_lengths`, `box_volume` |
| **Structural** | Trajectory Metadata | `time`, `potential_energy`, `kinetic_energy`, `total_energy`, `temperature` |
| **Structural** | Crystallographic & Bioassembly | `b_factor`, `alternate_location`, `occupancy` |
| **Structural** | Other Attributes | `atom_index`, `n_atoms` |

Coordinates use the structure-first shape `(n_structures, n_atoms, 3)`; boxes
use `(n_structures, 3, 3)`. Length and time fields carry PyUnitWizard quantities
with compatible units; nm and ps are defaults, not fixed output-unit choices.
Atom positions and counts do not establish names, groups or chemical identity.

These routes have different fidelity. Structures reads the available arrays;
MolSys embeds those structures with no inferred topology or chemistry and an
empty mechanical domain. The explicit Topology route creates atom-count-only
inventory, not chemical identity. The MolecularMechanics route returns an
empty parameter container. Same-form conversion uses the whole-payload extract
hook and does not implement subset extraction; convert to Structures before
selecting a subset. File routes require the fields and optional writer
dependencies they consume; YAML performs schema and unit validation.

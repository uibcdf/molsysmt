## Overview

`molsysmt.TopologyDict` is a native payload wrapper for stable inventory.
Versions 0.2 and 0.3 keep chemical state outside Topology; the legacy 0.1
reader also accepts chemical declarations. Select the destination domain
explicitly when converting legacy records.

its chemical declarations; a Topology projection contains inventory only.

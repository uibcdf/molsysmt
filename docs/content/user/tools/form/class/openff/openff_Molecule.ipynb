## Overview

An OpenFF molecular graph with optional coordinate conformers and partial charges. Native MolSys conversion separates stable inventory, ChemicalStates, available Structures and mechanical partial charges. Reading a declared charge is different from assigning a charge model or preparing the molecule.

This optional integration uses OpenFF Toolkit >=0.17.1 and OpenFF Units >=0.3.0.
The supported acquisition route is Conda-forge, for example
`mamba install -c conda-forge openff-toolkit-base openff-units`. These providers
are kept separate from MolSysMT's pip `soft` extra. Choose builds compatible
with your Python and Pint versions; the literal oldest releases have narrower
interpreter/unit-library constraints than the current providers.

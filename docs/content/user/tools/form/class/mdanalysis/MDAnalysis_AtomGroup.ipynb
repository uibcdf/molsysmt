## Overview

An ordered view of atoms belonging to an MDAnalysis Universe. Atom indices in this form are positions within that view, not necessarily indices in its parent Universe. Available attributes depend on the parent topology and reader.

Atom-index selections refer to positions within the input AtomGroup and preserve their requested order. Independent copying, extraction and conversion back to the same form retain source labels, formal charges, numerical bond orders, long or unknown stored bond type labels and per-bond guessed flags. Selected bonds retain their source declarations; no bond perception runs. Available coordinates, velocities, forces, cells and time follow the selected structures without changing the source Universe or its reader position.

Unrestricted extraction or same-form conversion can reuse the original AtomGroup with `copy_if_all=False`. Unsupported selected time axes raise explicitly; a native `molsysmt.MolSys` can retain irregular time. Provider extensions require their own preservation contract.

An AtomGroup copy clones its parent Universe and returns the same ordered view of that independent parent, keeping unselected parent atoms as context. Extraction and same-form conversion instead materialize the selected atoms in a new Universe.

Explicit empty atom or structure selections are unsupported by the in-memory subset builder. A full copy of a topology-only Universe retains its topology without creating structures.

The direct coordinate getter can delegate through a resident native MolSys. Conversion to native inventory has its own atom-order contract; do not infer it from same-form view/copy ordering. This adapter has no structure iterator export.

## Overview

An MDAnalysis molecular topology with an optional structure reader. Native MolSys conversion separates stable inventory, stored chemical declarations and available geometry. Provider positions and cells use angstroms, velocities angstroms per picosecond and time picoseconds; PyUnitWizard controls the resulting quantity policy.

Explicit atom-index selections preserve their requested order and retain source labels. Copying and extraction preserve formal charges, numerical bond orders, long or unknown stored bond type labels and each bond's guessed flag. Labels are retained without chemical reinterpretation. Available coordinates, velocities, forces, periodic cells and time follow the selected structures; the source and its reader position remain unchanged.

Unrestricted extraction or conversion to the same form can reuse the source with `copy_if_all=False`. A selected time axis that the MDAnalysis in-memory reader cannot represent raises instead of changing the stored times; convert to `molsysmt.MolSys` to retain irregular time. Provider extensions require their own preservation contract.

Explicit empty atom or structure selections are unsupported by the in-memory subset builder. A full copy of a topology-only Universe retains its topology without creating structures.

The structure iterator first materializes selected native Structures and only then yields native chunks. It does not bound the initial reader-to-native allocation. Coordinate, velocity, box and time getters restore the active source structure after reading. Native MolSys conversion uses canonical sorted/unique source atom indices with geometry aligned to that inventory.

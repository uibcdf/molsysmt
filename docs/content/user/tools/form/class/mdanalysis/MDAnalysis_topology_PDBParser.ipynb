## Overview

A filename-backed MDAnalysis PDB topology parser, not a resident molecular ensemble. It declares no direct molecular attributes in this adapter. Conversion to MDAnalysis Topology calls parse(); the native MolSys/Topology routes reopen its filename through a Universe. Copying the parser creates another parser for the same file, not an independent copy of its contents.

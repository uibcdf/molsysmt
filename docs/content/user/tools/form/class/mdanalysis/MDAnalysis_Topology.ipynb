## Overview

An MDAnalysis atom/group/segment hierarchy with optional stored chemical fields, without coordinate structures. Conversion to MolSys separates stable Topology from available ChemicalStates; a Topology-only projection excludes the chemical domain. MDAnalysis residues map to MolSysMT groups and segments supply represented chains.

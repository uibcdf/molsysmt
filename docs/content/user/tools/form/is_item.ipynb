# Is item

*Validating whether an object is a recognized molecular system item of any supported form.*

The function {func}`molsysmt.form.is_item` inspects a Python object and checks if it conforms to any registered form in MolSysMT.

:::{versionadded} 1.0.0
:::

:::{admonition} API documentation
:class: dropdown

Follow this link for a detailed description of the input arguments, raised errors, and returned objects of this function: {func}`molsysmt.form.is_item`.
:::


## Basic usage

Let's test objects against recognized item forms:


```python
import molsysmt as msm
```

```python
molsys = msm.convert(msm.systems['Barnase-Barstar']['barnase_barstar.h5msm'])
```

:::{note} Demo Systems Catalog
:class: dropdown

This tutorial uses demonstration datasets provided by MolSysMT. To explore the full catalog of bundled systems, forms, and file paths, visit the {ref}`Demo Systems <user-foundations-entrance-demo-systems>` guide.
:::


```python
print('Is molsys instance an item?:', msm.form.is_item(molsys))
print('Is file path an item?:', msm.form.is_item(msm.systems['Barnase-Barstar']['barnase_barstar.h5msm']))
print('Is plain integer an item?:', msm.form.is_item(42))
print('Is arbitrary dictionary an item?:', msm.form.is_item({'foo': 'bar'}))
```

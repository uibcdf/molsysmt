# Is string

*Checking whether a molecular system item or form name corresponds to an in-memory string format.*

The function {func}`molsysmt.form.is_string` tests if a form or item represents a string sequence, PDB text block, chemical SMILES, or accession ID.

:::{versionadded} 1.0.0
:::

:::{admonition} API documentation
:class: dropdown

## Basic usage

Let's test string form classifications:


```python
import molsysmt as msm
```

```python
print('Is "string:amino_acids_1" a string form?:', msm.form.is_string('string:amino_acids_1'))
print('Is "string:smiles" a string form?:', msm.form.is_string('string:smiles'))
print('Is "file:pdb" a string form?:', msm.form.is_string('file:pdb'))
print('Is "molsysmt.MolSys" a string form?:', msm.form.is_string('molsysmt.MolSys'))
```

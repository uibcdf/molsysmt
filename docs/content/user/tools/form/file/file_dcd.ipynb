Whole-file conversion accepts `output_filename`; omitted destinations are
session-owned temporaries. Explicit destinations reject source aliases.
Select atoms/structures when converting to `molsysmt.Structures` or H5MSM;
writing the selection back to this file format is unsupported. See
{ref}`Cookbook_Copying_Disk_Systems` for copy lifetime and selected output.

These links describe adapter hooks. Registration is not a promise of every
public workflow; required attributes, optional engines and supported operation
limits still apply. Public byte-copy ownership is described by
{func}`molsysmt.basic.copy`.

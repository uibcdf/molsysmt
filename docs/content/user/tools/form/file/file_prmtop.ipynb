## Overview

`file:prmtop` is a supported representation form in MolSysMT. AMBER parameter/topology file format

Box queries reuse the OpenMM AMBER reader and general periodic geometry tools.
A periodic topology supplies one box record; a nonperiodic topology supplies None.
This metadata does not create coordinates: conversion to MolSys still has zero
coordinate structures unless a coordinate source is supplied.

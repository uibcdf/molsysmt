## Overview

`file:trjpk` is a supported representation form in MolSysMT. Form adapter and capabilities for file:trjpk.

The existing pickle layout stores coordinates and box vectors in nm, time in ps,
and optional structure ID labels. Reading materializes the stored arrays. Queries
accept nonconsecutive or repeated positional indices; IDs remain separate labels.


## Supported attributes

The `file:trjpk` form container supports a total of **7 attributes**:

| Category | Subcategory / Level | Supported Attributes |
|---|---|---|
| **Topological / Structural** | Positional Atom Axis | `n_atoms`, `atom_index` |
| **Structural** | Structure Count | `n_structures` |
| **Structural** | Coordinates & Conformations | `structure_id`, `coordinates` |
| **Structural** | Periodic Box | `box` |
| **Structural** | Trajectory Metadata | `time` |


## Implemented operations

The adapter provides atom counts and positional indices from its header, and
queries for coordinates, box vectors, time and structure IDs through the existing
StructuresDict reader. Bulk structural queries reuse that conversion.

Export from StructuresDict through {func}`molsysmt.basic.convert` preserves
selected atom and structure order, repetitions and empty axes. The two header
counts describe the actual selected payload; box, time and ID arrays follow the
same structure selection. Absent optional fields stay None.

The legacy file extraction, append, merge, setters and iterator placeholders do
not provide subset editing or bounded streaming. To export a subset, load a
StructuresDict and convert that dictionary to a new TRJPK filename.

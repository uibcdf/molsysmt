## Overview

`file:crd` is a supported representation form in MolSysMT. CHARMM card (CRD) file format with coordinates.


## Supported attributes

The `file:crd` form container supports a total of **15 attributes**:

| Category | Subcategory / Level | Supported Attributes |
|---|---|---|
| **Topological** | Atom Level | `atom_index`, `atom_name`, `atom_id`, `atom_type` |
| **Topological** | Group Level | `group_index`, `group_name`, `group_id`, `group_type` |
| **Topological** | Element Counts | `n_atoms`, `n_groups` |
| **Structural** | Coordinates & Conformations | `structure_id`, `coordinates`, `n_structures` |
| **Structural** | Periodic Box | `box` |
| **Structural** | Trajectory Metadata | `time` |
| **Structural** | Other Attributes | `atom_index`, `n_atoms` |


An atom-index-only query reads the header count without loading coordinates or constructing topology. Atom and group metadata use the native `Topology` conversion route; atom, group and chain IDs are strings, while indices are positions. Queries preserve source positions and selection order, including repetitions and empty selections. Coordinates retain physical units, with angstroms read from CRD.

Converting a selected subset to `Topology` or `MolSys` uses the established native sorted atom-index order. In `MolSys`, coordinates follow that same order. A coordinate-only `Structures` conversion retains the requested atom order. Empty structure selections produce zero structures. This reader does not infer connectivity, a box, a time or a complete chemical state.

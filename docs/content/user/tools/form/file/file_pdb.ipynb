## Overview

`file:pdb` is a supported representation form in MolSysMT. Protein Data Bank file format


Native conversion to MolSys accepts `get_missing_bonds=False`.
This retains only PDB-declared edges and their explicit evidence, without
invoking OpenMM. File and PDBFileHandler routes honor the same option. Their
existing `True` default attempts optional OpenMM inference and can retain only
declared edges when the legacy engine is unavailable or fails. Neither reading
nor connectivity inference establishes complete bond orders, charges or
protonation. Coordinates and box lengths follow configured PyUnitWizard units.
See {ref}`cookbook-pdb-connectivity-policy` for explicit source preparation.

RCSB downloads and PDB-ID/AlphaFold conversions use private staging beside the
requested destination. Download or extraction failure preserves an existing
file; successful publication replaces it. The parent directory must be writable.
Cleanup errors remain visible, including after a completed publication.
The PDB-text adapter retires a generated output on failed write/close, but retains
explicit outputs, which may contain partial content. Successful files belong to
you. See {ref}`user-foundations-support-forms-files` for the file ownership contract.

The first ChemicalStates preparation report also retains typed PDB declaration
diagnostics. Inspect `report['declaration_issues']['records']` for original
line numbers, canonical endpoint indices, statuses, symmetry and alternate labels.
Missing/ambiguous endpoints and unrepresented symmetry images are distinguished
from redundant records and conditional canonical-site mappings. These historical
indices are not remapped after extraction; accepted edges do not establish bond
order, bond kind, valence or complete alternate-specific chemistry. Public
`msm.convert(..., return_report=True)` reports adapter limitations; `strict=True`
refuses them. See {ref}`cookbook-pdb-connectivity-policy`.

Native PDB conversion accepts `from_naming_scheme` before classifying fresh groups and running the selected native inference method. No convention is guessed. File-wide declarations apply to every group; use per-group native declarations for mixed sources. Context-dependent selections use the interpreted inventory. The historical reader report retains `source_naming_scheme`. Unsupported contextual OpenMM naming requires an adapter: explicit requests raise before provider execution, while the legacy optional route warns and archives the cause. Default engine choices remain unchanged. Pure Topology output projects stable inventory and naming declarations; bonds and preparation history belong to ChemicalStates in MolSys. See {ref}`user-foundations-group-naming-schemes` and {ref}`Tutorial_Infer_Covalent_Bonds`.

Opaque `REMARK` records without an integer number do not block coordinate
queries. The native PDB handler retains their full trimmed text with
`remarkNum=None`, including long GHECOM command lines. Numbered comments keep
their PDB fields, and structured assembly declarations are validated
independently. This does not interpret provider-specific metadata or promise
a comment roundtrip through other forms. See
{ref}`user-foundations-native-world-file-handlers-molsysmt-pdbfilehandler`.

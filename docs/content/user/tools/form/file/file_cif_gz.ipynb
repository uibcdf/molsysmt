## Overview

`file:cif.gz` is a supported representation form in MolSysMT. Form adapter and capabilities for file:cif.gz.

Reading a compressed CIF leaves the source and neighboring files unchanged.
The parser uses an independent temporary directory for each read and cleans it
on success or failure. You can read the same local file concurrently, including
from a directory without write permission. Temporary storage must be available;
parsing still loads the requested molecular data eagerly.

Occupancy is retained as a dimensionless floating-point array with shape `(n_structures, n_atoms)` in native `Structures`. Unknown source values (`?` or `.`) remain `NaN`; an entirely missing primary field remains `None`. Values are never replaced with an assumed occupancy of one.

Alternate atom sites are reduced separately within each model: the largest known occupancy is selected, with label A preferred among tied maxima, then source order. If every occupancy is unknown, the first source variant is retained with unknown occupancy. Original alternatives, coordinates and observations remain available in `alternate_location` for explicit resolution. This import policy differs from the half-occupancy tie convention of {ref}`Solve alternate locations <Tutorial_Solve_atoms_with_alternate_locations>`.

Atom extraction through `msm.convert()` follows native sorted source indices. Direct occupancy queries preserve the requested atom order and repetitions; both routes preserve nonconsecutive and repeated structure indices.

Native conversion to `molsysmt.MolSys` or `molsysmt.Topology` accepts
`get_missing_bonds=False`. This retains mapped `chem_comp_bond` and supported
`struct_conn` declarations while disabling local template, terminal-H,
peptide-link and distance-based inference. It does not discard a mapped
explicit edge when other source atoms lack a reference, and it does not certify
complete connectivity. The existing inference-enabled default remains `True`.
The same policy is forwarded through CIF, BinaryCIF, their gzip forms and
public PDB-identifier acquisition. Topology retains stable inventory only;
inspect chemical edges in the resulting MolSys chemical-state domain.


The shared native reader preserves deposited model labels as string
`structure_id` values, independently of positional structure indices. Labels
and coordinates follow the same selection/reordering, and missing labels
remain unknown. See the
[shared reader contract](../class/others/mmcif_PdbxContainers_DataContainer.ipynb).

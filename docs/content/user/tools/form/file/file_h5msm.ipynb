This is the adapter declaration union, not the field inventory of every file.
Availability depends on the root version, stored domains and represented
values. Mechanical persistence belongs to H5MSM 1.0 and explicit experimental
0.6; 0.5 does not store that domain. Stable inventory belongs to Topology,
chemical assignments/bonds/components to ChemicalStates and coordinates to
Structures. Stored explicit-H assignments are not recounted atom populations.

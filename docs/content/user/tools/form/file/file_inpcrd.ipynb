## Overview

`file:inpcrd` is a supported representation form in MolSysMT. AMBER ASCII restart/inpcrd file format


## Supported attributes

The `file:inpcrd` form container supports a total of **12 attributes**:

| Category | Subcategory / Level | Supported Attributes |
|---|---|---|
| **Topological** | Atom Level | `atom_index` |
| **Topological** | Element Counts | `n_atoms` |
| **Structural** | Coordinates & Conformations | `structure_index`, `structure_id`, `coordinates`, `velocities`, `n_structures` |
| **Structural** | Periodic Box | `box`, `box_shape`, `box_angles`, `box_lengths`, `box_volume` |
| **Structural** | Other Attributes | `atom_index`, `n_atoms` |

Use `msm.get(molsys, element="atom", atom_index=True)` to obtain source atom positions without loading the coordinate array. `selection=[2, 0, 2]` returns `[2, 0, 2]`, and an empty selection returns `[]`. These are positions, not atom IDs or atomic identities. The coordinate file alone does not provide chemical topology.

# Close

*Explicitly closing resource file handlers and open stream handles.*

The function {func}`molsysmt.form.close` safely closes open disk streams and underlying file handles for forms registered under `RESOURCE_FORMS`.

:::{versionadded} 1.0.0
:::

:::{admonition} API documentation
:class: dropdown

Follow this link for a detailed description of the input arguments, raised errors, and returned objects of this function: {func}`molsysmt.form.close`.
:::


## Basic usage

Let's open and close a file handler stream:


```python
import molsysmt as msm
```

```python
h5_handler = msm.convert(msm.systems['Barnase-Barstar']['barnase_barstar.h5msm'], to_form='molsysmt.H5MSMFileHandler')
```

:::{note} Demo Systems Catalog
:class: dropdown

This tutorial uses demonstration datasets provided by MolSysMT. To explore the full catalog of bundled systems, forms, and file paths, visit the {ref}`Demo Systems <user-foundations-entrance-demo-systems>` guide.
:::


```python
print('Handler open:', h5_handler)
msm.form.close(h5_handler)
print('Closed successfully.')
```

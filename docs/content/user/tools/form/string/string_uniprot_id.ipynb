## Overview

A UniProt accession, optionally prefixed with `uniprot_id:`, refers to a remotely retrieved sequence. It contains no direct molecular attributes or coordinates. Convert through FASTA to obtain a native group-only inventory; atomic construction is a separate build operation.

The accession has no direct attribute portfolio. Retrieve and convert its FASTA content before querying a molecular inventory. A FASTA sequence record is not a coordinate structure.

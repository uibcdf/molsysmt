*Molecular connectivity encoded as SMILES text.*

- **Form Type:** `string`
- **Origin / Library:** SMILES notation, with optional RDKit/OpenFF interpretation.

## Overview

SMILES text represents molecular connectivity rather than a coordinate structure. The optional `smiles:` prefix identifies the form; recognition alone is not full chemical validation. RDKit parses chemical conversions and atom-subset extraction. OpenFF targets require their own optional provider.

## Overview

An immutable one-letter sequence represents ordered molecular groups without stored atom or coordinate rows. Native recognition, group queries and string translation use the MolSysMT sequence codec; constructing atomic coordinates is a separate build operation.


Native string recognition, group extraction and one/three-letter translation
are independent of Biopython. Request Bio.Seq or Bio.SeqRecord explicitly when
that optional representation is needed. Selection addresses original groups
before three-letter terminal caps are omitted. Prefix markers never become
sequence letters; empty selections remain valid for object conversion.
See {ref}`user-foundations-support-forms-strings`.

### Native group inventories

These native conversions retain ordered groups and one declared
molecule/entity/chain slot, with no atom, bond or coordinate rows. Zero stored
atoms describes this partial representation, not physical chemical composition.
Use numeric group selections to retain requested order and repetitions. Three-
letter caps are preserved directly; conversion does not pass through one-letter
text. Native ambiguous/rare amino-acid codes retain their group names without
assigning complete chemistry. Gaps, stops, unsupported one-letter symbols and
incomplete three-letter tokens require explicit interpretation.

Native group/chain/molecule/entity queries and text selections operate on the
represented hierarchy. The ordered chain/group relation survives copying,
TopologyDict and H5MSM. Use `msm.extract(..., element="group", selection=[2, 0])`
for an ordered native group subset, or choose another represented hierarchy axis.
Numeric sequence conversion selections still preserve repetitions; extraction
requires unique positions. Generated chain IDs use `A`, `B`, through `Z`, then
`AA`, `AB`, and onward. Coordinate structures cannot be selected, and atomic
construction remains an explicit build operation.

## Overview

An AlphaFold model identifier, optionally prefixed with `alphafold_id:`, refers to remotely retrieved data. The identifier itself contains no atom inventory or coordinates. Recognizing its syntax does not verify that a model exists or that a remote service is reachable.

:::{warning}
The registered direct `file:fasta` route is unavailable and raises
`NotImplementedConversionError` before any download or file modification.
A real model-aware sequence export is deferred under MolSysMT #733; a returned
full UniProt sequence must not be assumed to match represented model groups.
:::

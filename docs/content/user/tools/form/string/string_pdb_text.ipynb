## Overview

PDB text carries source records in memory. Native conversion uses the PDB handler and its explicit naming, primary-observation and connectivity policies. A whole-text result preserves the immutable value; a selected molecular result is parsed and reserialized and need not preserve source bytes.

Native MolSys/Topology conversion accepts `from_naming_scheme` before group classification and native inference. Original names stay unchanged, and context-dependent selections use the interpreted inventory. Text-to-MolSys still defaults to `get_missing_bonds=False`; request True explicitly when choosing an inference engine. The source naming declaration is retained in reader history when returning MolSys. Unsupported contextual OpenMM naming is not translated: explicit requests raise, and the optional policy warns and records failure. A pure Topology retains inventory/declarations, not the chemical graph/history. See {ref}`user-foundations-group-naming-schemes` and {ref}`Tutorial_Infer_Covalent_Bonds`.

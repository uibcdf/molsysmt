# Get polarity

*Querying amino acid polarity indices from reference scales.*

The function {func}`molsysmt.physchem.get_polarity` determines residue polarity scores representing sidechain electrostatic and hydrogen-bonding propensities using two reference scales:
- `'grantham'`: Classic side-chain polarity scale from Grantham (1974).
- `'zimmerman'`: Polarity index from Zimmerman et al. (1968).

:::{versionadded} 1.0.0
:::

:::{admonition} API documentation
:class: dropdown

Follow this link for a detailed description of the input arguments, raised errors, and returned objects of this function: {func}`molsysmt.physchem.get_polarity`.
:::


## Basic usage

By default (`definition='grantham'`), Grantham polarity values are returned for each residue:


```python
import molsysmt as msm
```

```python
molsys = msm.convert(msm.systems['Trp-Cage']['1l2y.h5msm'])
```

:::{note} Demo Systems Catalog
:class: dropdown

This tutorial uses demonstration datasets provided by MolSysMT. To explore the full catalog of bundled systems, forms, and file paths, visit the {ref}`Demo Systems <user-foundations-entrance-demo-systems>` guide.
:::


We query the Grantham polarity values for all residues:


```python
pol_grantham = msm.physchem.get_polarity(molsys, element='group', definition='grantham')
print('Grantham polarity scores:\n', pol_grantham)
```

## Alternative polarity scales

Setting `definition='zimmerman'` evaluates polarity based on electric dipole and dielectric properties:


```python
pol_zimmerman = msm.physchem.get_polarity(molsys, element='group', definition='zimmerman')
print('Zimmerman polarity scores:\n', pol_zimmerman)
```

## Selecting the intended population

This example explicitly uses the optional MDTraj selection syntax. The selected
indices refer to the `group` axis; they are positions, not IDs. Numeric lists
keep their order and repetitions, and an empty supported population keeps a
well-defined empty result. The ordinary MolSysMT syntax remains the default.

Only `element='group'` is supported here. Other levels raise
`ArgumentChoiceError` before selection is interpreted. A group index cannot
silently become an atom index, and no arbitrary aggregate is supplied.

Unknown non-placeholder group names still raise their table diagnostic. These
scales do not validate a group's chemical state or prepare a molecular system.

For a combined example, see {ref}`cookbook-assigning-partial-charges` and the
population notes in {ref}`user-foundations-support-physchem`.


```python
import numpy as np

indices = msm.select(molsys, element='group', selection='resname GLY', syntax='MDTraj')
selected = msm.physchem.get_polarity(molsys, element='group',
    selection='resname GLY', syntax='MDTraj')
control = msm.physchem.get_polarity(molsys, element='group', selection=indices)
np.testing.assert_allclose(selected, control)
assert len(selected) == len(indices)
```

# Get electronegativity

*Querying Pauling electronegativities for constituent atoms.*

The function {func}`molsysmt.physchem.get_electronegativity` retrieves standard Pauling electronegativity values for each selected atom in a molecular system.

:::{versionadded} 1.0.0
:::

:::{admonition} API documentation
:class: dropdown

Follow this link for a detailed description of the input arguments, raised errors, and returned objects of this function: {func}`molsysmt.physchem.get_electronegativity`.
:::


## Basic usage

Let's extract atomic electronegativities from a peptide structure:


```python
import molsysmt as msm
```

```python
molsys = msm.convert(msm.systems['Met-enkephalin']['met_enkephalin.h5msm'])
```

:::{note}
:class: dropdown

**Demo Systems Catalog**

This tutorial uses demonstration datasets provided by MolSysMT. To explore the full catalog of bundled systems, forms, and file paths, visit the {ref}`Demo Systems <user-foundations-entrance-demo-systems>` guide.
:::


We query Pauling electronegativities for all atoms:


```python
en = msm.physchem.get_electronegativity(molsys, element='atom', definition='pauling')
print('First 10 atomic electronegativities (Pauling):', en[:10])
```

## Selecting the intended population

This example explicitly uses the optional MDTraj selection syntax. The selected
indices refer to the `atom` axis; they are positions, not IDs. Numeric lists
keep their order and repetitions, and an empty supported population keeps a
well-defined empty result. The ordinary MolSysMT syntax remains the default.

Only `element='atom'` is supported here. Other levels raise
`ArgumentChoiceError` before selection is interpreted. A group index cannot
silently become an atom index, and no arbitrary aggregate is supplied.

For a combined example, see {ref}`cookbook-assigning-partial-charges` and the
population notes in {ref}`user-foundations-support-physchem`.


```python
import numpy as np

indices = msm.select(molsys, element='atom', selection='name CA', syntax='MDTraj')
selected = msm.physchem.get_electronegativity(molsys, element='atom',
    selection='name CA', syntax='MDTraj')
control = msm.physchem.get_electronegativity(molsys, element='atom', selection=indices)
np.testing.assert_allclose(msm.pyunitwizard.get_value(selected, to_unit='dimensionless'),
    msm.pyunitwizard.get_value(control, to_unit='dimensionless'))
assert len(selected) == len(indices)
```

# Get bulkiness

*Reading unit-aware Zimmerman side-chain reference descriptors.*

{func}`molsysmt.physchem.get_bulkiness` looks up the twenty standard amino-acid
names in the existing reference table. It needs group identities, not coordinates.
This experimental reader does not measure the selected conformation's area.

:::{admonition} API documentation
{func}`molsysmt.physchem.get_bulkiness`
:::

:::{versionadded} 1.0.0
:::


## Reading selected groups

Use any supported form that supplies group names. Here the bundled peptide
provides a topology. Its terminal caps have no reference value, so select
amino-acid groups explicitly. Numeric indices retain their order and repetitions.


```python
import molsysmt as msm

molsys = msm.convert(
    msm.systems['pentalanine']['traj_pentalanine.h5msm'],
    to_form='molsysmt.Topology',
)
area = msm.physchem.get_bulkiness(molsys, selection=[1, 3, 1])
msm.pyunitwizard.get_value(area, to_unit='angstrom**2').round(2).tolist()
```

## Units and empty selections

The output is an area quantity standardized through the active PyUnitWizard
policy. ALA 11.5 Å² is 0.115 nm². Extract explicit units before passing floats
to a model so its feature scale is preserved. Empty selections keep a typed
zero-length group axis. External selection strings use the requested `syntax`.


```python
msm.pyunitwizard.get_value(area, to_unit='nm**2').tolist()
```

```python
empty = msm.physchem.get_bulkiness(molsys, selection=[])
empty.shape
```

## Scope and provenance

Only `element='group'` and `definition='zimmerman'` are implemented. A name-based
lookup does not validate the group's chemical composition. Modified groups,
terminal caps, unknown names and DUM/X placeholders raise an explicit diagnostic;
no parent value or zero area is invented. Group sums are not molecular surfaces.
Coordinates and chemical states remain unchanged.

The numbers match [AAindex1 ZIMJ680102](https://www.genome.jp/aaindex/) and
[official ProtScale](https://web.expasy.org/protscale/pscale/Bulkiness.html).
The original scale is [Zimmerman et al. (1968)](https://doi.org/10.1016/0022-5193(68)90069-6).
The area interpretation follows Table 1 of
[Proust et al. (2019)](https://doi.org/10.3389/fmicb.2019.00906).
The original 1968 full-text derivation and normalization have not been re-audited.
The evidence classification is partial; lookup agreement is not a geometric
validation on arbitrary molecular systems.

If Ackredit is available, a successful call contributes the scale and unit
interpretation references to the application's current session. Empty lookups
also retain attribution; unsupported-group failures do not credit a completed
lookup. Ackredit absence or failure does not change the numerical result.

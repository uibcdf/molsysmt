# Get surface area

*Retrieving tabulated isotropic surface-area descriptors for amino-acid groups.*

The function {func}`molsysmt.physchem.get_surface_area` looks up the historical
Collantes–Dunn isotropic surface-area (ISA) table. The
[original article](https://doi.org/10.1021/jm00014a022), *Journal of Medicinal
Chemistry* **38** (1995), 2705–2713, describes side-chain descriptors derived
from optimized amino-acid structures for QSAR. This lookup does not recompute
those descriptors on your coordinates.

:::{versionadded} 1.0.0
:::

## Basic usage

Retrieve one reference descriptor per selected amino-acid group:


```python
import molsysmt as msm
```

```python
molsys = msm.convert(msm.systems['Met-enkephalin']['met_enkephalin.h5msm'])
```

:::{note} Demo Systems Catalog
:class: dropdown

This tutorial uses demonstration datasets provided by MolSysMT. To explore the full catalog of bundled systems, forms, and file paths, visit the {ref}`Demo Systems <user-foundations-entrance-demo-systems>` guide.

Query the stored Collantes ISA values, retaining their legacy list format:


```python
sa = msm.physchem.get_surface_area(molsys, element='group', definition='collantes')
print('Raw ISA descriptors (Collantes):', sa)
```

## Keeping the group population explicit

These are tabulated amino-acid group descriptors, not geometric calculations
on the current coordinates. Numeric group indices retain order and repetitions,
and empty supported selections keep the same output container. Unknown
non-placeholder groups retain explicit table diagnostics.

Only `element='group'` is supported; other levels are rejected before selection.
String selections honor the requested syntax; this example uses optional MDTraj.

The Collantes table returns raw floats. Its physical unit/normalization is under
review in [#473](https://github.com/uibcdf/molsysmt/issues/473). Do not label these
values as nm squared or add a guessed conversion. Use `get_sasa` for the separate
coordinate-based solvent-accessible-area calculation.


A bare float does not establish that the descriptor is dimensionless. Do not
attach nm², Å² or convert these numbers with a length-unit factor until the
original table unit and normalization have been verified. That scientific
review remains open in [#473](https://github.com/uibcdf/molsysmt/issues/473).
For an actual coordinate-dependent solvent-accessible area, use
{func}`molsysmt.physchem.get_sasa` and its documented quantity contract.

```python
import numpy as np

indices = msm.select(molsys, element='group', selection='resname GLY', syntax='MDTraj')
selected = msm.physchem.get_surface_area(molsys, element='group',
    selection='resname GLY', syntax='MDTraj')
control = msm.physchem.get_surface_area(molsys, element='group', selection=indices)
np.testing.assert_allclose(selected, control)
assert len(selected) == len(indices)
```

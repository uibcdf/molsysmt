# Get atomic radius

*Extracting atomic van der Waals and effective radii for molecular structures.*

The function {func}`molsysmt.physchem.get_atomic_radius` retrieves atomic radii using standard element-based van der Waals definitions (`definition='vdw'`) or implicit-hydrogen-aware ProtOr parameter sets (`definition='protor'`).

:::{versionadded} 1.0.0
:::

:::{admonition} API documentation
:class: dropdown

Follow this link for a detailed description of the input arguments, raised errors, and returned objects of this function: {func}`molsysmt.physchem.get_atomic_radius`.
:::


## Basic usage: Element-based van der Waals radii

By default (`definition='vdw'`), MolSysMT extracts standard element-level van der Waals radii tabulated by Bondi and Mantina et al.:


```python
import molsysmt as msm
import pyunitwizard as puw
```

```python
molsys = msm.convert(msm.systems['Trp-Cage']['1l2y.h5msm'])
```

:::{note} Demo Systems Catalog
:class: dropdown

This tutorial uses demonstration datasets provided by MolSysMT. To explore the full catalog of bundled systems, forms, and file paths, visit the {ref}`Demo Systems <user-foundations-entrance-demo-systems>` guide.
:::


Let's extract standard element-based van der Waals radii for the first residue (`group_index==0`):


```python
radius_vdw = msm.physchem.get_atomic_radius(molsys, element='atom', selection='group_index==0', definition='vdw')
print('Element-based VdW radii (nm):\n', radius_vdw)
```

## ProtOr implicit-hydrogen-aware van der Waals radii

When dealing with united-atom representations or implicit-hydrogen models, the `definition='protor'` parameter adjusts heavy atom radii depending on the chemical hybridization and number of bonded hydrogens (Tsai et al.):


```python
radius_protor = msm.physchem.get_atomic_radius(molsys, element='atom', selection='group_index==0', definition='protor')
print('ProtOr VdW radii (nm):\n', radius_protor)
```

## Direct ProtOr type and radius helpers

MolSysMT also provides dedicated helper functions to determine the exact ProtOr atom classification codes (e.g. `N3H1`, `C4H1`, `C3H0`, `O1H0`) and their corresponding radii directly:


```python
protor_types, provenance = msm.physchem.get_protor_atom_type(molsys, selection='group_index==0')
print('ProtOr atom types:', protor_types)

direct_radii = msm.physchem.get_protor_vdw_radius(molsys, selection='group_index==0')
print('Direct ProtOr radii:', direct_radii)
```

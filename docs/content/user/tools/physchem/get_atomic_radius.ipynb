# Get atomic radius

*Extracting atomic van der Waals and effective radii for molecular structures.*

The function {func}`molsysmt.physchem.get_atomic_radius` retrieves atomic radii using standard element-based van der Waals definitions or implicit-hydrogen-aware ProtOr parameter sets.

:::{versionadded} 1.0.0
:::

:::{admonition} API documentation
:class: dropdown

Follow this link for a detailed description of the input arguments, raised errors, and returned objects of this function: {func}`molsysmt.physchem.get_atomic_radius`.
:::


## Basic usage

Let's extract standard element-based van der Waals radii from a protein structure:


```python
import molsysmt as msm
import pyunitwizard as puw
```

```python
molsys = msm.convert(msm.systems['Trp-Cage']['1l2y.h5msm'])
```

:::{note} Demo Systems Catalog
:class: dropdown

This tutorial uses demonstration datasets provided by MolSysMT. To explore the full catalog of bundled systems, forms, and file paths, visit the {ref}`Demo Systems <user-foundations-entrance-demo-systems>` guide.
:::


We query standard element-based van der Waals radii for the first residue:


```python
radius_vdw = msm.physchem.get_atomic_radius(molsys, element='atom', selection='group_id==0', definition='vdw')
print('Element-based VdW radii (nm):\n', radius_vdw)
```

## ProtOr implicit-hydrogen-aware van der Waals radii

When dealing with united-atom or implicit-hydrogen representations, the `definition='protor'` parameter adjusts heavy atom radii based on chemical context:


```python
radius_protor = msm.physchem.get_atomic_radius(molsys, element='atom', selection='group_id==0', definition='protor')
print('ProtOr VdW radii (nm):\n', radius_protor)
```

## Direct ProtOr type and radius helpers

MolSysMT provides dedicated helper functions to determine ProtOr atom classification codes and their associated radii directly:


```python
protor_types, provenance = msm.physchem.get_protor_atom_type(molsys, selection='group_id==0')
print('ProtOr atom types:', protor_types)
direct_radii = msm.physchem.get_protor_vdw_radius(molsys, selection='group_id==0')
print('Direct ProtOr radii:', direct_radii)
```

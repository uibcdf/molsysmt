# Get volume

*Querying amino acid sidechain steric volumes from reference scales.*

The function {func}`molsysmt.physchem.get_volume` evaluates residue sidechain volumes using established biochemical scales (e.g. Grantham).

:::{versionadded} 1.0.0
:::

:::{admonition} API documentation
:class: dropdown

Follow this link for a detailed description of the input arguments, raised errors, and returned objects of this function: {func}`molsysmt.physchem.get_volume`.
:::


## Basic usage

Let's extract Grantham sidechain volumes for residues in a molecular system:


```python
import molsysmt as msm
```

```python
molsys = msm.convert(msm.systems['Trp-Cage']['1l2y.h5msm'])
```

:::{note} Demo Systems Catalog
:class: dropdown

This tutorial uses demonstration datasets provided by MolSysMT. To explore the full catalog of bundled systems, forms, and file paths, visit the {ref}`Demo Systems <user-foundations-entrance-demo-systems>` guide.
:::


We query sidechain volumes using the Grantham scale:


```python
volumes = msm.physchem.get_volume(molsys, definition='grantham')
print('Sidechain volumes (Grantham):\n', volumes)
```

## Keeping the group population explicit

These are tabulated amino-acid group descriptors, not geometric calculations
on the current coordinates. Numeric group indices retain order and repetitions,
and empty supported selections keep the same output container. Unknown
non-placeholder groups retain explicit table diagnostics.

The public signature has no element argument: it queries groups. String
selections honor its syntax argument, as the optional MDTraj example shows.


```python
import numpy as np

indices = msm.select(molsys, element='group', selection='resname GLY', syntax='MDTraj')
selected = msm.physchem.get_volume(molsys,
    selection='resname GLY', syntax='MDTraj')
control = msm.physchem.get_volume(molsys, selection=indices)
np.testing.assert_allclose(selected, control)
assert len(selected) == len(indices)
```

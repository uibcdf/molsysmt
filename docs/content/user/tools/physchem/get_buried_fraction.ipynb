# Get buried fraction

*Determining statistical buried fractions of residues in folded proteins.*

The function {func}`molsysmt.physchem.get_buried_fraction` retrieves empirical molar percentage fractions of residues found in buried environments (Janin scale).

:::{versionadded} 1.0.0
:::

:::{admonition} API documentation
:class: dropdown

Follow this link for a detailed description of the input arguments, raised errors, and returned objects of this function: {func}`molsysmt.physchem.get_buried_fraction`.
:::


## Basic usage

Let's extract statistical buried fractions for residues in a protein system:


```python
import molsysmt as msm
```

```python
molsys = msm.convert(msm.systems['Trp-Cage']['1l2y.h5msm'])
```

:::{note}
:class: dropdown

**Demo Systems Catalog**

This tutorial uses demonstration datasets provided by MolSysMT. To explore the full catalog of bundled systems, forms, and file paths, visit the {ref}`Demo Systems <user-foundations-entrance-demo-systems>` guide.
:::


We query the Janin buried fraction percentages:


```python
bf = msm.physchem.get_buried_fraction(molsys, element='group', definition='janin')
print('Buried fractions (Janin, %):\n', bf)
```

## Keeping the group population explicit

These are tabulated amino-acid group descriptors, not geometric calculations
on the current coordinates. Numeric group indices retain order and repetitions,
and empty supported selections keep the same output container. Unknown
non-placeholder groups retain explicit table diagnostics.

Only `element='group'` is supported; other levels are rejected before selection.
String selections honor the requested syntax; this example uses optional MDTraj.


```python
import numpy as np

indices = msm.select(molsys, element='group', selection='resname GLY', syntax='MDTraj')
selected = msm.physchem.get_buried_fraction(molsys, element='group',
    selection='resname GLY', syntax='MDTraj')
control = msm.physchem.get_buried_fraction(molsys, element='group', selection=indices)
np.testing.assert_allclose(selected, control)
assert len(selected) == len(indices)
```

## The Essentials

The Root Mean Square Deviation (RMSD) measures the average distance between the atoms of two structures. You can compare different systems or frames of the same trajectory.

````python
import molsysmt as msm
from molsysmt import systems

trajectory = systems['alanine dipeptide']['alanine_dipeptide.h5msm']

# Calculate RMSD of the whole trajectory against the first frame
rmsd = msm.structure.get_rmsd(trajectory, selection='backbone', 
                              reference_molecular_system=trajectory, reference_structure_index=0)

print(f"RMSD values (first 5 frames): {rmsd[:5]}")
````

## Subject vs Reference

MolSysMT uses a clear naming convention for the comparison:
- `molecular_system`: The system you are analyzing (the "subject").
- `reference_molecular_system`: The structure used as the 0-deviation baseline.

```{warning}
Both the subject selection and the reference selection must have the same number of atoms. If they don't, MolSysMT will raise an `ArgumentError` through SMonitor.
```

## Heavy Mode & Performance

For massive trajectories, `get_rmsd()` can operate in **Heavy Mode**, processing chunks of the trajectory to avoid memory overflow.

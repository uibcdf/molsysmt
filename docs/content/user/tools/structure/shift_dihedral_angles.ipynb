# Shift dihedral angles

*Shifting dihedral angles by incremental angular amounts.*

The function {func}`molsysmt.structure.shift_dihedral_angles` rotates dihedral angles incrementally relative to their existing values, applying an angular displacement $\Delta \theta$ while preserving connected covalent blocks.

:::{versionadded} 1.0.0
:::

:::{admonition} API documentation
:class: dropdown

Follow this link for a detailed description of the input arguments, raised errors, and returned objects of this function: {func}`molsysmt.structure.shift_dihedral_angles`.
:::


## Basic usage

Let's show how to shift a specific backbone $\phi$ dihedral angle by $+90.0^\circ$ in Met-enkephalin:


```python
import molsysmt as msm
import pyunitwizard as puw
```

```python
molsys = msm.convert(msm.systems['Met-enkephalin']['met_enkephalin.h5msm'])
```

:::{note}
:class: dropdown

**Demo Systems Catalog**

This tutorial uses demonstration datasets provided by MolSysMT. To explore the full catalog of bundled systems, forms, and file paths, visit the {ref}`Demo Systems <user-foundations-entrance-demo-systems>` guide.
:::


We query the backbone $\phi$ quartets and measure the current value of the second $\phi$ angle:


```python
phi_quartets = msm.chemical_states.get_dihedral_quartets(molsys, phi=True)
initial_angle = msm.structure.get_dihedral_angles(molsys, dihedral_quartets=phi_quartets[1])
print('Initial phi angle:', puw.convert(initial_angle, to_unit='degrees'))
```

We inspect the initial 3D conformation of the peptide:


```python
molsysviewer_htmlfile = '_static/views/tools_structure_shift_dihedral_angles_1.html'
```

```python
msm.view(molsys)
```

We shift the dihedral angle by $+90.0^\circ$ using {func}`molsysmt.structure.shift_dihedral_angles`:


```python
molsys_shifted = msm.structure.shift_dihedral_angles(molsys, dihedral_quartets=phi_quartets[1], shifts='90.0 degrees')
```

We verify that the new angle reflects exactly the $+90.0^\circ$ angular increment:


```python
new_angle = msm.structure.get_dihedral_angles(molsys_shifted, dihedral_quartets=phi_quartets[1])
print('Shifted phi angle:', puw.convert(new_angle, to_unit='degrees'))
```

Visualizing the shifted system in 3D confirms the updated peptide backbone geometry:


```python
molsysviewer_htmlfile = '_static/views/tools_structure_shift_dihedral_angles_2.html'
```

```python
msm.view(molsys_shifted)
```

## Shifting multiple dihedral angles simultaneously

Multiple dihedral angles can be shifted concurrently by providing a list or array of angular shifts:


```python
shifts = puw.quantity([30.0, 30.0, 30.0, 30.0], 'degrees')
molsys_multi_shift = msm.structure.shift_dihedral_angles(molsys, dihedral_quartets=phi_quartets, shifts=shifts)
all_shifted = msm.structure.get_dihedral_angles(molsys_multi_shift, dihedral_quartets=phi_quartets)
print('All phi angles after multi-shift:', puw.convert(all_shifted, to_unit='degrees'))
```

## Explicit blocks and finite edits

`blocks` is a pool of nonempty atom-index lists. For each quartet, all blocks
containing its fourth atom contribute to the moving population. Every fourth
atom must belong to a block. This pool is shared across quartets; it is not an
array assigning one block to each quartet. `blocks=None` computes the covalent
blocks after cutting the central bond. A custom restricted population changes
which atoms move and can change distances across its boundary.

Both editors reject NaN, infinity, complex values, empty requests and incompatible
angular shapes before publishing coordinates. Size-one angular axes broadcast;
a vector specifies values per quartet and a column matrix specifies values per
structure. Rejection leaves coordinates, box, topology and interaction coverage
unchanged. A successful edit makes interactions on edited structures unevaluated,
including previously evaluated empty structures; no detector runs automatically.
The explicit-block route also accepts coordinate-only forms.


```python
import numpy as np

quartet = np.asarray(phi_quartets[1], dtype=int)
moving_block = sorted(next(
    block for block in msm.chemical_states.get_covalent_blocks(
        molsys, remove_bonds=quartet[1:3]
    ) if quartet[3] in block
))
before_edit = puw.get_value(msm.get(molsys, coordinates=True), to_unit='nm').copy()
molsys_explicit = msm.structure.shift_dihedral_angles(
    molsys, dihedral_quartets=quartet, shifts='15 degrees',
    blocks=[moving_block], structure_indices=[0], pbc=False
)
try:
    msm.structure.shift_dihedral_angles(
        molsys, dihedral_quartets=quartet, shifts=puw.quantity(np.nan, 'radians'),
        blocks=[moving_block], structure_indices=[0], pbc=False, in_place=True
    )
except msm.ArgumentError:
    print('Nonfinite edit rejected before coordinate publication.')
else:
    raise AssertionError('The edit should reject a nonfinite request.')
np.testing.assert_array_equal(
    puw.get_value(msm.get(molsys, coordinates=True), to_unit='nm'), before_edit
)
```

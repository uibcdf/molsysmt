# Align principal axes

*Aligning the principal inertia or geometric axes of a molecular system over reference coordinate axes.*

The function {func}`molsysmt.structure.align_principal_axes` rotates a molecular system so that its principal axes of inertia (or geometric axes) align with a specified target orthonormal coordinate basis.

:::{versionadded} 1.0.0
:::

:::{admonition} API documentation
:class: dropdown

Follow this link for a detailed description of the input arguments, raised errors, and returned objects of this function: {func}`molsysmt.structure.align_principal_axes`.
:::


## Basic usage

Let's demonstrate principal axes alignment using a POPC lipid molecule:


```python
import molsysmt as msm
import numpy as np
```

```python
crd = msm.systems['POPC']['popc.crd']
psf = msm.systems['POPC']['popc.psf']
molsys = msm.convert([crd, psf], to_form='molsysmt.MolSys')
```

:::{note} Demo Systems Catalog
:class: dropdown

This tutorial uses demonstration datasets provided by MolSysMT. To explore the full catalog of bundled systems, forms, and file paths, visit the {ref}`Demo Systems <user-foundations-entrance-demo-systems>` guide.
:::


We inspect the principal axes of inertia and their principal moments before alignment using {func}`molsysmt.structure.get_principal_axes`:


```python
axes, momenta = msm.structure.get_principal_axes(molsys)
```

```python
for ii in range(3):
    print(f'Axis {ii}: {axes[0, ii]} | Moment: {momenta[0, ii]:.2f}')
```

Now we align the principal axes to the canonical Cartesian basis $[[1,0,0], [0,1,0], [0,0,1]]$:


```python
molsys_aligned = msm.structure.align_principal_axes(molsys, axes=[[1,0,0], [0,1,0], [0,0,1]])
```

We verify that the principal axes of the aligned system match the canonical Cartesian directions:


```python
axes_aligned, momenta_aligned = msm.structure.get_principal_axes(molsys_aligned)
```

```python
for ii in range(3):
    print(f'Aligned Axis {ii}: {axes_aligned[0, ii]}')
```

# Add missing bonds

*Inferring and adding missing covalent bonds based on interatomic distances and chemical element types.*

The function {func}`molsysmt.build.add_missing_bonds` infers and adds missing covalent bonds in a molecular system by analyzing interatomic distances and standard valence rules for known elements. This function is particularly useful when working with coordinate files (`.pdb`, `.xyz`, or trajectory files) that lack explicit topology or bond definitions.

:::{versionadded} 1.0.0
:::

:::{admonition} API documentation
:class: dropdown

Follow this link for a detailed description of the input arguments, raised errors, and returned objects of this function: {func}`molsysmt.build.add_missing_bonds`.
:::


## Basic usage

Let's show how this function works using an alanine dipeptide system:


```python
import molsysmt as msm
```

```python
molsys = msm.systems['alanine dipeptide']['alanine_dipeptide.h5msm']
molsys = msm.convert(molsys)
```

:::{note} Demo Systems Catalog
:class: dropdown

This tutorial uses demonstration datasets provided by MolSysMT. To explore the full catalog of bundled systems, forms, and file paths, visit the {ref}`Demo Systems <user-foundations-entrance-demo-systems>` guide.
:::


To demonstrate {func}`molsysmt.build.add_missing_bonds`, we first save the original list of `bonded_atom_pairs` and remove all existing covalent bonds from the molecular system to simulate a topology lacking bond definitions:


```python
bonded_atom_pairs = msm.get(molsys, bonded_atom_pairs=True)
```

```python
molsys.topology.remove_bonds()
```

Let's verify that the system currently has zero covalent bonds:


```python
msm.get(molsys, n_bonds=True)
```

We now call {func}`molsysmt.build.add_missing_bonds` to automatically infer covalent bonds based on interatomic distances and element types:


```python
msm.build.add_missing_bonds(molsys)
```

Finally, let's verify that the total number of bonds was restored and check if the newly inferred bonded atom pairs match the original topology:


```python
msm.get(molsys, n_bonds=True)
```

```python
new_bonded_atom_pairs = msm.get(molsys, bonded_atom_pairs=True)
```

```python
new_bonded_atom_pairs == bonded_atom_pairs
```

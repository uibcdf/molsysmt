# Has hydrogens

*Checking whether a molecular system or selection contains hydrogen atoms.*

The function {func}`molsysmt.build.has_hydrogens` evaluates whether hydrogen atoms are present in a molecular system or within a specific selection, returning `True` or `False`.

:::{versionadded} 1.0.0
:::

:::{admonition} API documentation
:class: dropdown

Follow this link for a detailed description of the input arguments, raised errors, and returned objects of this function: {func}`molsysmt.build.has_hydrogens`.
:::


## Basic usage

Let's show how this function works using T4 lysozyme (PDB ID: 181L):


```python
import molsysmt as msm
```

```python
molsys = msm.convert(msm.systems['T4 lysozyme L99A']['181l.h5msm'], to_form='molsysmt.MolSys')
```

:::{note} Demo Systems Catalog
:class: dropdown

This tutorial uses demonstration datasets provided by MolSysMT. To explore the full catalog of bundled systems, forms, and file paths, visit the {ref}`Demo Systems <user-foundations-entrance-demo-systems>` guide.
:::


Let's check the number of hydrogen atoms using {func}`molsysmt.basic.get`:


```python
msm.get(molsys, selection='atom_type=="H"', n_atoms=True)
```

We check if the system contains hydrogen atoms using {func}`molsysmt.build.has_hydrogens`:


```python
msm.build.has_hydrogens(molsys)
```

We can also restrict the check to a specific selection (such as protein molecules):


```python
msm.build.has_hydrogens(molsys, selection='molecule_type=="protein"')
```

Now let's complete terminal cappings, heavy atoms, and add hydrogen atoms to the system:


```python
molsys = msm.build.add_missing_terminal_cappings(molsys)
molsys = msm.build.add_missing_heavy_atoms(molsys)
molsys = msm.build.add_missing_hydrogens(molsys)
```

We check again to confirm that hydrogen atoms are now present:


```python
msm.build.has_hydrogens(molsys)
```

```python
msm.build.has_hydrogens(molsys, selection='molecule_type=="protein"')
```

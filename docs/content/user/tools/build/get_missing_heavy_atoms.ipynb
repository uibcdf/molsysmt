# Get missing heavy atoms

*Identifying missing non-hydrogen heavy atoms in a molecular system.*

Structures solved by experimental methods such as X-ray crystallography often contain missing heavy atoms due to resolution limits or local disorder. The function {func}`molsysmt.build.get_missing_heavy_atoms` analyzes residue topologies against standard templates to identify non-hydrogen heavy atoms that are missing.

:::{versionadded} 1.0.0
:::

:::{admonition} API documentation
:class: dropdown

Follow this link for a detailed description of the input arguments, raised errors, and returned objects of this function: {func}`molsysmt.build.get_missing_heavy_atoms`.
:::


## Basic usage

Let's show how this function works using a protein structure loaded from PDB ID `1BRS`:


```python
import molsysmt as msm
```

```python
molsys = msm.convert('1BRS', selection='molecule_type=="protein"')
```

```python
msm.info(molsys, element='molecule')
```

We now use {func}`molsysmt.build.get_missing_heavy_atoms` to detect missing non-hydrogen atoms across all residues:


```python
missing_heavy_atoms = msm.build.get_missing_heavy_atoms(molsys)
```

```python
missing_heavy_atoms
```

We can identify which molecules contain these missing heavy atoms:


```python
residue_indices = list(missing_heavy_atoms.keys())
molecule_indices = msm.get(molsys, element='molecule', selection='group_index in @residue_indices', molecule_index=True)
```

```python
msm.info(molsys, element='molecule', selection='molecule_index in @molecule_indices')
```

Finally, let's check if the system has hydrogen atoms using {func}`molsysmt.build.has_hydrogens`:


```python
msm.build.has_hydrogens(molsys)
```

# Get non standard residues

*Identifying non-standard residues in a molecular system.*

Many biomolecular structures contain non-standard residues such as modified amino acids or post-translational modifications. The function {func}`molsysmt.build.get_non_standard_residues` identifies residues that differ from standard amino acid or nucleotide dictionaries.

:::{versionadded} 1.0.0
:::

:::{admonition} API documentation
:class: dropdown

Follow this link for a detailed description of the input arguments, raised errors, and returned objects of this function: {func}`molsysmt.build.get_non_standard_residues`.
:::


## Basic usage

Let's show how this function works using cAMP-dependent protein kinase, PDB entry `1ATP`, which MolSysMT ships as a demo system. It carries two post-translational modifications: a phosphothreonine in the activation loop and a phosphoserine in the turn motif.

```python
import molsysmt as msm
```

```python
from molsysmt import systems

molsys = msm.convert(systems['1ATP']['1atp.pdb'])
```

We now call {func}`molsysmt.build.get_non_standard_residues` to detect any non-standard residues.

The result maps each non-standard residue index to its nearest **standard sequence equivalent**, not to a chemical repair template. Here `TPO` (phosphothreonine, residue id 197) maps to `THR`, and `SEP` (phosphoserine, residue id 338) maps to `SER`. Their heavy-atom inventories are assessed by exact component templates in {func}`molsysmt.build.get_missing_heavy_atoms`; the standard sequence equivalent does not determine their repair chemistry.

The native backend now reads each group's stored naming convention in native
objects, native dictionaries and H5MSM 0.5. Without a declaration it uses only
the assessed unambiguous subset. Ambiguous labels require context; unassessed
default parents are omitted. An explicit but unsupported amino-acid interpretation
raises an error. The PDBFixer engine uses its own catalog and can disagree. See
{ref}`user-foundations-group-naming-schemes` and
{ref}`Tutorial_Get_standard_name`. Reading this report changes no names,
chemical assignments, atoms or coordinates.


```python
msm.build.get_non_standard_residues(molsys)
```

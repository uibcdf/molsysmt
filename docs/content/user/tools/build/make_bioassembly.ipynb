# Make bioassembly

*Constructing biological assemblies using symmetry transformations from PDB records.*

Biological assemblies (or biomolecules) represent the functional quaternary structure of a macromolecule, often generated by applying crystallographic symmetry operations to the asymmetric unit. The function {func}`molsysmt.build.make_bioassembly` constructs a full biological assembly from PDB or mmCIF records.

:::{versionadded} 1.0.0
:::

:::{admonition} API documentation
:class: dropdown

Follow this link for a detailed description of the input arguments, raised errors, and returned objects of this function: {func}`molsysmt.build.make_bioassembly`.
:::


## Basic usage

Let's show how to construct a biological assembly from PDB entry `1OUT`:


```python
import molsysmt as msm
```

```python
molsys = msm.convert('1OUT')
```

```python
molsysviewer_htmlfile = '_static/views/tools_build_make_bioassembly_1.html'
```

```python
msm.view(molsys)
```

Now we apply symmetry operations to construct biological assembly `'1'`:


```python
molsys = msm.build.make_bioassembly(molsys, bioassembly='1')
```

```python
molsysviewer_htmlfile = '_static/views/tools_build_make_bioassembly_2.html'
```

```python
msm.view(molsys)
```

Let's look at a second example using PDB entry `2BUK`:


```python
molsys = msm.convert('2BUK')
```

```python
molsysviewer_htmlfile = '_static/views/tools_build_make_bioassembly_3.html'
```

```python
msm.view(molsys)
```

We construct biological assembly `'1'` for `2BUK` and inspect its summary:


```python
assembled_molsys = msm.build.make_bioassembly(molsys, bioassembly='1')
```

```python
msm.info(assembled_molsys)
```

```python
molsysviewer_htmlfile = '_static/views/tools_build_make_bioassembly_4.html'
```

```python
msm.view(assembled_molsys)
```

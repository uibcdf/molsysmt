# Get missing bonds

*Identifying missing covalent bonds in a molecular system.*

Some molecular systems may lack explicit bond definitions in their topology. The function {func}`molsysmt.build.get_missing_bonds` returns candidate missing atom pairs from group templates and distance criteria without modifying the source. These heuristics do not certify a complete chemical graph.

Peptide candidates require consecutive **source group indices** in the same defined chain and a compatible C–N distance. A selection cannot make separated groups adjacent, and group IDs need not be consecutive. Integer selections are **atom indices**; both endpoints of every returned pair must belong to the selection. Distances use the requested structure, length units and PBC convention. Groups with missing or ambiguous chain membership do not contribute peptide candidates. File segment breaks cannot constrain candidates unless the source adapter preserves them.

:::{versionadded} 1.0.0
:::

With the optional PyTraj engine, the operation owns its intermediate PDB until
topology loading and bond-pair extraction finish. Success and write/read/extraction
failures retire that managed scratch; returned pairs remain usable. Cleanup
errors are visible. This does not change the engine's candidate criteria.


## Basic usage

Let's show how this function works by building a peptide and intentionally removing a covalent bond:


```python
import molsysmt as msm
```

```python
molsys = msm.build.build_peptide('AceAlaNme')
```

Let's inspect the first covalent bonds in the system using {func}`molsysmt.basic.get`:


```python
msm.get(molsys, element='bond', selection=[0, 1, 2], bonded_atoms=True)
```

Now we remove the first bond (index `0`) from the system topology:


```python
molsys.topology.remove_bonds([0])
```

We now run {func}`molsysmt.build.get_missing_bonds` to detect the missing covalent bond:


```python
msm.build.get_missing_bonds(molsys)
```

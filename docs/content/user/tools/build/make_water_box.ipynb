# Make water box

*Generating a box of water molecules with specified dimensions.*

The function {func}`molsysmt.build.make_water_box` fills a periodic box of specified dimensions with water molecules at standard density.

:::{versionadded} 1.0.0
:::

:::{admonition} API documentation
:class: dropdown

Follow this link for a detailed description of the input arguments, raised errors, and returned objects of this function: {func}`molsysmt.build.make_water_box`.
:::


## Basic usage

Let's show how to create a rectangular water box using `numpy` box dimensions:


```python
import molsysmt as msm
import numpy as np
```

We define a cubic box of length 2.0 nanometers:


```python
box = np.array([[[2.0, 0.0, 0.0], [0.0, 2.0, 0.0], [0.0, 0.0, 2.0]]]) * msm.pyunitwizard.unit('nm')
```

Now we pass the box dimensions to {func}`molsysmt.build.make_water_box` to generate the solvated box:


```python
molsys = msm.build.make_water_box(box)
```

```python
msm.info(molsys)
```

We can verify that the oxygen atom coordinates fall strictly within the box boundaries:


```python
coordinates = msm.get(molsys, selection='atom_type=="O"', coordinates=True)
```

```python
print(f'Min X: {coordinates[0,:,0].min()} and Max X: {coordinates[0,:,0].max()}')
```

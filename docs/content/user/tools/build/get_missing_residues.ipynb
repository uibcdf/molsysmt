# Get missing residues

*Identifying missing amino acid or nucleotide residues in a molecular system.*

Experimental structures often omit flexible loops or terminal residues due to unresolved electron density. The function {func}`molsysmt.build.get_missing_residues` compares resolved residues against reference sequence records (such as `_entity_poly_seq` in mmCIF or `SEQRES` in PDB files) to identify missing residues.

:::{versionadded} 1.0.0
:::

:::{admonition} API documentation
:class: dropdown

Follow this link for a detailed description of the input arguments, raised errors, and returned objects of this function: {func}`molsysmt.build.get_missing_residues`.
:::


## Basic usage

Let's show how this function works using a protein structure loaded from PDB entry `1BRS`:


```python
import molsysmt as msm
```

First, we convert PDB entry `1BRS` into a standard `molsysmt.MolSys` object:


```python
molsys = msm.convert('1BRS', selection='molecule_type=="protein"', to_form='molsysmt.MolSys')
```

Let's inspect the protein molecules in the system:


```python
msm.info(molsys, element='molecule')
```

Now we call {func}`molsysmt.build.get_missing_residues` using the PDB identifier `'1BRS'` to compare the resolved residues against the reference sequence records:


```python
missing_residues = msm.build.get_missing_residues('1BRS')
```

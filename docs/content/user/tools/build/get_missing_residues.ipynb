# Get missing residues

*Identifying missing amino acid or nucleotide residues in a molecular system.*

Experimental structures often omit flexible loops or terminal residues due to unresolved electron density. The function {func}`molsysmt.build.get_missing_residues` compares the resolved residues against reference sequence records (such as `_entity_poly_seq` in mmCIF files) to identify missing residues.

:::{versionadded} 1.0.0
:::

:::{admonition} API documentation
:class: dropdown

Follow this link for a detailed description of the input arguments, raised errors, and returned objects of this function: {func}`molsysmt.build.get_missing_residues`.
:::


## Basic usage

Let's show how this function works using a protein structure loaded from PDB entry `1BRS`:


```python
import molsysmt as msm
```

```python
molsys = msm.convert('1BRS', to_form='mmcif_PdbxContainers_DataContainer')
```

Now we run {func}`molsysmt.build.get_missing_residues` to find missing residues compared to the reference sequence:


```python
missing_residues = msm.build.get_missing_residues(molsys)
```

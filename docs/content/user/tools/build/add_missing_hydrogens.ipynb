# Add missing hydrogens

*Adding missing hydrogen atoms to a molecular system.*

Many molecular systems lack hydrogen atoms or require protonation adjustments to reflect a specific chemical or physical environment. The function {func}`molsysmt.build.add_missing_hydrogens` adds missing hydrogen atoms to any molecular system based on standard protonation rules and residue templates at a specified pH.

:::{versionadded} 1.0.0
:::

:::{admonition} API documentation
:class: dropdown

Follow this link for a detailed description of the input arguments, raised errors, and returned objects of this function: {func}`molsysmt.build.add_missing_hydrogens`.
:::


## Basic usage

Let's show how this function works using a protein structure loaded from T4 lysozyme (PDB ID: 181L):


```python
import molsysmt as msm
```

```python
molsys = msm.convert('181L', selection='molecule_type=="protein"')
```

```python
msm.info(molsys)
```

This protein system was obtained from a PDB X-ray structure and does not contain hydrogen atoms. Let's verify this using {func}`molsysmt.build.has_hydrogens`:


```python
msm.build.has_hydrogens(molsys)
```

Now we add the missing hydrogen atoms using standard protonation patterns at pH 7.4:


```python
molsys = msm.build.add_missing_hydrogens(molsys, pH=7.4)
```

Let's confirm the addition of hydrogen atoms by counting how many atoms of type `'H'` are now present:


```python
msm.get(molsys, element='atom', selection='atom_type=="H"', n_atoms=True)
```

## Preserving naming context

Native pH preparation reads stored group conventions. Ambiguous undeclared labels
fail before native component reconstruction. The eight explicit Amber states use
checked heavy connectivity and H counts per parent atom, preserving their named
side-chain inventories across pH choices. Heavy-only modified references, named-state
terminal H completion and incompatible layouts remain unassessed with a diagnostic.
The standard pH model and placement geometry retain their existing approximations.

OpenMM/PDBFixer wrappers refuse reviewed contextual labels and incompatible curated
modified-name declarations until source-naming adapters exist. Fixed-state RDKit
preparation follows ChemicalStates without chemically interpreting group names.

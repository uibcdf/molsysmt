# Editable

*Creating an editable molecular system builder to modify or construct topologies.*

The function {func}`molsysmt.build.editable` converts any molecular system (or creates a new empty instance) into a native {class}`molsysmt.MolSysBuilder` object. A builder enables dynamic programmatic modifications (such as adding or removing atoms, residues, chains, and bonds) before materializing a final native {class}`molsysmt.MolSys` instance via `.build()`.

:::{versionadded} 1.0.0
:::

:::{admonition} API documentation
:class: dropdown

Follow this link for a detailed description of the input arguments, raised errors, and returned objects of this function: {func}`molsysmt.build.editable`.
:::


## Basic usage

Let's show how to convert an existing molecular system into an editable builder using Met-enkephalin:


```python
import molsysmt as msm
```

```python
molsys = msm.convert(msm.systems['Met-enkephalin']['met_enkephalin.h5msm'])
```

:::{note} Demo Systems Catalog
:class: dropdown

This tutorial uses demonstration datasets provided by MolSysMT. To explore the full catalog of bundled systems, forms, and file paths, visit the {ref}`Demo Systems <user-foundations-entrance-demo-systems>` guide.
:::


We convert the molecular system into a {class}`molsysmt.MolSysBuilder` object using {func}`molsysmt.build.editable`:


```python
builder = msm.build.editable(molsys)
```

```python
type(builder)
```

## Modifying topology and bonds

Using the builder instance, we can add custom bonds or reassign residue groups to new chains:


```python
builder.add_bond(0, 10, bond_order=1)
```

```python
builder.assign_groups_to_new_chain(group_indices=[0, 1], chain_name='B')
```

## Materializing the final molecular system

Once edits are complete, we materialize a new native {class}`molsysmt.MolSys` object by calling `.build()`:


```python
new_molsys = builder.build()
```

```python
msm.info(new_molsys, element='chain')
```

## Creating a builder from scratch

Calling {func}`molsysmt.build.editable` without arguments returns an empty builder:


```python
empty_builder = msm.build.editable()
```

```python
type(empty_builder)
```

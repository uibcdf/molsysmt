# Editable

*Creating an editable molecular system builder to modify or construct topologies.*

The function {func}`molsysmt.build.editable` serves as the primary entry point to work with the native {class}`molsysmt.MolSysBuilder` container (detailed in the {ref}`MolSysBuilder <user-foundations-native-world-classes-molsysmt-molsysbuilder>` guide). While standard molecular system containers in MolSysMT are immutable for performance and safety, a builder provides a mutable staging environment to dynamically add, edit, or remove atoms, bonds, groups, and chains before materializing a final native {class}`molsysmt.MolSys` instance via `.build()`.

:::{versionadded} 1.0.0
:::

:::{admonition} API documentation
:class: dropdown

Follow this link for a detailed description of the input arguments, raised errors, and returned objects of this function: {func}`molsysmt.build.editable`.
:::


## Basic usage

Let's show how to make an existing molecular system editable using Met-enkephalin:


```python
import molsysmt as msm
```

```python
molsys = msm.convert(msm.systems['Met-enkephalin']['met_enkephalin.h5msm'])
```

:::{note} Demo Systems Catalog
:class: dropdown

This tutorial uses demonstration datasets provided by MolSysMT. To explore the full catalog of bundled systems, forms, and file paths, visit the {ref}`Demo Systems <user-foundations-entrance-demo-systems>` guide.
:::


Calling {func}`molsysmt.build.editable` on `molsys` returns an active {class}`molsysmt.MolSysBuilder` instance:


```python
builder = msm.build.editable(molsys)
```

```python
type(builder)
```

## Inspecting the builder

A builder can be queried directly with standard functions like {func}`molsysmt.basic.get` or {func}`molsysmt.basic.info`:


```python
msm.get(builder, n_atoms=True, n_bonds=True, n_chains=True)
```

## Modifying topology and bonds

Using the builder, we can add new covalent bonds between specific atom indices:


```python
builder.add_bond(atom_index_1=0, atom_index_2=10, bond_order=1, bond_type='covalent')
```

We can also remove specific covalent bonds by specifying their bond indices, or remove all bonds by passing `'all'`:


```python
builder.remove_bonds([0, 1])
```

Residue groups can be reassigned to a new chain:


```python
builder.assign_groups_to_new_chain(group_indices=[0, 1], chain_id='B', chain_name='B')
```

## Materializing the final molecular system

Once edits are completed, calling `.build()` compiles and validates the staging model into a new native {class}`molsysmt.MolSys` object:


```python
new_molsys = builder.build()
```

```python
msm.info(new_molsys, element='chain')
```

## Assembling a molecular system from scratch

Calling {func}`molsysmt.build.editable` without arguments creates an empty builder ready for incremental model declaration:


```python
empty_builder = msm.build.editable()
```

We declare atoms, group them into residues and chains, add covalent bonds, and assign 3D coordinates:


```python
import numpy as np

empty_builder.add_atom(atom_name='N', atom_type='N')
empty_builder.add_atom(atom_name='CA', atom_type='C')
empty_builder.add_atom(atom_name='C', atom_type='C')
empty_builder.add_group(atom_indices=[0, 1, 2], group_name='ALA', group_id='1', group_type='amino_acid')
empty_builder.add_chain(group_indices=[0], chain_id='A', chain_name='A')
empty_builder.add_bond(atom_index_1=0, atom_index_2=1, bond_order=1, bond_type='covalent')
empty_builder.add_bond(atom_index_1=1, atom_index_2=2, bond_order=1, bond_type='covalent')

coords = np.array([[[0.0, 0.0, 0.0], [0.145, 0.0, 0.0], [0.245, 0.12, 0.0]]]) * msm.pyunitwizard.unit('nm')
empty_builder.set_coordinates(coords)
```

```python
scratch_molsys = empty_builder.build()
```

```python
msm.info(scratch_molsys)
```

# Get disulfide bonds

*Identifying disulfide bridges between sulfur atoms in protein structures.*

Disulfide bonds are covalent links that typically form between the sulfur atoms of cysteine residues in proteins, playing a key role in stabilizing tertiary and quaternary structures. The function {func}`molsysmt.build.get_disulfide_bonds` detects disulfide bonds by identifying bonded sulfur atom pairs within a maximum distance threshold.

:::{versionadded} 1.0.0
:::

:::{admonition} API documentation
:class: dropdown

Follow this link for a detailed description of the input arguments, raised errors, and returned objects of this function: {func}`molsysmt.build.get_disulfide_bonds`.
:::


## Basic usage

Let's show how this function works using a protein structure loaded from PDB ID `5XJH`:


```python
import molsysmt as msm
```

```python
molsys = msm.convert('5XJH')
```

We now apply the function to retrieve disulfide bonds based on the default maximum sulfur–sulfur distance:


```python
msm.build.get_disulfide_bonds(molsys)
```

The default threshold for S–S bond detection is retrieved from a predefined dictionary of typical bond lengths:


```python
msm.element.bond.max_expected_bond_length['protein']['S']['S']
```

Let's check how many sulfur atoms are present in the system using {func}`molsysmt.basic.get`:


```python
msm.get(molsys, element='atom', selection='atom_type=="S"', n_atoms=True)
```

Next, we find pairs of sulfur atoms that are mutual nearest neighbors using {func}`molsysmt.structure.get_neighbors`:


```python
msm.structure.get_neighbors(molsys, selection='atom_type=="S"', n_neighbors=1, output_type='pairs', mutual_only=True, output_indices='atom')
```

By slightly increasing the maximum bond length threshold to `2.15 angstroms`, an additional disulfide bond is identified:


```python
msm.build.get_disulfide_bonds(molsys, max_bond_length='2.15 angstroms')
```

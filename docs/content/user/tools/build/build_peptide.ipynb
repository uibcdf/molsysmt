# Build peptide

*Building peptides from amino acid sequence.*

Peptides are essential building blocks in molecular modeling. The function {func}`molsysmt.build.build_peptide` constructs an all-atom molecular system with topology and coordinates directly from an amino acid sequence, supporting three-letter or single-letter codes and customizable terminal cappings.

:::{versionadded} 1.0.0
:::

:::{admonition} API documentation
:class: dropdown

Follow this link for a detailed description of the input arguments, raised errors, and returned objects of this function: {func}`molsysmt.build.build_peptide`.
:::


## Basic usage

Let's show how to build a capped dipeptide (`AceAlaNme`) from sequence:


```python
import molsysmt as msm
```

```python
molsys = msm.build.build_peptide('AceAlaNme')
```

```python
molsysviewer_htmlfile = '_static/views/tools_build_build_peptide_1.html'
```

```python
msm.view(molsys)
```

Next, let's build a longer cationic peptide sequence (`GRKFRRKFKK`) and solvate it in a simulation box:


```python
molsys = msm.build.build_peptide('GRKFRRKFKK')
```

We cap the terminal residues with `ACE` and `NME`:


```python
molsys = msm.build.add_missing_terminal_cappings(molsys, N_terminal='ACE', C_terminal='NME')
```

Center the peptide in space before solvation:


```python
molsys = msm.structure.center(molsys)
```

Solvate the system in a truncated octahedral box with 14.0 Å clearance:


```python
molsys = msm.build.solvate(molsys, box_shape='truncated octahedral', clearance='14.0 angstroms')
```

Wrap water molecules to the primary unit cell using the minimum image convention:


```python
molsys = msm.pbc.wrap_to_mic(molsys)
```

Finally, visualize the solvated system:


```python
molsysviewer_htmlfile = '_static/views/tools_build_build_peptide_2.html'
```

```python
msm.view(molsys)
```

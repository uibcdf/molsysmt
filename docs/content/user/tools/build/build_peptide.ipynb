# Build peptide

*Building peptides from amino acid sequence.*

Peptides are essential building blocks in molecular modeling. The function {func}`molsysmt.build.build_peptide` constructs an all-atom molecular system with topology and coordinates directly from an amino acid sequence, supporting three-letter or single-letter codes and terminal cappings specified in the sequence. It uses MolSysMT's native engine by default; pass `engine='LEaP'` to use the optional AmberTools executable. Both sequence notations work without Biopython.

The optional `engine="LEaP"` builds and converts the peptide inside an owned temporary directory. That directory is removed even if preparation, execution or conversion fails. File and cleanup errors remain visible; an unsuccessful call does not return a completed peptide.

## Basic usage

Let's show how to build a capped dipeptide (`AceAlaNme`) from sequence:


```python
import molsysmt as msm
```

```python
molsys = msm.build.build_peptide('AceAlaNme')
```

```python
molsysviewer_htmlfile = '_static/views/tools_build_build_peptide_1.html'
```

```python
msm.view(molsys)
```

Next, let's build a longer cationic peptide sequence (`GRKFRRKFKK`) and solvate it in a simulation box:


```python
molsys = msm.build.build_peptide('GRKFRRKFKK')
```

We cap the terminal residues with `ACE` and `NME`:


```python
molsys = msm.build.add_missing_terminal_cappings(molsys, N_terminal='ACE', C_terminal='NME')
```

Center the peptide in space before solvation:


```python
molsys = msm.structure.center(molsys)
```

Solvate the system in a truncated octahedral box with 14.0 Å clearance:


```python
molsys = msm.build.solvate(molsys, box_shape='truncated octahedral', clearance='14.0 angstroms')
```

Wrap water molecules to the primary unit cell using the minimum image convention:


```python
molsys = msm.pbc.wrap_to_mic(molsys)
```

Finally, visualize the solvated system:


```python
molsysviewer_htmlfile = '_static/views/tools_build_build_peptide_2.html'
```

```python
msm.view(molsys)
```

## Retaining source naming context

Construction reads exact labels rather than substituting their sequence parents.
For an ambiguous state label, declare the source convention. Native products
record the Amber vocabulary used by their construction references; the input is
unchanged. Only standard twenty names, eight reviewed Amber states and existing
caps have a supported construction route.


```python
molsys = msm.build.build_peptide(
    'amino_acids_3:ALAHIDALA', from_naming_scheme='amber')
print(molsys.topology.groups.group_name.tolist())
print(molsys.topology.group_naming_schemes)
print(msm.convert(molsys, to_form='string:amino_acids_1'))
```

The group convention does not force one spelling for hydrogen atom names.
The checked state layouts include historical and Amber construction H names;
preparation recognizes the observed layout without adding an alias duplicate.


```python
n_before = msm.get(molsys, n_atoms=True)
molsys = msm.build.add_missing_hydrogens(molsys, engine='MolSysMT')
print('No alias hydrogens added:', msm.get(molsys, n_atoms=True) == n_before)
```

Native construction supplies initial coordinates and atom/bond inventory;
it does not certify formal charges, bond orders, disulfide pairing, terminal
chemistry or environmental protonation. Geometry remains approximate. Optional
Ackredit credits the actual bundled construction data, without claiming execution
of Amber/LEaP. The historical asset has no pinned upstream revision.

A Topology/TopologyDict output retains naming context. Full MolSysDict/H5MSM
transport still rejects nonempty MolecularMechanics; omit that domain explicitly
when it is outside your intended output. No mechanics data are silently dropped.
See {ref}`user-foundations-group-naming-schemes`.

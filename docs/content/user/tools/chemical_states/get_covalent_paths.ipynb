# Get covalent paths

*Finding paths of covalently bonded atoms matching specified element patterns.*

The function {func}`molsysmt.chemical_states.get_covalent_paths` identifies all linear sequences of covalently bonded atoms that match a specified pattern of atom names or element criteria.

:::{versionadded} 1.0.0
:::

:::{admonition} API documentation
:class: dropdown

Follow this link for a detailed description of the input arguments, raised errors, and returned objects of this function: {func}`molsysmt.chemical_states.get_covalent_paths`.
:::


## Basic usage

Let's show how to find covalent paths matching atom name patterns using TcTIM (PDB ID: 1TCD):


```python
import molsysmt as msm
```

```python
molsys = msm.convert(msm.systems['TcTIM']['1tcd.h5msm'], to_form='molsysmt.MolSys')
```

:::{note}
:class: dropdown

**Demo Systems Catalog**

This tutorial uses demonstration datasets provided by MolSysMT. To explore the full catalog of bundled systems, forms, and file paths, visit the {ref}`Demo Systems <user-foundations-entrance-demo-systems>` guide.
:::


We inspect structural information with {func}`molsysmt.basic.info`:


```python
msm.info(molsys)
```

We search for all 3-atom covalent paths matching the sequence `C -> N -> CA` using {func}`molsysmt.chemical_states.get_covalent_paths`:


```python
covalent_paths = msm.chemical_states.get_covalent_paths(molsys, path=['atom_name=="C"', 'atom_name=="N"', 'atom_name=="CA"'])
```

```python
covalent_paths.shape
```

```python
covalent_paths
```

We check the atom names along the first retrieved covalent path using {func}`molsysmt.basic.get`:


```python
msm.get(molsys, element='atom', selection=covalent_paths[0], name=True)
```

## Using multi-name atom selections

Selection strings at each position can match multiple atom names:


```python
covalent_paths = msm.chemical_states.get_covalent_paths(molsys, path=['atom_name==["C", "N"]', 'atom_name=="CA"', 'atom_name=="C"'])
```

```python
covalent_paths
```

```python
msm.get(molsys, element='atom', selection=covalent_paths[0], name=True)
```

```python
msm.get(molsys, element='atom', selection=covalent_paths[1], name=True)
```

## Source indices and empty patterns

Each entry in `path` describes one position in a bonded walk. Numeric selectors refer to the complete source system, even when `selection` restricts the candidate atoms. A one-position pattern such as `[[0, 1]]` returns two one-atom walks; it does not ask for the shortest path between those atoms. Walks can revisit atoms.

An empty position, or candidate positions without connecting bonds, gives no matches. The result keeps one column per pattern position and an `int64` dtype, so downstream code can safely inspect either column. All selectors are validated even if another position is empty. `path=None` and an empty outer pattern are invalid.

```python
single_atom_walks = msm.chemical_states.get_covalent_paths(molsys, path=[[0, 1]])
single_atom_walks
```

```python
no_matches = msm.chemical_states.get_covalent_paths(molsys, path=[[0], []])
no_matches.shape, str(no_matches.dtype)
```

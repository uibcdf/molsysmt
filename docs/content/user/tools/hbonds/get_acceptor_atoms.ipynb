# Get acceptor atoms

*Identifying hydrogen bond acceptor atoms based on chemical rules and electronegative centers.*

The function {func}`molsysmt.hbonds.get_acceptor_atoms` identifies potential hydrogen bond acceptor atoms within molecular systems using customizable chemical inclusion and exclusion selection criteria.

:::{versionadded} 1.0.0
:::

:::{admonition} API documentation
:class: dropdown

Follow this link for a detailed description of the input arguments, raised errors, and returned objects of this function: {func}`molsysmt.hbonds.get_acceptor_atoms`.
:::


## Basic usage

Let's identify the hydrogen bond acceptor atoms in a protein structure:


```python
import molsysmt as msm
```

```python
molsys = msm.convert(msm.systems['chicken villin HP35']['chicken_villin_HP35.h5msm'])
```

:::{note} Demo Systems Catalog
:class: dropdown

This tutorial uses demonstration datasets provided by MolSysMT. To explore the full catalog of bundled systems, forms, and file paths, visit the {ref}`Demo Systems <user-foundations-entrance-demo-systems>` guide.
:::


We inspect the built-in default inclusion and exclusion rules defined in MolSysMT:


```python
print('Default inclusion rules:', msm.hbonds.acceptor_inclusion_rules)
print('Default exclusion rules:', msm.hbonds.acceptor_exclusion_rules)
```

Now let's find all acceptor atoms within the first three residues (`group_index in [1, 2, 3]`):


```python
acceptor_atoms = msm.hbonds.get_acceptor_atoms(molsys, selection='group_index==[1, 2, 3]')
print('Acceptor atom indices:', acceptor_atoms)
```

## Inspecting acceptor atom metadata

We can query detailed attribute information for the identified acceptor atoms:


```python
msm.info(molsys, element='atom', selection=acceptor_atoms)
```

## Custom inclusion and exclusion rules

Users can specify custom inclusion or exclusion rules to tailor acceptor definitions for non-standard molecules or specific chemical centers:


```python
custom_acceptors = msm.hbonds.get_acceptor_atoms(molsys, selection='group_index==[1, 2, 3]',
                                                 inclusion_rules="atom_name=='O'")
msm.info(molsys, element='atom', selection=custom_acceptors)
```

# Get donor atoms

*Identifying hydrogen bond donor pairs (heavy atom and bonded hydrogen).*

The function {func}`molsysmt.hbonds.get_donor_atoms` detects donor heavy atoms along with their covalently bonded hydrogen atoms for hydrogen-bond geometric analysis.

:::{versionadded} 1.0.0
:::

:::{admonition} API documentation
:class: dropdown

Follow this link for a detailed description of the input arguments, raised errors, and returned objects of this function: {func}`molsysmt.hbonds.get_donor_atoms`.
:::


## Basic usage

Let's identify hydrogen bond donor pairs in a protein system:


```python
import molsysmt as msm
```

```python
molsys = msm.convert(msm.systems['chicken villin HP35']['chicken_villin_HP35.h5msm'])
```

:::{note} Demo Systems Catalog
:class: dropdown

This tutorial uses demonstration datasets provided by MolSysMT. To explore the full catalog of bundled systems, forms, and file paths, visit the {ref}`Demo Systems <user-foundations-entrance-demo-systems>` guide.
:::


We inspect the built-in default inclusion and exclusion rules for hydrogen bond donors:


```python
print('Default donor inclusion rules:', msm.hbonds.donor_inclusion_rules)
print('Default donor exclusion rules:', msm.hbonds.donor_exclusion_rules)
```

We detect donor pairs for the first three residues (`group_index in [1, 2, 3]`):


```python
donor_pairs = msm.hbonds.get_donor_atoms(molsys, selection='group_index==[1, 2, 3]')
print('Donor pairs (heavy_atom, bonded_H):\n', donor_pairs)
```

## Inspecting donor atom metadata

We inspect the heavy atoms (column 0) and bonded hydrogens (column 1) comprising the detected donor pairs:


```python
heavy_donors = donor_pairs[:, 0]
bonded_hydrogens = donor_pairs[:, 1]

print('Heavy donor atoms:')
msm.info(molsys, element='atom', selection=heavy_donors)
```

## Custom inclusion and exclusion rules

Custom donor rules allow focusing exclusively on backbone amide donors or specific residue types:


```python
backbone_donors = msm.hbonds.get_donor_atoms(molsys, selection='group_index==[1, 2, 3]',
                                             inclusion_rules="atom_name=='N'")
msm.info(molsys, element='atom', selection=backbone_donors[:, 0])
```

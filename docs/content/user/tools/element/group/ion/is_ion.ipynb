# Is ion

*Validating whether a chemical group represents a recognized monoatomic or polyatomic ion.*

The function {func}`molsysmt.element.group.ion.is_ion` checks if a group name matches standard ion definitions.

## Basic usage

Let's test various ion symbols:


```python
import molsysmt as msm
```

```python
print('Is NA an ion?:', msm.element.group.ion.is_ion('NA'))
print('Is CL an ion?:', msm.element.group.ion.is_ion('CL'))
print('Is MG an ion?:', msm.element.group.ion.is_ion('MG'))
print('Is ALA an ion?:', msm.element.group.ion.is_ion('ALA'))
```

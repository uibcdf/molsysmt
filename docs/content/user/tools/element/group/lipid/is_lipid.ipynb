# Is lipid

*Validating whether a residue is a recognized lipid molecule.*

The function {func}`molsysmt.element.group.lipid.is_lipid` checks group names against standard lipid definitions (e.g. `POPC`, `POPE`, `DPPC`, `CHOL`).

## Basic usage

Let's test lipid names:


```python
import molsysmt as msm
```

```python
print('Is POPC a lipid?:', msm.element.group.lipid.is_lipid('POPC'))
print('Is CHOL a lipid?:', msm.element.group.lipid.is_lipid('CHOL'))
print('Is ALA a lipid?:', msm.element.group.lipid.is_lipid('ALA'))
```

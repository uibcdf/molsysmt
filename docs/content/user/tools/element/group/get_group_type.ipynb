# Get group type

*Querying the chemical group category (amino acid, water, ion, lipid, etc.) for residue names.*

The function {func}`molsysmt.element.group.get_group_type` categorizes group or residue names into standard structural classifications.

:::{versionadded} 1.0.0
:::

:::{admonition} API documentation
:class: dropdown

## Basic usage

Let's identify the group classification for common residue codes:


```python
import molsysmt as msm
```

```python
residues = ['ALA', 'HOH', 'NA', 'POPC', 'DA', 'ACE', 'NME']
for res in residues:
    g_type = msm.element.group.get_group_type(res)
    print(f'Group {res:<4} -> Group type: {g_type}')
```

# Get group type

*Querying group types from molecular systems and topologies.*

The function {func}`molsysmt.element.group.get_group_type` retrieves or infers the structural group classification (`amino acid`, `water`, `ion`, `lipid`, `nucleotide`, etc.) for residues in a molecular system.

:::{versionadded} 1.0.0
:::

:::{admonition} API documentation
:class: dropdown

Follow this link for a detailed description of the input arguments, raised errors, and returned objects of this function: {func}`molsysmt.element.group.get_group_type`.
:::


## Basic usage

Let's retrieve the group types across all residues of the Barnase-Barstar complex:


```python
import molsysmt as msm
```

```python
molsys = msm.convert(msm.systems['Barnase-Barstar']['barnase_barstar.h5msm'])
```

:::{note}
:class: dropdown

**Demo Systems Catalog**

This tutorial uses demonstration datasets provided by MolSysMT. To explore the full catalog of bundled systems, forms, and file paths, visit the {ref}`Demo Systems <user-foundations-entrance-demo-systems>` guide.
:::


```python
group_types = msm.element.group.get_group_type(molsys, element='group', selection='all')
print('Total residues:', len(group_types))
print('Unique group classifications:', set(group_types))
print('First 5 group types:', group_types[:5])
```

## Querying group types per atom

We can also query the group classification corresponding to individual atoms:


With `redefine_types=True`, this query infers types using local evidence and
each group's source naming declaration; otherwise it returns stored types.
Same-name groups can have different meanings under Amber and CCD. Inference
leaves the input unchanged and never repartitions its stable molecules. To
write inferred group types, explicitly call `Topology.rebuild_groups` with
`redefine_types=True`. Assigning a naming declaration alone does not rewrite
types. See {ref}`user-foundations-group-naming-schemes`.


```python
atom_group_types = msm.element.group.get_group_type(molsys, element='atom', selection=[0, 1, 2, 3])
print('Group types for atoms 0-3:', atom_group_types)
```

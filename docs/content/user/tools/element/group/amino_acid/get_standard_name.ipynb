# Get standard name

*Reporting reference amino-acid names under source naming conventions.*

The function {func}`molsysmt.element.group.amino_acid.get_standard_name` reports the assessed parent of a modified amino-acid group. A parent is useful for sequence comparison; it does not establish chemical equivalence or permission to use the parent's repair template.

:::{versionadded} 1.0.0
:::

:::{admonition} API documentation
:class: dropdown

Follow this link for a detailed description of the input arguments, raised errors, and returned objects of this function: {func}`molsysmt.element.group.amino_acid.get_standard_name`.
:::

## Basic usage

The four reviewed modified names below have matching parents in the supported default subset. Canonical and unassessed default names return `None`. Ambiguous names such as `GLH` require `naming_scheme`. See {ref}`user-foundations-group-naming-schemes` for bounded coverage and source revisions.

```python
import molsysmt as msm
```

```python
modified = ['MSE', 'SEP', 'TPO', 'PTR']
for group_name in modified:
    standard = msm.element.group.amino_acid.get_standard_name(group_name)
    print(f'{group_name} -> {standard}')
```

## Declaring context

Amber and the historical legacy table give different parents to `GLH`. CCD `CYX` is a non-peptide component, so it has no reference amino-acid parent. An explicit unsupported profile/name pair raises an error; there is no fallback to another convention.

This lookup leaves the original label and chemical assignments unchanged. Optional Ackredit records the reached datasets, without running Amber or MDTraj.

```python
for naming_scheme in ('amber', 'legacy'):
    print(naming_scheme, msm.element.group.amino_acid.get_standard_name(
        'GLH', naming_scheme=naming_scheme))
print('ccd CYX', msm.element.group.amino_acid.get_standard_name(
    'CYX', naming_scheme='ccd'))
```

# Get standard name

*Mapping modified or non-canonical amino acid residue names to their parent canonical standard residue.*

The function {func}`molsysmt.element.group.amino_acid.get_standard_name` resolves modified amino acids (e.g. `MSE`, `SEP`, `TPO`, `PTR`) to their standard 3-letter equivalent (`MET`, `SER`, `THR`, `TYR`).

:::{versionadded} 1.0.0
:::

:class: dropdown

:::


## Basic usage

Let's map modified residue names to their standard counterparts:


```python
import molsysmt as msm
```

```python
modified = ['MSE', 'SEP', 'TPO', 'PTR', 'HSD', 'HSE']
for res in modified:
    std = msm.element.group.amino_acid.get_standard_name(res)
    print(f'Modified residue {res:<4} -> Standard residue: {std}')
```

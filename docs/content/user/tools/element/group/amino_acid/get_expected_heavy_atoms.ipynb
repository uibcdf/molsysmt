# Get expected heavy atoms

*Retrieving the expected set of non-hydrogen heavy atoms for an amino acid residue.*

The function {func}`molsysmt.element.group.amino_acid.get_expected_heavy_atoms` returns the heavy atoms expected from an exact residue template. It returns `None` when no supported reference is available. MSE, SEP, TPO and MLY have their own CCD-derived heavy-atom references; a sequence equivalent does not establish chemical equivalence.

:::{versionadded} 1.0.0
:::

## Basic usage

Let's retrieve the expected heavy atoms for alanine and phenylalanine:


```python
import molsysmt as msm
```

```python
ala_heavy = msm.element.group.amino_acid.get_expected_heavy_atoms('ALA')
print('ALA expected heavy atoms:', sorted(list(ala_heavy)))
phe_heavy = msm.element.group.amino_acid.get_expected_heavy_atoms('PHE')
print('PHE expected heavy atoms:', sorted(list(phe_heavy)))
```

## Declaring chemical identity

The standard twenty names use the historical atom/bond catalogue. MSE, SEP,
TPO and MLY use separate CCD-derived heavy-only references under `ccd` or the
unambiguous default. Other modified groups can be unassessed; a parent name
never authorizes substituting a standard template.

Explicit Amber state labels use only catalogue variants whose heavy connectivity
and H counts per parent atom match the bundled Meeko-derived state fragments.
Other raw variants under the same label can describe different compounds.
CCD HID is non-peptide; CCD HIP and LYN also cannot reuse the Amber references.
An incompatible supplied heavy-atom layout returns `None` instead of falling
back to the first variant. Bare ambiguous labels raise an argument error.

Optional Ackredit records the catalogue and actual verification data, without
claiming that Meeko or Amber ran. Inventory support does not imply that coordinate
placement, bond orders, charge or stereochemistry are validated. See
{ref}`user-foundations-group-naming-schemes`.


```python
amber_glh = msm.element.group.amino_acid.get_expected_heavy_atoms(
    'GLH', naming_scheme='amber')
print('Amber GLH:', sorted(amber_glh))
print('CCD HID:', msm.element.group.amino_acid.get_expected_heavy_atoms(
    'HID', naming_scheme='ccd'))
print('Exact MSE has SE:', 'SE' in msm.element.group.amino_acid.get_expected_heavy_atoms('MSE'))
```

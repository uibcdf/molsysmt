# Get bonded atom pairs

*Extracting covalent bond index pairs for a standard amino acid residue.*

The function {func}`molsysmt.element.group.amino_acid.get_bonded_atom_pairs` computes covalent bond connections among the given atom names.

:::

## Basic usage

Let's retrieve the bond pairs for alanine:


```python
import molsysmt as msm
```

```python
atom_names = ['N', 'CA', 'C', 'O', 'CB']
bonds = msm.element.group.amino_acid.get_bonded_atom_pairs('ALA', atom_names)
print('Bond index pairs in ALA backbone + CB:', bonds)
```

# Get bonded atom pairs

*Looking up reference bond index pairs for an amino-acid group.*

The function {func}`molsysmt.element.group.amino_acid.get_bonded_atom_pairs` looks up bundled reference connections among the supplied atom names.

:::{versionadded} 1.0.0
:::

:::{admonition} API documentation
:class: dropdown

Follow this link for a detailed description of the input arguments, raised errors, and returned objects of this function: {func}`molsysmt.element.group.amino_acid.get_bonded_atom_pairs`.

The endpoints are integer indices, not atom names. Without `atom_indices`,
these are local positions in your supplied name sequence. With an explicit map,
supply one integer per name in matching order. Each pair puts the smaller mapped
index first; `sorted=True` additionally sorts the rows. A known empty name list
gives `[]`; unknown group names or unmatched layouts give `None`.

Repeated names expand a reference edge over the matching supplied occurrences;
the lookup does not decide which alternate locations coexist. Caller inputs are
preserved, and no bonds or chemical assignments are added to a molecular system.
See {ref}`Tutorial_Get_bonded_atom_pairs` for the other families' conventions,
including oxygen-first water pairs.

Explicit maps are validated before reference access: use one nonnegative
int64-range integer per name. Repeated indices are allowed. Invalid values
raise `ArgumentError`; missing or extra entries raise `ArgumentLengthError`.
With `skip_digestion=True`, provide inputs already satisfying this contract.


## Basic usage

Let's retrieve the bond pairs for alanine:


```python
import molsysmt as msm
```

```python
atom_names = ['N', 'CA', 'C', 'O', 'CB']
bonds = msm.element.group.amino_acid.get_bonded_atom_pairs('ALA', atom_names)
print('Bond index pairs in ALA backbone + CB:', bonds)
```

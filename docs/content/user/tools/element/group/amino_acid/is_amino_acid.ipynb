# Is amino acid

*Recognizing assessed amino-acid labels under source conventions.*

The function {func}`molsysmt.element.group.amino_acid.is_amino_acid` recognizes amino-acid names within the selected bounded naming profile. Recognition does not imply that an exact chemical template is available.

:::{versionadded} 1.0.0
:::

:::{admonition} API documentation
:class: dropdown

Follow this link for a detailed description of the input arguments, raised errors, and returned objects of this function: {func}`molsysmt.element.group.amino_acid.is_amino_acid`.
:::

## Basic usage

Unambiguous assessed names can be queried without a convention. `False` can mean a known non-peptide component or an unassessed name; it is not proof that an unknown compound cannot be an amino acid.

```python
import molsysmt as msm
```

```python
print('Is ALA an amino acid?:', msm.element.group.amino_acid.is_amino_acid('ALA'))
print('Is MSE an amino acid?:', msm.element.group.amino_acid.is_amino_acid('MSE'))
print('Is HOH an amino acid?:', msm.element.group.amino_acid.is_amino_acid('HOH'))
```

## Comparing conventions

The same `HID` label is a histidine state in Amber and a non-peptide component in the reviewed CCD snapshot. Calling this scalar query without a convention raises an argument error. See {ref}`user-foundations-group-naming-schemes` for the complete scope. Optional Ackredit credits reached reference data.

```python
for naming_scheme in ('amber', 'ccd'):
    print(naming_scheme, msm.element.group.amino_acid.is_amino_acid(
        'HID', naming_scheme=naming_scheme))
print('Unassessed:', msm.element.group.amino_acid.is_amino_acid('PCA'))
```

# Get expected hydrogens

*Retrieving the expected hydrogen atom names for an amino acid under given chemical conditions.*

The function {func}`molsysmt.element.group.amino_acid.get_expected_hydrogens` queries bounded H-name references under explicit group naming context. Standard labels retain the approximate pH model; named Amber states constrain the side-chain inventory.

:::{versionadded} 1.0.0
:::

:class: dropdown

## Basic usage

Let's query the expected hydrogens for alanine:


```python
import molsysmt as msm
```

```python
h_atoms = msm.element.group.amino_acid.get_expected_hydrogens('ALA')
print('ALA expected hydrogens at pH 7.4:', h_atoms)
```

## Declaring a named state

Standard amino-acid names retain the existing fixed-pKa approximation; it is
not an environmental pKa prediction. The eight explicit Amber state labels
GLH/HID/HIE/HIP/ASH/CYM/CYX/LYN instead constrain their side-chain inventory.
The reference variants are checked against the bundled Meeko-derived heavy
connectivity and H counts per parent atom; no sequence-parent substitution occurs.

`None` means unassessed, including modified heavy-only references, unsupported
profile/name combinations, incompatible heavy-atom layouts and named-state
terminal H completion. `[]` would mean that a supported reference expects no H.
These references currently cover the named states as internal peptide groups;
do not infer terminal completeness from them. Raw catalogue variants can describe
different compounds under the same label. Optional Ackredit records the data
actually used. See {ref}`user-foundations-group-naming-schemes` and
{ref}`Tutorial_Fixed_State_Hydrogens` for prepared ChemicalStates workflows.


```python
glh_h = msm.element.group.amino_acid.get_expected_hydrogens(
    'GLH', naming_scheme='amber', pH=12.0)
print('Named GLH keeps HE2:', 'HE2' in glh_h)
print('Named HID terminus:', msm.element.group.amino_acid.get_expected_hydrogens(
    'HID', naming_scheme='amber', is_n_terminal=True))
print('Heavy-only MSE:', msm.element.group.amino_acid.get_expected_hydrogens('MSE'))
```

*Inspecting the internal topological database record for an amino acid group.*

## Basic usage

Let's inspect the database entry for alanine:


```python
import molsysmt as msm
```

```python
db = msm.element.group.amino_acid.get_group_db('ALA')
print('Keys in ALA database entry:', list(db.keys()))
print('Atoms in template:', db.get('atoms', [])[:6])
```

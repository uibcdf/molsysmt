# Is terminal capping

*Validating whether a residue is a recognized peptide terminal capping group.*

The function {func}`molsysmt.element.group.terminal_capping.is_terminal_capping` checks for N-terminal and C-terminal capping groups (`ACE`, `NME`, `NH2`, etc.).

## Basic usage

Let's test capping group names:


```python
import molsysmt as msm
```

```python
print('Is ACE a terminal capping?:', msm.element.group.terminal_capping.is_terminal_capping('ACE'))
print('Is NME a terminal capping?:', msm.element.group.terminal_capping.is_terminal_capping('NME'))
print('Is ALA a terminal capping?:', msm.element.group.terminal_capping.is_terminal_capping('ALA'))
```

# Is water

*Validating whether a group name corresponds to a recognized water model representation.*

The function {func}`molsysmt.element.group.water.is_water` tests group names against standard water models (e.g. `HOH`, `WAT`, `TIP3`, `SOL`).

## Basic usage

Let's test water residue names:


```python
import molsysmt as msm
```

```python
print('Is HOH water?:', msm.element.group.water.is_water('HOH'))
print('Is WAT water?:', msm.element.group.water.is_water('WAT'))
print('Is TIP3 water?:', msm.element.group.water.is_water('TIP3'))
print('Is ALA water?:', msm.element.group.water.is_water('ALA'))
```

# Get group type from group name

*Recognizing group types under source naming conventions.*

The function {func}`molsysmt.element.group.get_group_type_from_group_name` reports recognized types for group labels. Naming-dependent amino-acid meanings use the shared bounded resolver; existing independent water, ion and other non-peptide catalogs remain available.

:::{versionadded} 1.0.0

## Basic usage

Common supported group names can be recognized directly. Missing or unassessed names return `unknown`; no chemical identity is invented.

```python
import molsysmt as msm
```

```python
group_names = ['ALA', 'HOH', 'NA', 'POPC', 'DA', 'ACE', 'NME']
for group_name in group_names:
    group_type = msm.element.group.get_group_type_from_group_name(group_name)
    print(f'{group_name:<4} -> {group_type}')
```

## Declaring context

An explicit scalar query requires a convention for ambiguous names. Raw molecular readers instead preserve these labels with an unknown type, so they can be loaded before you declare their source vocabulary. Native type reconstruction uses the stored declaration of each group, including groups sharing the same label. See {ref}`user-foundations-group-naming-schemes`.

Changing the declaration alone does not rewrite stored types, chemistry or stable molecule membership. Querying with `get_group_type(..., redefine_types=True)` returns interpreted types without editing the input; an explicit `Topology.rebuild_groups(redefine_types=True)` writes reconstructed group types. Neither operation reassigns stable molecules.

```python
for naming_scheme in ('amber', 'ccd'):
    print(naming_scheme, msm.element.group.get_group_type_from_group_name(
        'HID', naming_scheme=naming_scheme))
print('Missing:', msm.element.group.get_group_type_from_group_name(None))
```

# Get group type from group name

*Inferring the group classification from a standard residue or chemical group name string.*

The function {func}`molsysmt.element.group.get_group_type_from_group_name` translates residue names (e.g. `ALA`, `HOH`, `POPC`, `ACE`) into standard group classifications.

:::{versionadded} 1.0.0
:::

## Basic usage

Let's identify the group classification for common residue codes:


```python
import molsysmt as msm
```

```python
residues = ['ALA', 'HOH', 'NA', 'POPC', 'DA', 'ACE', 'NME']
for res in residues:
    g_type = msm.element.group.get_group_type_from_group_name(res)
    print(f'Group {res:<4} -> Group type: {g_type}')
```

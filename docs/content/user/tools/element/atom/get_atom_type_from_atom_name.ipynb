# Get atom type from atom name

*Inferring an atom type from a standard atom name.*

`get_atom_type_from_atom_name()` returns the element symbol associated with a recognized atom name. For an unknown name, it emits `UnknownAtomNameWarning` and returns the explicit fallback `"UNK"`; provide `atom_type` directly when working with a non-standard naming convention.


```python
msm.element.atom.get_atom_type_from_atom_name('H2')
```

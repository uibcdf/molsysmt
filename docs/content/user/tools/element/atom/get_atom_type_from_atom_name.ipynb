# Get atom type from atom name

*Inferring the chemical element symbol from a standard atom name string.*

The function {func}`molsysmt.element.atom.get_atom_type_from_atom_name` translates PDB atom names (such as `CA`, `HB2`, `OW`) to their canonical chemical element symbol.

:::{versionadded} 1.0.0
:::

:class: dropdown

## Basic usage

Let's look up the chemical element types for several standard biomolecular atom names:


```python
import molsysmt as msm
```

```python
names = ['CA', 'N', 'O', 'CB', 'H', 'OW', 'HW1', 'CL']
for name in names:
    atom_type = msm.element.atom.get_atom_type_from_atom_name(name)
    print(f'Atom {name:<4} -> Element {atom_type}')
```

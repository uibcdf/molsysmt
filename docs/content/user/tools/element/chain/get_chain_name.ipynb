# Get chain name

*Extracting descriptive names for chains in molecular systems.*

The function {func}`molsysmt.element.chain.get_chain_name` retrieves human-readable names assigned to chains.

:::{versionadded} 1.0.0
:::

:::{admonition} API documentation
:class: dropdown

Follow this link for a detailed description of the input arguments, raised errors, and returned objects of this function: {func}`molsysmt.element.chain.get_chain_name`.
:::


## Basic usage

Let's query the names across the system:


```python
import molsysmt as msm
```

```python
molsys = msm.convert(msm.systems['Barnase-Barstar']['barnase_barstar.h5msm'])
```

:::{note} Demo Systems Catalog
:class: dropdown

This tutorial uses demonstration datasets provided by MolSysMT. To explore the full catalog of bundled systems, forms, and file paths, visit the {ref}`Demo Systems <user-foundations-entrance-demo-systems>` guide.
:::


```python
names = msm.element.chain.get_chain_name(molsys, element='chain', selection='all')
print('Chain names:', names)
```

# Get n components

*Counting the total number of components in a molecular system.*

The function {func}`molsysmt.element.component.get_n_components` counts components satisfying given selection criteria.

## Basic usage

Let's count elements in the system:


```python
import molsysmt as msm
```

```python
molsys = msm.convert(msm.systems['Barnase-Barstar']['barnase_barstar.h5msm'])
```

:::{note} Demo Systems Catalog
:class: dropdown

This tutorial uses demonstration datasets provided by MolSysMT. To explore the full catalog of bundled systems, forms, and file paths, visit the {ref}`Demo Systems <user-foundations-entrance-demo-systems>` guide.
:::


```python
n = msm.element.component.get_n_components(molsys)
print('Total components:', n)
```

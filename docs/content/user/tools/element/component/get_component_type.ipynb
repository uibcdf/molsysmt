# Get component type

*Determining the structural and chemical classification of components.*

The function {func}`molsysmt.element.component.get_component_type` infers or retrieves the type classification (`protein`, `water`, `ion`, `small molecule`, etc.) for components.

:::{versionadded} 1.0.0
:::

:::{admonition} API documentation
:class: dropdown

Follow this link for a detailed description of the input arguments, raised errors, and returned objects of this function: {func}`molsysmt.element.component.get_component_type`.
:::


## Basic usage

Let's determine the types across the system:


```python
import molsysmt as msm
```

```python
molsys = msm.convert(msm.systems['Barnase-Barstar']['barnase_barstar.h5msm'])
```

:::{note} Demo Systems Catalog
:class: dropdown

This tutorial uses demonstration datasets provided by MolSysMT. To explore the full catalog of bundled systems, forms, and file paths, visit the {ref}`Demo Systems <user-foundations-entrance-demo-systems>` guide.
:::


```python
types = msm.element.component.get_component_type(molsys, element='component', selection='all')
print('Component types:', types)
```

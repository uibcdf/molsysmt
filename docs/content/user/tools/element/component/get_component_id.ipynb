# Get component id

*Extracting component identifier strings from molecular systems.*

The function {func}`molsysmt.element.component.get_component_id` retrieves the assigned ID strings for components in a molecular system.

:::{versionadded} 1.0.0
:::

:::{admonition} API documentation
:class: dropdown

Follow this link for a detailed description of the input arguments, raised errors, and returned objects of this function: {func}`molsysmt.element.component.get_component_id`.
:::


## Basic usage

Let's extract the IDs for a molecular complex:


```python
import molsysmt as msm
```

```python
molsys = msm.convert(msm.systems['Barnase-Barstar']['barnase_barstar.h5msm'])
```

:::{note} Demo Systems Catalog
:class: dropdown

This tutorial uses demonstration datasets provided by MolSysMT. To explore the full catalog of bundled systems, forms, and file paths, visit the {ref}`Demo Systems <user-foundations-entrance-demo-systems>` guide.
:::


```python
ids = msm.element.component.get_component_id(molsys, element='component', selection='all')
print('Component IDs:', ids)
```

# Is molecule type

*Validating whether a string represents a recognized molecule classification type.*

The function {func}`molsysmt.element.molecule.is_molecule_type` checks if a type name (e.g. `protein`, `lipid`, `water`, `small molecule`) is a recognized molecule classification.

## Basic usage

Let's test molecule classification strings:


```python
import molsysmt as msm
```

```python
print('Is "protein" a molecule type?:', msm.element.molecule.is_molecule_type('protein'))
print('Is "lipid" a molecule type?:', msm.element.molecule.is_molecule_type('lipid'))
print('Is "custom_species" a molecule type?:', msm.element.molecule.is_molecule_type('custom_species'))
```

# Get potential energy

*Evaluating total potential energy and force field energy components for molecular systems.*

The function {func}`molsysmt.molecular_mechanics.get_potential_energy` computes total potential energies and individual bonded and non-bonded force group contributions.

:::{versionadded} 1.0.0
:::

:::{admonition} API documentation
:class: dropdown

Follow this link for a detailed description of the input arguments, raised errors, and returned objects of this function: {func}`molsysmt.molecular_mechanics.get_potential_energy`.
:::


:::{note} Molecular Mechanics Backend Engine
:class: dropdown

Molecular mechanics and force evaluations in MolSysMT are executed using backend simulation engines (currently **OpenMM** by default). MolSysMT intentionally unifies the Pythonic interface and data structures while delegating energy evaluations, parameterizations, and gradient calculations to high-performance, battle-tested molecular simulation engines.
:::


## Basic usage

Let's compute the total potential energy of the Barnase–Barstar complex:


```python
import molsysmt as msm
import pyunitwizard as puw
```

```python
molsys = msm.convert(msm.systems['Barnase-Barstar']['barnase_barstar.h5msm'])
```

:::{note} Demo Systems Catalog
:class: dropdown

This tutorial uses demonstration datasets provided by MolSysMT. To explore the full catalog of bundled systems, forms, and file paths, visit the {ref}`Demo Systems <user-foundations-entrance-demo-systems>` guide.
:::


We calculate the total potential energy of the system:


```python
u_total = msm.molecular_mechanics.get_potential_energy(molsys)
print('Total potential energy:', u_total)
```

## Energy decomposition into force terms

Setting `decomposition=True` breaks down the total energy into individual force contributions (such as bonds, angles, torsions, and non-bonded interactions):


```python
u_decomp = msm.molecular_mechanics.get_potential_energy(molsys, decomposition=True)
for force_name, energy_val in u_decomp.items():
    print(f'{force_name}: {energy_val}')
```

## Selection-specific potential energy

We can compute the isolated internal potential energy of specific molecules or chains:


```python
u_barnase = msm.molecular_mechanics.get_potential_energy(molsys, selection='molecule_name=="BARNASE"')
u_barstar = msm.molecular_mechanics.get_potential_energy(molsys, selection='molecule_name=="BARSTAR"')

print('Barnase internal energy:', u_barnase)
print('Barstar internal energy:', u_barstar)
```

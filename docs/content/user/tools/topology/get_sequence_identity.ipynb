# Get sequence identity

*Computing the sequence identity percentage and matching residue index mappings between molecular systems.*

The function {func}`molsysmt.topology.get_sequence_identity` calculates the sequence identity percentage between polymer chains and returns corresponding aligned residue index maps.

:::{versionadded} 1.0.0
:::

:::{admonition} API documentation
:class: dropdown

:::


## Basic usage

Let's show how to compute sequence identity between T4 lysozyme variants (PDB IDs: 181L and 1L17):


```python
import molsysmt as msm
```

```python
molsys_1 = msm.convert('pdb_id:181l', to_form='molsysmt.MolSys')
molsys_2 = msm.convert('pdb_id:1l17', to_form='molsysmt.MolSys')
```

We compute the percentage identity and residue index mappings using {func}`molsysmt.topology.get_sequence_identity`:


```python
identity, intersection, ref_intersection = msm.topology.get_sequence_identity(molsys_1, selection='molecule_type=="protein"',
                                           reference_molecular_system=molsys_2, reference_selection='molecule_type=="protein"')
```

```python
identity
```

```python
print(intersection)
```

```python
print(ref_intersection)
```

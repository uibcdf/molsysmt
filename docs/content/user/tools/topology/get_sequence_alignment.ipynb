# Get sequence alignment

*Computing the amino acid or nucleotide sequence alignment between molecular systems.*

The function {func}`molsysmt.topology.get_sequence_alignment` aligns sequences extracted from polymer chains across two molecular systems using pairwise sequence alignment algorithms.

:::{versionadded} 1.0.0
:::

:::{admonition} API documentation
:class: dropdown

Follow this link for a detailed description of the input arguments, raised errors, and returned objects of this function: {func}`molsysmt.topology.get_sequence_alignment`.
:::


## Basic usage

Let's show how to align sequences of T4 lysozyme variants (PDB IDs: 181L and 1L17):


```python
import molsysmt as msm
```

```python
molsys_1 = msm.convert('pdb_id:181l', to_form='molsysmt.MolSys')
molsys_2 = msm.convert('pdb_id:1l17', to_form='molsysmt.MolSys')
```

We inspect the residue counts for both protein structures using {func}`molsysmt.basic.get`:


```python
msm.get(molsys_1, element='atom', selection='molecule_type=="protein"', n_groups=True)
```

```python
msm.get(molsys_2, element='atom', selection='molecule_type=="protein"', n_groups=True)
```

We display the prettyprinted sequence alignment using {func}`molsysmt.topology.get_sequence_alignment`:


```python
msm.topology.get_sequence_alignment(molsys_1, selection='molecule_type=="protein"',
                                    reference_molecular_system=molsys_2, reference_selection='molecule_type=="protein"',
                                    prettyprint=True)
```

Alternatively, we can retrieve the aligned sequence strings as a tuple:


```python
seq, seq_ref = msm.topology.get_sequence_alignment(molsys_1, selection='molecule_type=="protein"',
                                    reference_molecular_system=molsys_2, reference_selection='molecule_type=="protein"')
```

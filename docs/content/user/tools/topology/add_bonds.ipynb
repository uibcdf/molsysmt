# Add bonds

*Adding covalent bonds between specified atom pairs in a molecular system.*

Covalent bonds in a native `molsysmt.Topology` or `molsysmt.MolSys` object can be added directly via the `topology.add_bonds()` method, specifying pairs of atom indices. For a full mutable staging workflow that supports adding atoms, residues, and bonds interactively, see also {func}`molsysmt.build.editable` ({ref}`Editable <Tutorial_Editable>`).

:::{versionadded} 1.0.0
:::

:::{admonition} API documentation
:class: dropdown

Follow this link for a detailed description of the input arguments, raised errors, and returned objects of this method: {meth}`molsysmt.Topology.add_bonds`.
:::


## Basic usage

Let's show how to add covalent bonds using an alanine dipeptide system:


```python
import molsysmt as msm
```

```python
molsys = msm.systems['alanine dipeptide']['alanine_dipeptide.h5msm']
molsys = msm.convert(molsys)
```

:::{note} Demo Systems Catalog
:class: dropdown

This tutorial uses demonstration datasets provided by MolSysMT. To explore the full catalog of bundled systems, forms, and file paths, visit the {ref}`Demo Systems <user-foundations-entrance-demo-systems>` guide.
:::


First, let's remove all existing covalent bonds from the system to simulate a topology lacking bond definitions:


```python
molsys.topology.remove_bonds()
```

Let's confirm that the system currently has zero covalent bonds:


```python
msm.get(molsys, n_bonds=True)
```

Now we add explicit covalent bonds between specified atom index pairs:


```python
molsys.topology.add_bonds([[0, 1], [0, 2], [1, 4]])
```

Let's check the total number of bonds and inspect the newly defined bonded atom pairs:


```python
msm.get(molsys, n_bonds=True)
```

```python
bonded_atom_pairs = msm.get(molsys, bonded_atom_pairs=True)
```

```python
bonded_atom_pairs
```

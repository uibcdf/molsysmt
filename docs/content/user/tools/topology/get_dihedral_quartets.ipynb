# Get dihedral quartets

*Getting the quartets of atoms defining specific dihedral angles in a molecular system.*

The function {func}`molsysmt.topology.get_dihedral_quartets` returns atom quartets defining backbone ($\phi$, $\psi$, $\omega$) and sidechain ($\chi_1$ to $\chi_5$) dihedral angles without requiring manual atom name specification.

:::{versionadded} 1.0.0
:::

:::{admonition} API documentation
:class: dropdown

Follow this link for a detailed description of the input arguments, raised errors, and returned objects of this function: {func}`molsysmt.topology.get_dihedral_quartets`.
:::


## Basic usage

Let's show how to extract backbone and sidechain dihedral quartets using TcTIM (PDB ID: 1TCD):


```python
import molsysmt as msm
```

```python
molsys = msm.convert(msm.systems['TcTIM']['1tcd.h5msm'], to_form='molsysmt.MolSys')
```

:::{note} Demo Systems Catalog
:class: dropdown

This tutorial uses demonstration datasets provided by MolSysMT. To explore the full catalog of bundled systems, forms, and file paths, visit the {ref}`Demo Systems <user-foundations-entrance-demo-systems>` guide.
:::


We extract quartets for all $\phi$ backbone angles across the system using {func}`molsysmt.topology.get_dihedral_quartets`:


```python
phi_chains = msm.topology.get_dihedral_quartets(molsys, phi=True)
```

```python
phi_chains
```

We can restrict the search to specific residue selections (e.g. residues 10 to 15):


```python
psi_chains = msm.topology.get_dihedral_quartets(molsys, selection='10<=group_index<=15', psi=True)
```

```python
psi_chains
```

Sidechain dihedral angles (such as $\chi_5$) can also be queried:


```python
chi5_chains = msm.topology.get_dihedral_quartets(molsys, chi5=True)
```

```python
print(len(chi5_chains))
```

```python
n_args = msm.get(molsys, element='group', selection='group_name=="ARG"', n_groups=True)
print(n_args)
```

Multiple dihedral angles can be queried simultaneously:


```python
phi, psi = msm.topology.get_dihedral_quartets(molsys, phi=True, psi=True)
```

```python
phi
```

```python
psi
```

```python
msm.get(molsys, element='atom', selection=phi[0], name=True)
```

## Summary of dihedral angle definitions

The tables below summarize the standard atom quartet definitions and ranges for amino acid dihedral angles:

### Backbone angles

#### $\phi$ (`phi`)

| Residue | Atoms | Zero value | Range (degrees)|
| :---: | :---: | :---: | :---: |
| all but PRO | C-N-CA-C | C cis to C | [-180, 180) |
| PRO | C-N-CA-C | C cis to C | ~-90 |

#### $\psi$ (`psi`)

| Residue | Atoms | Zero value | Range (degrees)|
| :---: | :---: | :---: | :---: |
| all | N-CA-C-N | N cis to N | [-180, 180) |

#### $\omega$ (`omega`)

| Residue | Atoms | Zero value | Range (degrees)|
| :---: | :---: | :---: | :---: |
| all | CA-C-N-CA | CA cis to CA | ~180 |
| all | CH3-C-N-CA | CA cis to CA | ~180 |
| all | CA-C-N-CH3 | CA cis to CA | ~180 |

## Extracting atom blocks attached to dihedral quartets

Using `with_blocks=True`, {func}`molsysmt.topology.get_dihedral_quartets` returns the two atom blocks that rotate around each dihedral central bond:


```python
molsys = msm.convert(msm.systems['Met-enkephalin']['met_enkephalin.h5msm'], to_form='molsysmt.MolSys')
```

```python
phi_chains, phi_blocks = msm.topology.get_dihedral_quartets(molsys, with_blocks=True, phi=True)
```

```python
phi_chains[2]
```

```python
phi_blocks[2]
```

```python
selection_quartet = msm.select(molsys, selection=phi_chains[2])
selection_block_0 = msm.select(molsys, selection=list(phi_blocks[2][0]))
selection_block_1 = msm.select(molsys, selection=list(phi_blocks[2][1]))
```

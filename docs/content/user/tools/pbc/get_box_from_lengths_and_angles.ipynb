# Get box from lengths and angles

*Building 3x3 periodic box matrices from lengths and angles.*

The function {func}`molsysmt.pbc.get_box_from_lengths_and_angles` constructs a standard $3\times 3$ box matrix (or array of matrices) from edge lengths $(a, b, c)$ and crystallographic angles $(\alpha, \beta, \gamma)$.

:::{versionadded} 1.0.0
:::

:::{admonition} API documentation
:class: dropdown

Follow this link for a detailed description of the input arguments, raised errors, and returned objects of this function: {func}`molsysmt.pbc.get_box_from_lengths_and_angles`.
:::


## Basic usage

Let's construct an orthorhombic periodic box matrix from specified dimensions and 90-degree angles:


```python
import molsysmt as msm
import pyunitwizard as puw
```

```python
lengths = puw.quantity([[4.0, 5.0, 6.0]], 'nm')
angles = puw.quantity([[90.0, 90.0, 90.0]], 'degrees')
box = msm.pbc.get_box_from_lengths_and_angles(lengths, angles)
print('Constructed box matrix:\n', box)
```

## Constructing triclinic box matrices

Non-orthogonal angles generate corresponding triclinic/skewed cell matrices:


```python
lengths_tri = puw.quantity([[3.0, 3.0, 3.0]], 'nm')
angles_tri = puw.quantity([[60.0, 60.0, 90.0]], 'degrees')
box_tri = msm.pbc.get_box_from_lengths_and_angles(lengths_tri, angles_tri)
print('Triclinic box matrix:\n', box_tri)
```

## Using orthogonal defaults and shared angles

Omitting `box_angles`, or passing `None`, uses 90-degree angles. One input triple
returns shape `(1, 3, 3)`; each length row adds an output structure. A single
angle triple is shared across length rows, while multiple angle rows must match
their count. Explicit angular units are converted to radians without requiring
an application angle standard. The output uses your active length standard.
The public matrix is rounded to six decimal places in nanometers.


```python
import numpy as np

box_default = msm.pbc.get_box_from_lengths_and_angles(puw.quantity([2, 3, 4], 'nm'))
assert box_default.shape == (1, 3, 3)
np.testing.assert_allclose(puw.get_value(box_default, to_unit='nm'), [np.diag([2, 3, 4])])
box_series = msm.pbc.get_box_from_lengths_and_angles(
    puw.quantity([[2, 3, 4], [5, 6, 7]], 'nm'), box_angles=None)
assert box_series.shape == (2, 3, 3)
np.testing.assert_allclose(puw.get_value(box_series, to_unit='nm'),
    [np.diag([2, 3, 4]), np.diag([5, 6, 7])])
```

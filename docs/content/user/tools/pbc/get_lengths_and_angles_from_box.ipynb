# Get lengths and angles from box

*Extracting edge lengths and crystallographic angles from box matrices.*

The function {func}`molsysmt.pbc.get_lengths_and_angles_from_box` returns both the edge lengths $(a, b, c)$ and the crystallographic angles $(\alpha, \beta, \gamma)$ from a periodic box matrix.

:::{versionadded} 1.0.0
:::

:::{admonition} API documentation
:class: dropdown

Follow this link for a detailed description of the input arguments, raised errors, and returned objects of this function: {func}`molsysmt.pbc.get_lengths_and_angles_from_box`.
:::


## Basic usage

Let's extract both lengths and angles from a truncated octahedral box:


```python
import molsysmt as msm
import pyunitwizard as puw
```

```python
box = msm.pbc.get_box_with_shape(shape='truncated octahedral', length='4.0 nm')
lengths, angles = msm.pbc.get_lengths_and_angles_from_box(box)

print('Box lengths:', lengths)
print('Box angles (degrees):', puw.convert(angles, to_unit='degrees'))
```

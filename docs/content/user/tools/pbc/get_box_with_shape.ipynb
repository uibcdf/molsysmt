# Get box with shape

*Constructing periodic box matrices for standard space-filling geometries.*

The function {func}`molsysmt.pbc.get_box_with_shape` builds canonical $3\times 3$ box matrices for standard unit cell shapes including cubic, truncated octahedral, and rhombic dodecahedral.

:::{versionadded} 1.0.0
:::

:::{admonition} API documentation
:class: dropdown

Follow this link for a detailed description of the input arguments, raised errors, and returned objects of this function: {func}`molsysmt.pbc.get_box_with_shape`.
:::


## Basic usage

Let's construct standard cubic, truncated octahedral, and rhombic dodecahedral periodic boxes:


```python
import molsysmt as msm
import pyunitwizard as puw
```

```python
box_cubic = msm.pbc.get_box_with_shape(shape='cubic', length='2.0 nm')
box_octahedral = msm.pbc.get_box_with_shape(shape='truncated octahedral', length='2.0 nm')
box_dodecahedral = msm.pbc.get_box_with_shape(shape='rhombic dodecahedral', length='2.0 nm')

print('Cubic box:\n', box_cubic)
print('Truncated octahedral box:\n', box_octahedral)
print('Rhombic dodecahedral box:\n', box_dodecahedral)
```

## Generating multi-structure box arrays

Using the `n_structures` argument, we can generate homogeneous box arrays with shape `(n_structures, 3, 3)`:


```python
box_traj = msm.pbc.get_box_with_shape(shape='cubic', length='3.0 nm', n_structures=10)
print('Multi-structure box shape:', box_traj.shape)
```

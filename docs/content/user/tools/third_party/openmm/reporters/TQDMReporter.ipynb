# TQDMReporter

*Displaying dynamic progress bars during OpenMM simulations.*

The `TQDMReporter` class integrates `tqdm` progress bars with simulation time estimations and completion progress during OpenMM dynamics.

:::{versionadded} 1.0.0
:::

:::{admonition} API documentation
:class: dropdown

Follow this link for a detailed description of the input arguments, raised errors, and returned objects: {class}`molsysmt.third_party.openmm.reporters.TQDMReporter`.
:::


## Basic usage

Let's attach a dynamic TQDM progress bar to an OpenMM simulation:

```python
import molsysmt as msm

molsys = msm.convert(msm.systems['Trp-Cage']['1l2y.h5msm'], to_form='molsysmt.MolSys')
simulation = msm.convert(molsys, to_form='openmm.Simulation', temperature='300 K')

reporter = msm.third_party.openmm.reporters.TQDMReporter(reportInterval=10, total_n_steps=20)
simulation.reporters.append(reporter)
simulation.step(20)
print('Simulation step completed.')
```

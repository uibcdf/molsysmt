# Add constant force

*Applying constant pulling or pushing force vectors to target atoms in OpenMM simulations.*

The function {func}`molsysmt.third_party.openmm.forces.add_constant_force` introduces a constant external force vector acting on selected atoms during molecular dynamics.

:::{versionadded} 1.0.0
:::

:::{admonition} API documentation
:class: dropdown

Follow this link for a detailed description of the input arguments, raised errors, and returned objects: {func}`molsysmt.third_party.openmm.forces.add_constant_force`.
:::


## Basic usage

Let's apply a constant external force vector to selected atoms:

```python
import molsysmt as msm

molsys = msm.convert(msm.systems['Trp-Cage']['1l2y.h5msm'], to_form='molsysmt.MolSys')
force = msm.third_party.openmm.forces.add_constant_force(molsys, selection=[0], force='[10,0,0] kJ/(mol*nm)', return_force=True)
print('Constant pulling force configured:', type(force))
```

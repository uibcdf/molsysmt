# Toolbox overview

MolSysMT groups public operations by purpose. The `basic` namespace handles forms, selections, attributes, and composition. The `structure` namespace handles measurements and coordinate transformations. The `topology`, `build`, `pbc`, and `molecular_mechanics` namespaces cover connectivity, system preparation, periodic boundaries, and mechanics.

Structural transformations follow explicit geometric contracts. Coordinates use nm with shape `(n_structures, n_atoms, 3)`. The shared `atom_index` and `n_atoms` attributes connect that atom axis to the stable topological inventory, so tools can work with topology-only, structures-only, or combined forms without changing the meaning of an atom position. Rotation matrices are proper, right-handed 3 × 3 matrices. Principal axes are returned as rows together with their ordered moments. Least-RMSD fitting requires at least three non-collinear fit atoms. These constraints prevent ambiguous transformations from silently producing plausible-looking coordinates.

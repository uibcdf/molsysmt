# `molsysmt.MolSys`

`MolSys` is MolSysMT's materialized native molecular-system form. It combines topology and per-structure data while keeping coordinates in nm, time in ps, and element IDs as strings.

Use `MolSysBuilder` when constructing a system explicitly: declare topology first, assign atom-dependent arrays, and call `build()`. Atom subsets are materialized in canonical increasing source-index order so topology and coordinates remain aligned.

Before materialization, `msm.get(builder, ...)` can read every attribute declared by the builder's stored chemical/topological and structural components. The builder does not store molecular-mechanics attributes or per-structure chemical-state associations.

```python
import numpy as np
import molsysmt as msm
from molsysmt import pyunitwizard as puw

builder = msm.MolSysBuilder()
carbon = builder.add_atom(atom_id=10, atom_name='C', atom_type='C')
oxygen = builder.add_atom(atom_id=11, atom_name='O', atom_type='O')
builder.add_group([carbon, oxygen], group_name='CO')
builder.add_bond(carbon, oxygen, bond_order=2, bond_type='covalent')
builder.set_coordinates(puw.quantity(np.array([[0.0, 0.0, 0.0], [0.123, 0.0, 0.0]]), 'nm'))
molsys = builder.build()
msm.info(molsys)
```

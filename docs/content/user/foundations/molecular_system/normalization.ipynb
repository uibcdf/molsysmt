# Topological Normalization

Molecular systems are chaotic. Depending on the force field (AMBER, CHARMM, GROMOS) or the source (PDB, BinaryCIF, PubChem), the same atom might have different names: `H1`, `HN1`, `H`, or `HA`. This inconsistency is a common cause of failure in automated analysis.

MolSysMT implements a **Topological Normalization Engine** that ensures your analysis works, regardless of the nomenclature of your input files.

### 1. The "Pacification" Process
When a system is loaded or converted, MolSysMT performs a quiet but essential task: it maps the input names to a canonical standard. This allows you to select `atom_name=='H'` and find all backbone hydrogens, even if the file calls them `HN`.

### 2. Standard Residue Naming
Following our "Group vs Residue" policy, MolSysMT standardizes residue names (groups) to their IUPAC/PDB 3-letter codes. It also handles the translation of protonation states (e.g., `HIE`, `HID`, `HIP` $ightarrow$ `HIS`).

### 3. Benefits for the Researcher
- **Interoperability:** Mix AMBER and CHARMM structures in the same script without manual renaming.
- **Script Reusability:** Your analysis functions will work on any PDB, because the selection language speaks to the normalized system, not the raw file.

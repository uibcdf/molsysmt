# Items and Forms

In MolSysMT, a clear distinction is made between a physical **item** and its data representation **form**:

- **Item**: A concrete data container holding molecular information. An item can be a file path on disk (such as `'181l.bcif.gz'`), an in-memory Python object (such as an `mdtraj.Trajectory` instance), a web database identifier (such as `'181L'`), or a text string (such as an amino acid sequence).
- **Form**: The standardized type tag or schema label that identifies how an item's data is structured (such as `'file:bcif'`, `'molsysmt.MolSys'`, `'mdtraj.Trajectory'`, `'string:pdb_id'`).

Functions in MolSysMT inspect items dynamically to determine their form. You can query the form of any item using {func}`molsysmt.basic.get_form`:

```python
import molsysmt as msm

item = msm.systems['T4 lysozyme L99A']['181l.bcif.gz']
msm.get_form(item)
```

## Form Classifications

MolSysMT categorizes supported forms into primary representation types: **files**, **classes**, and **strings** (which include database IDs, text formats, and sequence strings). The full list of supported forms for each category can be queried dynamically using {func}`molsysmt.supported.forms`.

## Files

File forms represent molecular structures or trajectories stored in disk files.

```python
msm.supported.forms(form_type='file')
```

## Classes

Class forms represent in-memory Python objects from third-party libraries (such as MDTraj, PyTraj, MDAnalysis, OpenMM, and ParmEd) as well as native MolSysMT classes.

```python
msm.supported.forms(form_type='class')
```

## Strings and Sequences

String forms represent character strings including database IDs (such as `'pdb:id'`, `'uniprot:id'`), raw text records (such as `'string:pdb_text'`), and sequence strings (such as `'string:amino_acids_1'`).

```python
msm.supported.forms(form_type='string')
```

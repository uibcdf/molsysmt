# Performance and Scalability

Molecular trajectories can be massive (Gigabytes or Terabytes). MolSysMT is built from the ground up to handle Big Data without saturating your computer's resources.

### 1. Lazy Loading (Smarter, not harder)
Unlike traditional libraries that load the whole file into RAM, MolSysMT uses **Lazy Loading**. When you open a trajectory, we only read the metadata. The coordinates are only fetched from the disk when you specifically ask for them (e.g., during a calculation).

### 2. Native CPU Parallelization
Selected analysis functions such as `get_distances`, `get_rmsd`, and `get_sasa` distribute independent work with Rayon. MolSysMT preserves sequential inner numerical loops so the Rust compiler can also vectorize them.

### 3. Chunked Execution
For systems larger than RAM, MolSysMT implements a **Chunked Engine**. It processes the trajectory in "bites" (chunks), performs the calculation, and releases the memory before moving to the next bite. This ensures that you can analyze a 1TB simulation on a laptop.

```{admonition} How to enable?
:class: info
Set the session default with `msm.configure.set_parallelization(parallel='auto', num_threads=8)`. A supported function can override it for one call, for example `msm.structure.get_distances(system, parallel=True, num_threads=2)`. Use `parallel=False` to force a one-thread execution. `None` inherits the session, and `num_threads=-1` uses all processors available to the process.
```

--- 

```{key-takeaway}
Memory is finite, but science shouldn't be. MolSysMT's scalability ensures that your computer stays fast even with the largest simulations.
```

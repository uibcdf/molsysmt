# `molsysmt.MolSys`

`molsysmt.MolSys` is MolSysMT's primary native, in-memory molecular system class. It serves as a unified umbrella container that brings together three decoupled sub-objects:

- **`topology`** (`molsysmt.Topology`): Manages atom inventories, group topologies, component and chain assignments, molecule/entity hierarchies, and chemical state attributes.
- **`structures`** (`molsysmt.Structures`): Stores 3D atomic coordinates, periodic boundary box vectors, time series, and trajectory frame metadata.
- **`molecular_mechanics`** (`molsysmt.MolecularMechanics`): Holds force field definitions, partial charges, atomic masses, non-bonded interaction parameters, and energy terms.

Any molecular system represented in any supported form (such as PDB files, mmCIF files, trajectory files, MDAnalysis universes, OpenMM topologies, or ParmEd structures) can be converted into an in-memory `molsysmt.MolSys` instance using `msm.convert()`.

## Converting to `molsysmt.MolSys`

Converting a molecular system into `molsysmt.MolSys` materializes all available topological, structural, and mechanics data into MolSysMT's native internal formats:

```python
import molsysmt as msm

molsys = msm.convert(msm.systems['T4 lysozyme L99A']['181l.bcif.gz'], to_form='molsysmt.MolSys')
msm.info(molsys)
```

## Inspecting Sub-Objects

The sub-objects composing a `molsysmt.MolSys` instance can be accessed directly via attributes:

```python
print("Topology sub-object:", type(molsys.topology))
print("Structures sub-object:", type(molsys.structures))
print("MolecularMechanics sub-object:", type(molsys.molecular_mechanics))
```

## Querying Attributes

You can query attributes from a `molsysmt.MolSys` object seamlessly using standard MolSysMT functions like `msm.get()` or `msm.select()`:

```python
n_atoms, n_groups, n_structures = msm.get(molsys, n_atoms=True, n_groups=True, n_structures=True)
print(f"Atoms: {n_atoms}, Groups: {n_groups}, Structures: {n_structures}")
```

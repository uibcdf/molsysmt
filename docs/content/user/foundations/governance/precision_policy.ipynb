# Precision Policy

Numerical noise is the silent enemy of Computational Structural Biology. In MolSysMT, we handle structural data from many sources, each with its own precision standards. This document defines our policy for handling floating-point precision.

### 1. The Single vs Double Precision Trade-off
MolSysMT recognizes two main operational modes:
1.  **Single Precision (`float32`):** Used by default in some trajectory formats (like XTC or DCD) to save disk space and accelerate visualization. While efficient, it can introduce noise in the 7th or 8th decimal place.
2.  **Double Precision (`float64`):** Our native standard for internal calculations and native objects (`molsysmt.MolSys`). This ensures that your analysis is as mathematically rigorous as possible.

### 2. Handling Numerical Noise
When setting coordinates via strings (e.g., `'1.0 angstroms'`), MolSysMT internally ensures that the conversion to Nanometers maintains `float64` precision unless the target form specifically requires `float32`.

### 3. Precision Agnosticism
MolSysMT is designed to be **Precision Agnostic** in its API. Functions like `get_rmsd()` or `get_distances()` will automatically detect the precision of the input arrays and perform the calculations accordingly, promoting the result to `float64` for maximum safety.

```{admonition} Rule of Thumb
:class: warning
If you are performing high-precision geometric analysis, we recommend converting your system to a native **`molsysmt.MolSys`** object first to ensure all data is stored in double precision.
```

--- 

```{key-takeaway}
Honesty in precision is honesty in science. MolSysMT gives you the tools to know exactly how precise your data is.
```

# Demo Systems Catalog

MolSysMT includes a built-in collection of pre-packaged molecular systems accessible via `molsysmt.systems`. These datasets are bundled directly with the library to enable offline testing, hands-on tutorials, benchmarks, and reproducible recipes without relying on external network downloads.

This page serves as the authoritative reference for all bundled demo systems referenced throughout the **User Guide**, **Cookbook**, **Showcase**, and **Master Course**.

---

## How to Access Demo Systems

All demo systems are stored in a nested dictionary structure inside `molsysmt.systems`, where the top-level keys represent system names and the inner keys represent available file names.

### 1. Listing Available Systems
You can list all bundled system names in Python:

```python
import molsysmt as msm
print(list(msm.systems.keys()))
```

### 2. Accessing a System File Path
To retrieve the absolute `Path` of a specific file:

```python
file_path = msm.systems['Trp-Cage']['1l2y.h5msm']
```

### 3. Loading a System Directly into MolSysMT
You can pass the dictionary item directly into `msm.convert()` or any other tool function:

```python
mol = msm.convert(msm.systems['Trp-Cage']['1l2y.h5msm'])
```

````python
import molsysmt as msm

# Load a demo system and inspect basic attributes
mol = msm.convert(msm.systems['Trp-Cage']['1l2y.h5msm'])
n_atoms, n_groups, n_structures = msm.get(mol, n_atoms=True, n_groups=True, n_structures=True)

print(f"Trp-Cage loaded: {n_atoms} atoms, {n_groups} residues, {n_structures} structure(s).")
````

## Complete Catalog of Bundled Systems

The table below is generated dynamically from `molsysmt.systems`, ensuring an up-to-date catalog of all available demo and test datasets:

````python
import pandas as pd
import molsysmt as msm

rows = []
for system_name, files in sorted(msm.systems.items()):
    file_list = ", ".join(f"<code>{f}</code>" for f in sorted(files.keys()))
    sample_file = sorted(files.keys())[0]
    example_code = f"<code>msm.systems['{system_name}']['{sample_file}']</code>"
    rows.append({
        "System Name": f"<b>{system_name}</b>",
        "Files Included": file_list,
        "Invocation Example": example_code
    })

df = pd.DataFrame(rows)
from IPython.display import HTML, display
display(HTML(df.to_html(escape=False, index=False)))
````

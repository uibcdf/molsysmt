# What is MolSysMT?

**MolSysMT** (*Molecular Systems Multi-Toolkit*) is a Python library designed to build, prepare, query, transform, analyze, and visualize molecular systems through **one uniform API**.

In computational structural biology, taking a system from raw coordinates to simulation and analysis typically requires four or five different software libraries with incompatible data models. Research groups spend up to 70% of their time writing brittle "glue code", translating file formats, repairing missing atoms, and resolving inconsistent physical units. **MolSysMT was built to eliminate this technical friction once and for all.**

---

## The Form-Agnostic Paradigm

At the core of MolSysMT is **Form Agnosticism**. MolSysMT does not care how your molecular data is stored — whether it is a PDB file on disk, an OpenMM `Topology` in memory, an MDAnalysis `Universe`, or an NGLView structure.

Instead of learning separate accessors for each library, MolSysMT provides a single set of core operations (`get`, `set`, `select`, `convert`, `build`, `view`, `compare`) that work identically across **89 supported forms** (with 75 Tier-1 stable forms).

```python
import molsysmt as msm

# Load a molecular system and inspect topology attributes in a single call
mol = msm.convert(msm.systems['Trp-Cage']['1l2y.h5msm'])
n_atoms, n_groups, n_chains = msm.get(mol, n_atoms=True, n_groups=True, n_chains=True)

print(f"Atoms: {n_atoms}, Residues: {n_groups}, Chains: {n_chains}")
```

---

## Core Architectural Pillars

### 1. Universal Interoperability & Zero-Loss Bridges
MolSysMT connects files, web databases, and third-party object models without forcing you to abandon your preferred software stack. Conversions carry explicit fidelity reporting:

```python
# Convert to a sequence string and an MDTraj Trajectory seamlessly
seq = msm.convert(mol, to_form='string:amino_acids_1')
traj = msm.convert(mol, to_form='mdtraj.Trajectory')

print(f"Sequence: {seq}")
print(f"MDTraj Trajectory: {traj}")
```

### 2. Native Structure Preparation Pipeline
Structure preparation (diagnosing missing heavy atoms, adding terminal cappings, protonating at specific pH, solvating, and adding counterions) is built directly into MolSysMT. You can prepare systems natively without needing external tools like PDBFixer or OpenMM:

```python
# Diagnose missing atoms or cappings natively
missing_heavy = msm.build.get_missing_heavy_atoms(mol)
print(f"Missing heavy atoms: {missing_heavy}")
```

### 3. Native High-Performance Compute in Rust
Analytical functions — including pairwise distances, neighbor lists, RMSD, superposition, radius of gyration, RMSF, principal component axes, SASA, and dihedral angles — run on precompiled Rust compute kernels. They execute with **zero JIT compilation warmup**, zero startup delay, and configurable parallel execution.

### 4. Physical Safety & Unit Consistency
All physical quantities in MolSysMT carry explicit physical units powered by `pyunitwizard` (nanometers, picoseconds, radians, elementary charges, Kelvin). Unit conversions, shape validations, and dimensional consistency are enforced automatically.

---

## The Engine of MolSysSuite

MolSysMT serves as the foundational "Operating System" for molecular data within the [**MolSysSuite**](https://www.uibcdf.org/) framework developed at the UIB-CDF. Higher-level specialized toolkits — such as **TopoMet** for topological analysis, **ElastNetMT** for elastic network models, and **MolSysViewer** for 3D web visualization — rely on MolSysMT as their underlying data engine.

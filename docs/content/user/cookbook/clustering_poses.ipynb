# Comparing and clustering saved poses

*Keeping original pose keys while composing general geometric tools.*

First declare one fixed ligand state and a common rigid receptor Cartesian frame.
Prepare whole periodic molecules explicitly when needed. Different states remain
separate populations. For independent saved systems, use public correspondence
and atom permutation tools to establish an explicit shared axis, retain each
source map and then compose compatible structures. Equal counts/IDs or successful
concatenation do not establish chemistry, alignment or source identity.

This analytic example already has a common axis. It compares original positions
without fitting, then groups the resulting length matrix. No receptor coordinates
are moved or scores interpreted. The initial DockingMT receiving budget of 32 poses
is a consumer budget, not a hard universal MolSysMT limit.


```python
import numpy as np
import molsysmt as msm
from molsysmt.native import Structures
puw = msm.pyunitwizard
coordinates = np.zeros((3, 2, 3))
coordinates[:, :, 0] = np.array([0., 1., 2.])[:, None]
molsys = Structures(coordinates=puw.quantity(coordinates, 'nm'))
poses = msm.structure.get_pairwise_rmsd(
    molsys, selection='all', member_keys=['saved/a', 'saved/b', 'saved/c'])
clusters = msm.structure.get_distance_clusters(
    poses['rmsd'], '1 nm', member_keys=poses['member_keys'])
for start, stop in zip(clusters['cluster_offsets'][:-1], clusters['cluster_offsets'][1:]):
    members = clusters['member_indices'][start:stop]
    print([poses['member_keys'][index] for index in members])
assert clusters['cluster_indices'].tolist() == [0, 0, 1]
```

## Reusing the chemical profile

For a chemically equivalent population, build maps with
`chemical_states.get_atom_correspondences` or let `get_pairwise_rmsd` construct them
with `method='symmetry_corrected'`. Retain fixed-state evidence, original atom maps,
`search_scope`, method, versions and detached references with your report.
Full chemical context is required even for a heavy-atom RMSD population. Different
indexed-H representations need explicit preparation; no chemistry is inferred.
Directional/nonreflexive candidate sets cannot define the symmetric matrix.

The shared table and `pair_indices` retain explicitly oriented winners. Member keys
remain opaque caller labels. Neither keys nor hashes authenticate which molecule
or receptor axes you supplied. Result dictionaries are detached analysis outputs;
this feature adds no H5MSM persistence schema for RMSD/clustering reports.

## Respecting resources and scientific meaning

Pairwise output is dense and selected coordinates remain resident. Set `max_work`
and session RAM limits before extending a population; input streaming does not
make the result out-of-core. Nonfinite coordinates/distances, asymmetric matrices,
unsupported chemistry and invalid bounds raise, rather than being repaired.

The matrix and cutoff convert directly to float64 nm for the inclusive comparison.
Changing your output-unit policy changes the returned quantities, not cluster
membership for these supplied inputs. Earlier conversions can have rounded the
inputs; no implicit tolerance repairs them.

Complete linkage limits cluster diameter; single linkage may form chains. Returned
representatives are geometric medoids or earliest members. Score comparability and
score-selected representatives belong to the consuming workflow. A cluster count
or size is not an affinity, probability, free energy or validation of a pose.

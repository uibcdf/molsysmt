# Trajectory Performance Analysis

**Task:** Calculate the RMSD of a multi-frame trajectory against a reference structure and plot the results in a single workflow.

MolSysMT is optimized for speed. This recipe demonstrates how to process thousands of frames and visualize structural stability effortlessly.

````python
import molsysmt as msm
from molsysmt import systems
import matplotlib.pyplot as plt

# Load a solvated trajectory (Villin Headpiece)
topology = systems['chicken villin HP35']['chicken_villin_HP35_solvated.h5msm']
trajectory = systems['chicken villin HP35']['traj_chicken_villin_HP35_solvated.dcd']
````

### 1. Fast RMSD Calculation
We calculate the Root Mean Square Deviation for the backbone atoms over the entire trajectory.

````python
times, rmsd = msm.structure.get_rmsd(trajectory, selection='backbone', 
                                     reference_molecular_system=topology, time=True)

print(f"Processed {len(rmsd)} frames.")
````

### 2. Plotting Stability
Using standard plotting tools with MolSysMT data is straightforward.

````python
plt.plot(times, rmsd)
plt.xlabel('Time (ps)')
plt.ylabel('RMSD (nm)')
plt.title('Structural Stability of Villin HP35')
plt.show()
````

```{key-takeaway}
MolSysMT bridges the gap between heavy simulation data and scientific visualization, providing high-performance analytical kernels like `get_rmsd`.
```

# Path C - Module 37: Hydrogen Bonds & Salt Bridges

The secret of inhibitor affinity is in the **Hydrogen Bonds**. The N3 inhibitor mimics a peptide substrate to form multiple H-bonds with the protease backbone. 

In this module, you will learn to identify the exact residues that anchor the drug to the virus. For trajectories, the default Buch result keeps rectangular arrays when counts match and aligned lists with empty entries when they vary. The optional `output_type="molsysmt.Interactions"` returns an experimental sparse analysis with H-A distances in nm, evaluated coverage, and atom/structure queries. It supports automatic roles in one selection or two disjoint participant universes and uses eager execution. See [Querying interaction results](../../user/tools/interactions/result.md) for explicit attachment to `MolSys`. Attaching an analysis declares that its local atom and structure indices match the target system. You are responsible for that correspondence when loading an independent file; source labels and equal axis sizes do not authenticate molecular origin. Each analysis also retains producer versions in `software`; saving or loading with another version does not replace them. An empty dictionary means unknown producer versions.

```python
import molsysmt as msm

molsys = msm.convert('pdb:6LU7', to_form='molsysmt.MolSys')
msm.build.add_missing_hydrogens(molsys)

# Find H-bonds between drug and protein
hbonds, _ = msm.interactions.hbonds.get_buch_hbonds(molsys, selection='molecule_type=="small molecule"', 
                               selection_2='molecule_type=="protein"')

print(f"The drug is anchored by {len(hbonds[0])} hydrogen bonds.")
```

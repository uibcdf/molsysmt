# Path B - Module 36: PBC Geometry & Conventions

To simulate your industrial enzyme, you need to define the "World" it lives in. In molecular dynamics, this is the **Simulation Box**. Managing its geometry and Periodic Boundary Conditions (PBC) is essential to ensure that your calculations represent a continuous bulk liquid.

In this module, you will learn to master the simulation box of your engineered PETase.

```python
import molsysmt as msm
from molsysmt import systems

# Load your solvated system (Enzyme + Ions + Water)
petase = msm.convert('pdb:6EQE', to_form='molsysmt.MolSys', selection='molecule_type=="protein"')
solvated_system = msm.build.solvate(petase, box_shape='cubic', clearance='1.0 nm')
```

### 1. Verifying the World's Boundaries
Does your industrial model have a defined box? Without a box, you cannot apply PBC.

```python
has_box = msm.pbc.has_pbc(solvated_system)
print(f"Is the Periodic Boundary Condition active? {has_box}")

# Get the box matrix
box = msm.get(solvated_system, element='system', box=True)
print(f"Box Matrix (nm):\n{box}")
```

### 2. Box Shape and Volume
Efficient simulations in industrial biocatalysis often use non-cubic boxes to save computation time. Let's identify the current shape and its volume.

```python
shape = msm.pbc.get_shape_from_box(box)
volume = msm.pbc.get_volume_from_box(box)

print(f"Box Geometry: {shape}")
print(f"Reactor Volume: {volume}")
```

### 3. Lengths and Angles
Engineers often describe boxes using side lengths (a, b, c) and angles (alpha, beta, gamma). MolSysMT provides easy access to these parameters.

```python
lengths, angles = msm.pbc.get_lengths_and_angles_from_box(box)

print(f"Lengths: {lengths}")
print(f"Angles: {angles}")
```

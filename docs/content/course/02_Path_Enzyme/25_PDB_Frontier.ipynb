# Path B - Module 25: PDB Bioassemblies & AltLocs

Many high-resolution industrial enzymes are solved using X-ray crystallography. These files often contain **Alternate Locations (AltLocs)**, which are ghost-like representations of atoms that exist in more than one position in the crystal.

In this module, you will learn to clean these ambiguities to have a physically consistent starting point for your PETase simulation.

```python
import molsysmt as msm
from molsysmt import systems

# Load a PETase crystal structure from the cloud
pdb_id = 'pdb:5XJH'
molsys = msm.convert(pdb_id, to_form='molsysmt.MolSys')
```

### 1. Identifying the Disorder
Let's see if our system has atoms with multiple locations. These are problematic for most force fields.

```python
altlocs = msm.get(molsys, element='atom', alternate_location=True)
# Filter only those that are not 'None' or empty
real_altlocs = [a for a in altlocs if a and a != ' ']

print(f"Found {len(real_altlocs)} atoms with multiple positions in the crystal.")
```

### 2. Solving AltLocs
The function `solve_atoms_with_alternate_location()` chooses the best position for each atom (usually the one with the highest occupancy or label 'A').

```python
print(f"Atoms before cleaning: {msm.get(molsys, element='system', n_atoms=True)}")

msm.build.solve_atoms_with_alternate_location(molsys)

print(f"Atoms after cleaning: {msm.get(molsys, element='system', n_atoms=True)}")
```

### 3. Biological Assembly check
Is the PETase active as a monomer or a dimer? MolSysMT can generate the functional unit using symmetry records.

```python
# Make the biological assembly
full_unit = msm.build.make_bioassembly(molsys)

msm.info(full_unit, element='chain')
```

# Path B - Module 43: Trajectory Management (Slicing)

In an industrial pipeline, you might simulate hundreds of different PETase mutants. Storing every single frame of every simulation is inefficient. You need to know how to filter, slice, and sub-sample your data to keep only the scientifically relevant parts.

In this module, you will learn to manage trajectory fragments using MolSysMT.

```python
import molsysmt as msm
from molsysmt import systems

# Load a simulation trajectory of our enzyme
villin_traj = systems['chicken villin HP35']['traj_chicken_villin_HP35_solvated.dcd']
villin_topo = systems['chicken villin HP35']['chicken_villin_HP35_solvated.h5msm']
molsys = [villin_topo, villin_traj]
```

### 1. Temporal Slicing
You can extract specific time windows from your trajectory. For example, let's keep only the last 50 frames, where the system is presumably equilibrated.

```python
n_frames = msm.get(molsys, element='system', n_structures=True)
print(f"Total simulation frames: {n_frames}")

# Extract the equilibrated part (last 50 frames)
equilibrated_traj = msm.extract(molsys, structure_indices=range(n_frames-50, n_frames))

print(f"Equilibrated frames: {msm.get(equilibrated_traj, element='system', n_structures=True)}")
```

### 2. Striding (Data Reduction)
To generate a movie or a quick report, you can take every 10th frame. This reduces the file size by 90% while maintaining the overall narrative of the motion.

```python
# Sub-sample the trajectory taking 1 frame every 10
short_traj = msm.extract(molsys, structure_indices=range(0, n_frames, 10))

print(f"Reduced trajectory frames: {msm.get(short_traj, element='system', n_structures=True)}")
```

### 3. Merging Trajectory Fragments
If your supercomputer job crashed and you had to restart it, you can join the two resulting files into a single virtual object.

```python
# Simulate two parts of a trajectory
part_a = msm.extract(molsys, structure_indices=range(0, 10))
part_b = msm.extract(molsys, structure_indices=range(10, 20))

# Concatenate them
full_run = msm.structure.concatenate_structures([part_a, part_b])
print(f"Total concatenated frames: {msm.get(full_run, element='system', n_structures=True)}")
```

# Path D - Module 24: The MolSysBuilder API

In some biophysical studies, you need to add specific solvent structures. In this module, you will learn to build a "Hydrated Potassium Ion" (an ion surrounded by 8 water molecules in a specific geometry) from scratch.

```python
import molsysmt as msm
molsys = msm.convert(None, to_form='molsysmt.MolSys')
builder = msm.build.editable(molsys)

# Add a central Potassium
builder.add_atom(atom_name='K', atom_type='K', coordinates=[[0,0,0]], entity_name='Solvated Ion')
print("Hydrated ion scaffold started.")
```

# Module 16: Merging and Growing Systems

**Welcome back, Apprentice Master.** In [Module 15: Semantic Labeling](../Common_Core/15_Semantic_Labeling.ipynb), you mastered generating human-readable nomenclature strings using `msm.get_label()`. Now we transition to system composition: **Merging and Growing Systems**.

Complex molecular simulations frequently require assembling systems from separate components: fusing a receptor protein and a co-crystallized ligand, appending solvent ions into a box, or joining sequential trajectory segments along a timeline. MolSysMT provides four dedicated composition functions in `msm.basic` categorized along two operational axes:

| Composition Dimension | Out-of-Place (Returns New Object) | In-Place (Modifies Target) |
| :--- | :--- | :--- |
| **Topology & Elements** (Atoms/Molecules) | **`msm.merge()`** | **`msm.add()`** |
| **Structures & Frames** (Trajectories) | **`msm.concatenate_structures()`** | **`msm.append_structures()`** |


### 1. Merging Independent Systems

Let's begin by importing MolSysMT, loading a protein system (Villin Headpiece) and a small molecule (Benzamidine), and converting them to native `MolSys` objects.

````python
import molsysmt as msm
from molsysmt import systems

# Load protein and ligand systems
protein = msm.convert(systems['chicken villin HP35']['1vii.pdb'], to_form='molsysmt.MolSys')
ligand = msm.convert(systems['benzamidine']['benzamidine.pdb'], to_form='molsysmt.MolSys')
````

The function **`msm.merge()`** takes a list of independent molecular systems and fuses their topology and spatial coordinates into a **new** combined system object, leaving the input systems unchanged:

````python
# Merge protein and ligand out-of-place into a new complex
complex_sys = msm.merge([protein, ligand])

# Inspect topology summary of the merged complex
msm.info(complex_sys)
````

:::{hint}
:class: dropdown
**msm.merge()**: Fuses multiple independent molecular systems into a new target system in memory. See API doc: {func}`molsysmt.basic.merge`.
:::

### 2. Adding Elements In-Place

While `msm.merge()` creates a new combined system from a list, **`msm.add()`** appends selected elements or molecules directly into an existing target system in place:

````python
# Print atom count of target protein before addition
print(f"Protein atom count before addition: {msm.get(protein, element='system', n_atoms=True)}")

# Add ligand elements directly into the protein system in place
msm.add(protein, ligand)

# Print updated atom count
print(f"Protein atom count after in-place addition: {msm.get(protein, element='system', n_atoms=True)}")
````

:::{hint}
:class: dropdown
**msm.add()**: Appends elements or molecules from a source system into an existing target system. See API doc: {func}`molsysmt.basic.add`.
:::

### 3. Concatenating Trajectory Frames

When you have multiple trajectory files representing sequential simulation segments for systems with matching topology, **`msm.concatenate_structures()`** combines their coordinate frames out-of-place into a **new** system along the time/structure axis:

````python
# Build two conformational snapshots of alanine dipeptide
snap1 = msm.build.build_peptide('AceAlaNme')
snap2 = msm.structure.translate(snap1, translation='[0.1, 0.1, 0.1] nanometers')

# Concatenate snapshots out-of-place into a new multi-structure system
multi_traj = msm.concatenate_structures([snap1, snap2])

n_frames = msm.get(multi_traj, element='system', n_structures=True)
print(f"Concatenated trajectory total frame count: {n_frames}")
````

:::{hint}
:class: dropdown
**msm.concatenate_structures()**: Combines structural frames from a list of matching systems into a new system. See API doc: {func}`molsysmt.basic.concatenate_structures`.
:::

### 4. Appending Frames In-Place

To append new coordinate frames directly into an existing trajectory object along the structure axis in place, use **`msm.append_structures()`**:

````python
# Print frame count of snap1 before appending
print(f"Frame count of snap1 before appending: {msm.get(snap1, element='system', n_structures=True)}")

# Append snap2 frames directly into snap1 in place
msm.append_structures(snap1, snap2)

# Print updated frame count
print(f"Frame count of snap1 after in-place appending: {msm.get(snap1, element='system', n_structures=True)}")
````

:::{hint}
:class: dropdown
**msm.append_structures()**: Appends structural frames from a source system onto an existing target system in place. See API doc: {func}`molsysmt.basic.append_structures`.
:::

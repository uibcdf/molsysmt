# Path A - Module 45: Performance Optimization

Speed is a feature. When you are analyzing a microsecond-scale trajectory of an Alzheimer's fibril, every millisecond counts. 

MolSysMT uses **Numba**, a Just-In-Time (JIT) compiler, to run heavy mathematical kernels at the speed of C++. In this module, you will learn to optimize the framework's performance using the **Warmup** mechanism.

```python
import molsysmt as msm
import time
from molsysmt import systems
```

### 1. Understanding the JIT Lag
Numba compiles Python code into machine code the first time a function is called. This means the first run is always slower than the subsequent ones.

```python
lysozyme = systems['T4 lysozyme L99A']['181l.h5msm']

# First call (Compiling + Executing)
start = time.time()
msm.get_distances(lysozyme, selection='atom_index==[0,1,2]', selection_2='atom_index==[10,11,12]')
end = time.time()
print(f"First call time: {end - start:.4f} seconds")

# Second call (Executing only)
start = time.time()
msm.get_distances(lysozyme, selection='atom_index==[0,1,2]', selection_2='atom_index==[10,11,12]')
end = time.time()
print(f"Second call time: {end - start:.4f} seconds")
```

### 2. Pre-compilation with `warmup_numba()`
In a production environment, you don't want your users (or your pipeline) to experience that first-call lag. You can trigger a full warmup of the most critical kernels at the beginning of your script.

```python
print("Starting Numba Warmup...")
# This will compile the most important structural kernels
msm.warmup_numba()
print("Warmup completed. All engines ready for high-speed analysis.")
```

### 3. When to use Warmup?
- Use it at the **top of your notebooks** or scripts.
- Use it in **HPC environments** (supercomputers) before launching massive parallel analyses.
- Don't worry about it if you are just doing simple visualizations.

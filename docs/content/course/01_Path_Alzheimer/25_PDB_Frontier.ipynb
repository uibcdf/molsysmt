# Path A - Module 25: PDB Bioassemblies & AltLocs

To finish your preparation as a Molecular Architect, you must master the two "hidden" features of the PDB format: **Alternate Locations** and **Biological Assemblies**.

If you ignore these, your simulations might crash or you might be studying only a fragment of the real protein.

```python
import molsysmt as msm
from molsysmt import systems
```

### 1. Solving Alternate Locations (AltLocs)
In high-resolution X-ray structures, some atoms can be seen in two or more positions. These are labeled as 'A', 'B', etc. MolSysMT can automatically keep only the most occupied one to avoid "ghost atoms".

```python
# Let's use a different PDB that is known to have AltLocs (e.g., 181L)
molsys_alt = msm.convert('pdb:181L', to_form='molsysmt.MolSys')

print(f"Atoms before solving AltLocs: {msm.get(molsys_alt, element='system', n_atoms=True)}")

# Solve them (it keeps the default 'A' or the highest occupancy)
msm.build.solve_atoms_with_alternate_location(molsys_alt)

print(f"Atoms after solving AltLocs: {msm.get(molsys_alt, element='system', n_atoms=True)}")
```

### 2. Making the Biological Assembly
Sometimes, the PDB file only contains the "Asymmetric Unit" (e.g., one chain), but the functional protein is a tetramer. The file contains symmetry instructions to build the full unit.

```python
# 2BEG is already a full assembly, but let's try to trigger the logic
fibril = msm.convert('pdb:2BEG', to_form='molsysmt.MolSys')
full_fibril = msm.build.make_bioassembly(fibril)

msm.info(full_fibril, element='chain')
```

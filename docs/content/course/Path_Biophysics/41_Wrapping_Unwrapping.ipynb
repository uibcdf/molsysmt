Restoring trajectory continuity. In this module, you will unwrap atom trajectories for diffusion measurements and distinguish this operation from single-frame molecular reconstruction.

leave the true image history ambiguous; nearest-image continuity does not
recover arbitrary motion across several cells. Component wrapping remains the
separate operation for reconstructing provided covalent connectivity within

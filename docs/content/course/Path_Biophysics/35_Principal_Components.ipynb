# Path D - Module 35: Principal Components (PCA) & Axes

Identify the **Gating Modes** of the channel. In this module, you will learn to use PCA to visualize the opening and closing of the pore.

Numerical PCA weights are lists, tuples or arrays in selected atom order. They
are finite, nonnegative and have a positive sum. Weighting uses square roots
without normalization by the weight sum; eigenvalues have squared-length units.
Use Cartesian blocks (all x, then y, then z) when projecting coordinates. See
{ref}`Cookbook_Weighting_PCA` for an independently checked example.

This bounded membrane example measures coordinate dispersion of eight phosphorus sites
across three selected observations. For a biological gating analysis, first
choose the relevant protein ensemble, fit rigid motion and reconstruct wrapped
participants as required by that scientific model.


```python
import molsysmt as msm
from molsysmt import systems
# A lipid membrane has phosphorus sites, not protein alpha carbons.
molsys = [systems['POPC membrane']['popc_membrane.psf'],
          systems['POPC membrane']['popc_membrane.dcd']]
atom_indices = msm.select(molsys, selection='atom_type=="P"')[:8]
eigenvectors, eigenvalues = msm.structure.principal_component_analysis(
    molsys, selection=atom_indices, structure_indices=[2, 0, 2],
    use_gpu=False)
print("PCA modes:", eigenvectors.shape)
print("Leading variance:", eigenvalues[0])
```

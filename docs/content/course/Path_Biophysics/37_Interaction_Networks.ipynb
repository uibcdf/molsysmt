# Path D - Module 37: Hydrogen Bonds & Salt Bridges

Map the H-bond network that coordinates K+ ions in the selectivity filter. In this module, you will learn to identify the exact chemical anchors of the pore. For Buch hydrogen bonds across structures, equal counts give rectangular arrays and varying counts give aligned lists with empty entries. Choose `output_type="molsysmt.Interactions"` for an experimental sparse analysis with H-A distances in nm, evaluated coverage, and participant queries. Automatic roles in one selection or two disjoint participant universes are supported; calculation remains eager. See [Querying interaction results](../../user/tools/interactions/result.md) for attaching named analyses to `MolSys`.

```python
import molsysmt as msm
molsys = msm.convert('pdb:1K4C', to_form='molsysmt.MolSys')
msm.build.add_missing_hydrogens(molsys)
hb, _ = msm.interactions.hbonds.get_buch_hbonds(molsys, selection='group_id==[70:80]')
```

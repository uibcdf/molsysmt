# Module 11: Hierarchical Levels

Molecular systems are not just lists of atoms. They have a biological structure. MolSysMT recognizes 6 main hierarchical levels (Elements):

```{admonition} Glossary: Hierarchy Element
:class: info
An **Element** is a biological unit at a specific scale. MolSysMT uses 6 nested levels: Atom < Group < Component < Molecule < Chain < Entity.
```

--- 

```{admonition} A Note on Terminology: Why "Group" and not "Residue"?
:class: info
You might be used to other softwares calling every collection of atoms a "residue". In MolSysMT, we avoid this for the sake of scientific precision. 

A **residue** is a biological term specifically for the building blocks of proteins or nucleic acids. Calling a single Sodium ion or a Water molecule a "residue" is technically inaccurate. Therefore, we use the universal term **Group** to represent any atomic unit or set, reserving the biological intuition for when it truly applies.
```

````python
import molsysmt as msm
from molsysmt import systems

lysozyme = systems['T4 lysozyme L99A']['181l.bcif.gz']
````

### 1. Changing the Perspective with `info()`
The `element` argument in `info()` acts as a "zoom". You can see the inventory of the whole system or drill down into specific biological levels.

````python
# Visual scale: What are the unique biological identities in this system?
msm.info(lysozyme, element='entity')
````

````python
# Closer look: Let's see the first 5 groups of the system
msm.info(lysozyme, element='group', selection=[0, 1, 2, 3, 4])
````

### 2. Multi-level Queries with `get()`
Just like `info()`, the `get()` function allows you to count and extract data at different scales.

````python
# Scale: System. How many atoms, groups, and entities in total?
n_atoms, n_groups, n_entities = msm.get(lysozyme, element='system', n_atoms=True, n_groups=True, n_entities=True)
print(f"The whole system has {n_atoms} atoms and {n_groups} groups.")

# Scale: Groups. How many atoms are in the first three groups?
n_atoms_per_group = msm.get(lysozyme, element='group', selection=[0, 1, 2], n_atoms=True)
print(f"Number of atoms in the first 3 groups: {n_atoms_per_group}")
````

### 3. Biological Content Audit
Now that you understand the hierarchy, you can audit the system's composition using high-level biological questions.

- `contains()`: Checks if the system contains at least one instance of an element type.
- `is_composed_of()`: Checks if the system matches an exact biological composition.

````python
print(f"Does it contain water? {msm.contains(lysozyme, water=True)}")
print(f"Does it contain lipids? {msm.contains(lysozyme, lipid=True)}")

# Is it composed of protein, 2 small molecules, waters, and ions?
print(f"Is it the full complex? {msm.is_composed_of(lysozyme, protein=True, small_molecule=2, water=True, ion=True)}")
````

### 4. Interactive Anatomy

Don't just trust the index. Use your eyes and your mouse to verify the hierarchy.

````python
view = msm.view(lysozyme)
view
````

**Instruction:** Click on any atom in the viewer above, then execute the cell below.

````python
picked_indices = view.active_selection.atom_indices
if len(picked_indices) > 0:
    entity_name, chain_id = msm.get(lysozyme, selection=picked_indices[0], entity_name=True, chain_id=True)
    print(f"You clicked an atom that belongs to Entity: {entity_name} (Chain {chain_id})")
````

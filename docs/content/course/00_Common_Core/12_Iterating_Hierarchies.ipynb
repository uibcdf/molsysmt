# Module 12: Iterating over Hierarchies with `Iterator()`

Extracting all data at once with `get()` is fast, but sometimes you need to process a system piece by piece. You might want to loop over every residue to perform a custom calculation, or process a massive trajectory frame by frame without loading it all into RAM.

MolSysMT provides a powerful **`Iterator()`** for these cases.

```python
import molsysmt as msm
from molsysmt import systems

lysozyme = systems['T4 lysozyme L99A']['181l.bcif.gz']
```

### 1. Simple Biological Loops
You can iterate over any hierarchical level (`group`, `molecule`, `chain`, etc.). The iterator will yield the requested attributes for each unit.

```python
# Let's iterate over the first 5 groups (residues)
for name, atoms in msm.Iterator(lysozyme, element='group', selection=[0,1,2,3,4], 
                                group_name=True, atom_index=True):
    print(f"Processing group {name}... it has {len(atoms)} atoms.")
```

### 2. The Power of Chunking
For very large systems, you can process multiple units at a time using the `chunk` argument.

```python
# Process residues in chunks of 50
for names in msm.Iterator(lysozyme, element='group', chunk=50, group_name=True):
    print(f"Chunk received with {len(names)} residues.")
```

### 3. Iterating over Time (Structures)
This is the most common use of iterators: processing trajectories frame by frame.

```python
villin_traj = systems['chicken villin HP35']['traj_chicken_villin_HP35_solvated.dcd']

# Iterate over structures (frames)
for frame_idx, coords in msm.Iterator(villin_traj, element='structure', coordinates=True):
    # Note: frame_idx is automatically provided in structure iteration
    if frame_idx % 5 == 0:
        print(f"Processing frame {frame_idx}... Coordinates shape: {coords.shape}")
```

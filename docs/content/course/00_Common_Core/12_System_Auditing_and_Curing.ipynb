# Module 11: System Auditing and Curing

**Welcome back, Apprentice Master.** In [Module 11: Iterating over Hierarchies](../00_Common_Core/12_Iterating_over_Hierarchies.ipynb), you mastered memory-efficient streaming with `msm.Iterator()`. Now we focus on structural quality assurance: **System Auditing and Curing**.

In computational structural biology, a single missing side-chain atom, unresolved loop gap, or missing covalent bond can cause molecular dynamics simulations to crash or produce physical artifacts. Before performing simulations or binding calculations, auditing system completeness and identifying structural defects is mandatory.

MolSysMT provides dedicated quality audit tools inside the **`msm.build`** submodule to detect non-standard groups, sequence gaps, incomplete side chains, and missing covalent bonds.

### 1. Identifying Non-Standard Groups

Let's begin by importing MolSysMT and loading the Triosephosphate Isomerase (TcTIM) structure.

````python
import molsysmt as msm
from molsysmt import systems

# Load TcTIM enzyme structure
enzyme = systems['TcTIM']['1tcd.pdb']
````

Many experimental PDB files contain co-crystallized ligands, ions, or non-canonical modified amino acids. You can identify groups that differ from standard biological building blocks using `msm.build.get_non_standard_residues()`:

````python
# Audit non-standard groups in TcTIM
non_std_groups = msm.build.get_non_standard_residues(enzyme)
print(f"Non-standard groups detected: {non_std_groups}")
````

:::{hint}
:class: dropdown
**msm.build.get_non_standard_residues()**: Scans a molecular system topology to identify non-standard amino acids, modified nucleotides, or co-crystallized ligands. See API doc: {func}`molsysmt.build.get_non_standard_residues`.
:::

### 2. Detecting Missing Gaps & Atoms

X-ray crystallography and cryo-EM structures frequently have unobserved flexible loops or unresolved side-chain atoms. MolSysMT provides functions to audit sequence gaps and incomplete side-chains:

````python
# Detect sequence gaps (missing residues) in TcTIM
gaps = msm.build.get_missing_residues(enzyme)
print(f"Missing sequence gaps: {gaps}")

# Audit missing heavy atoms in Alzheimer Fibril (PDB 2BEG)
alzheimer = 'pdb_id:2BEG'
missing_heavy = msm.build.get_missing_heavy_atoms(alzheimer)
print(f"Groups with missing heavy atoms in 2BEG: {len(missing_heavy)}")
````

### 3. Auditing Covalent Bonds

Standard PDB files often lack explicit CONECT records for protein backbones or side chains. You can audit missing covalent bonds and locate disulfide bridges:

````python
# Audit missing covalent bonds in the Alzheimer PDB file
missing_bonds = msm.build.get_missing_bonds(alzheimer)
print(f"Missing covalent bonds in 2BEG PDB: {len(missing_bonds)}")

# Detect disulfide bridges (S-S bonds)
disulfide_bridges = msm.build.get_disulfide_bonds(alzheimer)
print(f"Disulfide bridges detected: {len(disulfide_bridges)}")
````

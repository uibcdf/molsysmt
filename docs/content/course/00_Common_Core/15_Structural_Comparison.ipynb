# Module 15: Structural Comparison & Identity

In Computational Biology, you often need to verify if two models are consistent. Did you lose any atoms during a file conversion? Is the topology of your simulation identical to the starting PDB? 

The function `compare()` is a high-level validator that checks for topological and structural identity between any two systems.

```python
import molsysmt as msm
from molsysmt import systems

lysozyme = systems['T4 lysozyme L99A']['181l.h5msm']
```

### 1. Basic Comparison
By default, `compare()` checks if the two systems have the same fundamental topology (number of atoms, groups, chains, etc.).

```python
# Let's compare the system with a copy of itself. It should be True.
lysozyme_copy = msm.copy(lysozyme)
print(f"Are they identical? {msm.compare(lysozyme, lysozyme_copy)}")
```

### 2. Detecting Differences
Let's create a sub-system (only the protein) and compare it with the full system (protein + ligand + water).

```python
protein_only = msm.extract(lysozyme, selection='molecule_type=="protein"')

print(f"Is protein_only identical to the full system? {msm.compare(lysozyme, protein_only)}")
```

### 3. Comparing Specific Attributes
You can focus the comparison on specific properties. For example, do they have the same box dimensions? Or the same number of atoms?

```python
# This will return False because the full system has many more atoms
print(f"Same number of atoms? {msm.compare(lysozyme, protein_only, n_atoms=True)}")
```

### 4. Inverse Logic (Negation Flags)
You can use `compare()` to validate that two systems are **different** in a specific way. This is very useful for unit testing.

```python
# We expect them to have a DIFFERENT number of atoms. 
# Passing n_atoms=False means 'check that n_atoms is NOT the same'.
validation = msm.compare(lysozyme, protein_only, n_atoms=False)

print(f"Validation (Expected Difference): {validation}")
```

# Module 11: Sequences & Identity

Molecular biology is not just about 3D coordinates; it is also about the 1D code of life: the **Sequence**. 

MolSysMT allows you to extract sequences from your models and perform sequence-level analysis without leaving the framework.

```python
import molsysmt as msm
from molsysmt import systems

lysozyme = systems['T4 lysozyme L99A']['181l.h5msm']
```

### 1. Extracting Sequences
You can get the sequence of any protein or nucleic acid in your system. By default, it returns a list of strings (one per entity or chain).

```python
# Get the sequence of the main protein (Chain 0)
sequence = msm.get_sequence(lysozyme, selection='chain_index==0')

print(f"The sequence is {len(sequence[0])} residues long.")
print(f"Sequence (1-letter code):\n{sequence[0]}")
```

### 2. Sequence Identity
How similar are two systems in terms of sequence? Let's compare our Lysozyme with another system from the database: the **TcTIM** enzyme.

```python
tctim = systems['TcTIM']['1tcd.h5msm']

# Calculate sequence identity between Lysozyme and TcTIM
identity = msm.get_sequence_identity(lysozyme, tctim)

print(f"Sequence Identity: {identity:.2f} (0 to 1 scale)")
```

### 3. Sequence Alignment
MolSysMT can align the sequences of two systems to see where they match.

```python
# Get the alignment summary
alignment = msm.get_sequence_alignment(lysozyme, tctim)
print("Alignment result:")
print(alignment)
```

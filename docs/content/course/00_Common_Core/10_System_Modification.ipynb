# Module 9: System Modification

Analysis often requires editing the system: moving atoms, renaming groups, or changing chain identifiers. In MolSysMT, this is done with the **`set()`** function.

> **Crucial Concept:** Not all molecular system forms are mutable. For example, you cannot modify a file on disk directly using `set()`. You first need to convert your system into a mutable form in memory, like **`molsysmt.MolSys`**.

````python
import molsysmt as msm
from molsysmt import systems

# Load and convert to a mutable native object
lysozyme = systems['T4 lysozyme L99A']['181l.bcif.gz']
molsys = msm.convert(lysozyme, to_form='molsysmt.MolSys')
````

### 1. Modifying Coordinates
You can change the position of any atom using any unit library or even plain strings.

````python
# Move the first atom to the origin
msm.set(molsys, element='atom', selection=0, coordinates='[0.0, 0.0, 0.0] nm')

# Check the new position
new_pos = msm.get(molsys, selection=0, coordinates=True)
print(f"New position: {new_pos}")
````

### 2. Renaming Entities and Chains
Sometimes PDB files have confusing names. You can standardize them programmatically.

````python
# Rename Chain 'A' to 'PROTEIN'
msm.set(molsys, element='chain', selection='chain_id=="A"', chain_id='PROTEIN')

# Verify with info()
msm.info(molsys, element='chain')
````

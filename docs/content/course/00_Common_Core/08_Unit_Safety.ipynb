# Module 8: Unit Safety with PyUnitWizard

In Computational Structural Biology, numbers without units are dangerous. Is that distance in Angstroms or Nanometers? If you get it wrong, your simulation will explode. 

MolSysMT uses [**PyUnitWizard**](https://www.uibcdf.org/pyunitwizard) to ensure total physical safety. PyUnitWizard is the universal unit engine of the [**MolSysSuite**](https://www.uibcdf.org/), and you will find it in every tool of our ecosystem.

````python
import molsysmt as msm
from molsysmt import pyunitwizard as puw
from molsysmt import systems

lysozyme = msm.convert(systems['T4 lysozyme L99A']['181l.bcif.gz']) # By default conversions are to form molsysmt.MolSys
````

### 1. Agnostic Input, Structured Output

MolSysMT is **Quantity Agnostic** in the input. When a function expects a physical value, you don't need to worry about the specific library or unit. As long as the **dimensionality** (length, time, energy) is correct, PyUnitWizard will handle the translation automatically.

You can even pass physical values as simple **strings**!

```{admonition} Glossary: Quantity Agnosticism
:class: info
The design principle that allows MolSysMT to accept physical values in any format (strings, Pint, OpenMM, etc.). As long as the dimensionality is correct, the library handles the conversion automatically.
```

````python
# Set the coordinates of atom 45 using a string instead of a formal Quantity object
msm.set(lysozyme, element='atom', selection=45, coordinates='[1.0, 4.0, -2.0] angstroms')

# When we get it back, MolSysMT returns a formal Quantity (using Pint by default)
coords = msm.get(lysozyme, selection=45, coordinates=True)

print(f"New coordinates: {coords}")
print(f"Is it a quantity? {puw.is_quantity(coords)}")
````

### 2. Changing Global Preferences

MolSysMT doesn't store its own unit settings; it simply respects the global configuration of **PyUnitWizard**. If you want to change the default units or the library used for the output, you do it directly through the engine's configuration.

````python
# Let's see the current standard units registered in the engine
print(f"Standard units: {puw.configure.get_standard_units()}")

# Change the standard unit for length to Angstroms
puw.configure.set_standard_units(['angstrom', 'ps', 'mol', 'K', 'nm**2', 'rad'])

# Now, any MolSysMT call will return values in Angstroms
coords_ang = msm.get(lysozyme, selection=45, coordinates=True)
print(f"Coordinates in Angstroms: {coords_ang}")
````

### 3. Basic Arithmetic and Safety
The power of PyUnitWizard is that it prevents you from performing illegal physical operations. You cannot add a length to a time, and you cannot subtract two quantities with different dimensionalities.

````python
d1 = puw.quantity(1.0, 'nm')
d2 = puw.quantity(5.0, 'angstrom')

total = d1 + d2
print(f"Total distance: {total}")
````

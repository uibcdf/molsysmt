```{admonition} Glossary: MolSysMT Submodules & msm.build
:class: dropdown info
MolSysMT organizes specialized capabilities into domain-specific submodules:
- **`msm.basic`**: Core form-agnostic operations (conversion, selection, attribute extraction, modification, streaming).
- **`msm.build`**: System editing, residue/atom addition, solvation, synthesis, and quality auditing.
- **`msm.topology`**: Sequence alignment, covalent graphs, and secondary structure.
- **`msm.structure`**: Spatial alignment, RMSD, contacts, and geometric transformations.
- **`msm.physchem`**: Physical and chemical properties (charge, mass, SASA).

The **`msm.build`** submodule encompasses three major functional pillars:
1. **Audit & Inspection**: `get_non_standard_residues()`, `get_missing_residues()`, `get_missing_heavy_atoms()`, `get_missing_bonds()`, `get_disulfide_bonds()`, `get_missing_terminal_cappings()`, `has_hydrogens()`, `is_solvated()`.
2. **Repair & Completion**: `add_missing_heavy_atoms()`, `add_missing_hydrogens()`, `add_missing_bonds()`, `add_missing_terminal_cappings()`, `solve_atoms_with_alternate_location()`.
3. **Synthesis & Solvation**: `build_peptide()`, `solvate()`, `make_water_box()`, `mutate()`, `make_bioassembly()`, `remove_overlapping_molecules()`.
```

### 1. Identifying Non-Standard Groups

Let's begin by importing MolSysMT and loading our T4 Lysozyme demonstration system.

````python
import molsysmt as msm
from molsysmt import systems

# Load T4 Lysozyme system
lysozyme = systems['T4 lysozyme L99A']['181l.bcif.gz']
````

Many experimental PDB and mmCIF files contain co-crystallized ligands, ions, or non-canonical modified amino acids. You can identify groups that differ from standard biological building blocks using `msm.build.get_non_standard_residues()`:

````python
# Audit non-standard groups in T4 Lysozyme
non_std_groups = msm.build.get_non_standard_residues(lysozyme)
print(f"Non-standard groups detected: {non_std_groups}")
````

:::{hint}
:class: dropdown
**msm.build.get_non_standard_residues()**: Scans a molecular system topology to identify non-standard amino acids, modified nucleotides, or co-crystallized ligands. See API doc: {func}`molsysmt.build.get_non_standard_residues`.
:::

### 2. Detecting Missing Gaps & Atoms

X-ray crystallography and cryo-EM structures frequently have unobserved flexible loops or unresolved side-chain atoms. MolSysMT provides functions to audit sequence gaps and incomplete side-chains:

````python
# Detect sequence gaps (missing residues) in T4 Lysozyme using PDB 181L
gaps = msm.build.get_missing_residues('pdb_id:181L')
print(f"Missing sequence gaps: {gaps}")

# Audit missing heavy atoms in T4 Lysozyme
missing_heavy = msm.build.get_missing_heavy_atoms('pdb_id:181L')
print(f"Groups with missing heavy atoms: {len(missing_heavy)}")
````

### 3. Auditing Covalent Bonds

Standard PDB files often lack explicit CONECT records for protein backbones or side chains. You can audit missing covalent bonds and locate disulfide bridges:

````python
# Audit missing covalent bonds in the T4 Lysozyme PDB file
missing_bonds = msm.build.get_missing_bonds('pdb_id:181L')
print(f"Missing covalent bonds in PDB 181L: {len(missing_bonds)}")

# Detect disulfide bridges (S-S bonds)
disulfide_bridges = msm.build.get_disulfide_bonds(lysozyme)
print(f"Disulfide bridges detected: {len(disulfide_bridges)}")
````

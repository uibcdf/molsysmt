# Module 14: Inferred Connectivity (Heuristics)

In an ideal world, every file would tell us which atoms are bonded. In reality, standard **PDB** files often omit this information for the protein, leaving only the coordinates. 

MolSysMT has a built-in inference engine that uses chemical heuristics to reconstruct missing bonds. In this module, we will learn how to bring back these "invisible wires".

```python
import molsysmt as msm
from molsysmt import systems
```

### 1. Detecting Missing Bonds
Let's load a raw PDB from the cloud. PDB files usually only define bonds for non-standard residues (CONECT records), but not for the protein backbone and sidechains.

```python
raw_pdb = 'pdb:181L'
has_bonds = msm.has_attribute(raw_pdb, 'bonded_atom_pairs')
print(f"Does the raw PDB ID have a full list of bonds? {has_bonds}")
```

### 2. Reconstructing the Graph with `add_missing_bonds()`
If your system is missing connectivity, you can tell MolSysMT to compute it based on the distances between atoms.

```python
# We convert to a native object and compute bonds
molsys = msm.convert(raw_pdb, to_form='molsysmt.MolSys')
msm.build.add_missing_bonds(molsys)

new_has_bonds = msm.has_attribute(molsys, 'bonded_atom_pairs')
n_bonds = msm.get(molsys, element='system', n_bonds=True)

print(f"Does it have bonds now? {new_has_bonds}")
print(f"Computed {n_bonds} bonds.")
```

### 3. Detecting Disulfide Bridges
Special bonds like Disulfide bridges (S-S) are critical for protein stability. MolSysMT can find them automatically.

```python
ss_bonds = msm.get_disulfide_bonds(molsys)
print(f"Found {len(ss_bonds)} disulfide bridges in T4 Lysozyme (raw).")

if ss_bonds:
    print(f"First S-S bond between atoms: {ss_bonds[0]}")
```

### 4. Automatic Inference in Functions
Many functions in MolSysMT that require bonds (like `get_sequence` or `info(element='molecule')`) will try to trigger this inference automatically if you haven't done it yet. This is part of the "Smart Engine" of the framework.

# Module 4: Interactive Selection

One of the biggest pain points in molecular modeling is identifying specific atoms or residues by eye and then having to find their indices manually to perform a calculation. 

MolSysMT solves this by syncing with the 3D Viewer. You can pick residues with your mouse and bring them back into your Python script instantly.

````python
import molsysmt as msm
from molsysmt import systems

lysozyme = systems['T4 lysozyme L99A']['181l.h5msm']
````

### 1. Launching the Interactive Viewer
When you call `view()`, it returns a **viewer object**. We need to store this object in a variable to talk to it later.

````python
view = msm.view(lysozyme)
view
````

### 2. The "Pick and Fetch" Workflow

**Instruction for you:** 
1. Look at the viewer above.
2. Use your mouse to **click** on 3 or 4 different residues.
3. Once you have selected them, execute the cell below.

````python
# Recover the active selection from the viewer
view.active_selection.save(tag='my_selection')

# Now the selection is stored in the viewer's registry
my_indices = view.selections['my_selection'].atom_indices

print(f"You picked {len(my_indices)} atoms interactively!")
print(f"Indices: {my_indices}")
````

### 3. Closing the Loop
Now that those indices are in a Python variable (`my_indices`), you can use them for any analysis. For example, let's find out which residues those atoms belong to.

````python
res_names = msm.get(lysozyme, selection=my_indices, group_name=True)
print(f"The atoms you clicked belong to these residues: {set(res_names)}")
````

# Module 17: Semantic Labeling

Computers love indices (0, 1, 2...), but humans love names. If you are writing a paper or a report, saying "Carbon Alpha of Lysine 15" is much better than "Atom 142".

MolSysMT has a semantic labeling engine that converts any selection into a professional nomenclature string.

````python
import molsysmt as msm
from molsysmt import systems

lysozyme = systems['T4 lysozyme L99A']['181l.bcif.gz']
````

### 1. The Default Label
The function `get_label()` provides a standard, unambiguous string for any atom or group.

````python
# Get the label of the first atom
label = msm.get_label(lysozyme, selection='atom_index==0')
print(f"Default atom label: {label}")

# Get the label of a whole group (group)
res_label = msm.get_label(lysozyme, element='group', selection='group_index==10')
print(f"Default group label: {res_label}")
````

### 2. Designing Your Own Labels
You don't have to stick with the default format. You can use any attribute as a variable inside a string template using curly braces `{}`. This works like Python f-strings but for molecular data.

````python
# Custom format: Name of group and its ID
custom_1 = msm.get_label(lysozyme, element='group', selection=10, label='{group_name}:{group_id}')
print(f"Custom label 1: {custom_1}")

# Complex format: Atom name, group name, and chain ID
template = '{atom_name} from {group_name}{group_id} (Chain {chain_id})'
custom_2 = msm.get_label(lysozyme, selection=0, label=template)
print(f"Custom label 2: {custom_2}")
````

### 3. Batch Labeling
You can generate labels for a whole list of items at once. This is perfect for generating professional CSV reports or plot legends.

````python
# Label all C-alphas of the first 5 groups using a custom template
ca_labels = msm.get_label(lysozyme, 
                          selection='atom_name=="CA" and group_index==[0,1,2,3,4]',
                          label='Alpha-C of {group_name}-{group_id}')

for l in ca_labels:
    print(f" - {l}")
````

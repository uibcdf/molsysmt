# Module 12: Navigating Between Levels

In MolSysMT, you don't need to manually keep track of which atom belongs to which biological unit. The framework acts like a smart map. You can easily ask: "Who is the parent of this atom?" or "Who are the children of this chain?"

This ability to jump between hierarchical levels is one of the most powerful features of the `get()` function.

```python
import molsysmt as msm
from molsysmt import systems

lysozyme = systems['T4 lysozyme L99A']['181l.bcif.gz']
```

### 1. Mapping Upwards (Atom to Group)
Imagine you have a list of atom indices. You want to know which **groups** they belong to. Remember that in MolSysMT, we use the term **group** for any collection of atoms that form a unit (groups, ions, or small molecules).

```python
# Let's pick atoms 10, 50, and 100
my_atoms = [10, 50, 100]

# We ask for the group name and index of those specific atoms
group_names, group_indices = msm.get(lysozyme, element='atom', selection=my_atoms, 
                                   group_name=True, group_index=True)

print(f"Atom 10 belongs to group: {group_names[0]} (index {group_indices[0]})")
print(f"Atom 50 belongs to group: {group_names[1]} (index {group_indices[1]})")
```

### 2. Mapping Downwards (Entity to Atoms)
You can also go the other way. If you have a biological entity (like a ligand), you can find all the atoms that compose it.

```python
# Find the atoms that compose the BENZENE entity
atom_indices = msm.get(lysozyme, element='entity', selection='entity_name=="BENZENE"', atom_index=True)

print(f"The Benzene entity is composed of atoms: {atom_indices}")
```

### 3. Inverse Relational Logic
Finding what is **NOT** part of a relationship is very useful. For example, finding all atoms that do not belong to the main protein chain.

```python
# Select atoms where their parent chain index is NOT 0
non_chain_0 = msm.select(lysozyme, selection='not chain_index==0')

print(f"Found {len(non_chain_0)} atoms outside Chain 0 (water, ions, and ligands).")
```

### 4. Cross-Level Summaries
How many groups are in each chain?

```python
chain_indices, n_groups = msm.get(lysozyme, element='chain', chain_index=True, n_groups=True)

for idx, count in zip(chain_indices, n_groups):
    print(f"Chain {idx} has {count} groups.")
```

# Module 12: Canonical Data & Standard Naming

Biological data is messy. Depending on which force field or software was used to generate a file, the same residue might be called `HIS`, `HIE`, `HID`, or `HSP`. This makes automation very difficult.

MolSysMT has an internal **Knowledge Base** of canonical residue and atom names. It can tell you what something *is*, regardless of what the file *says* it is.

```python
import molsysmt as msm
from molsysmt import systems

lysozyme = systems['T4 lysozyme L99A']['181l.h5msm']
```

### 1. The Standard Identity
The function `get_standard_name()` looks up the residue in the MolSysMT database and returns its canonical name.

```python
# Let's pick a residue and see its name in the file
file_name = msm.get(lysozyme, element='group', selection='group_index==0', group_name=True)[0]

# Now let's ask for its standard name using the specific amino_acid module
from molsysmt.element.group.amino_acid import get_standard_name
standard_name = get_standard_name(file_name)

print(f"File name: {file_name}")
print(f"Standard canonical name: {standard_name}")
```

### 2. Identifying Non-Standard Residues
You can quickly find things that MolSysMT doesn't recognize as standard amino acids or nucleotides (like ligands, cofactors, or modified residues).

```python
# Identifying Non-Standard Residues
group_names = msm.get(lysozyme, element='group', group_name=True)
from molsysmt.element.group.amino_acid import get_standard_name

# Check which residues have a name different from their standard one
for name in set(group_names):
    std = get_standard_name(name)
    if std and std != name:
        print(f"Residue {name} is standardized to {std}")
```

### 3. Atom Naming Consistency
Standardization also applies to atoms. This is vital when converting between libraries (e.g., from MDTraj to OpenMM).

```python
# Get atom names as they appear in the file
atom_names = msm.get(lysozyme, element='atom', selection='group_index==0', atom_name=True)
print(f"Original atom names: {atom_names[:5]}")

# Get standard atom names according to the canonical template
std_atom_names = msm.get(lysozyme, element='atom', selection='group_index==0', atom_standard_name=True)
print(f"Standard atom names: {std_atom_names[:5]}")
```

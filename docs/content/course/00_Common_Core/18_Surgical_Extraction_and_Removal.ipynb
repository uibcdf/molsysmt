# Module 18: Surgical Extraction and Removal

**Welcome back, Apprentice Master.** In [Module 17: Merging and Growing Systems](../00_Common_Core/17_Merging_and_Growing_Systems.ipynb), you learned how to assemble complex molecular systems using `msm.merge()` and `msm.add()`. Now we explore the inverse operations: **Surgical Extraction and Removal**.

Large simulation boxes often contain tens of thousands of water molecules, counterions, or bulk solvent. Analyzing a binding pocket or isolated protein domain requires cutting away non-essential components without corrupting atomic indices or structural topology. MolSysMT provides dedicated extraction and removal engines (**`msm.extract()`**, **`msm.remove()`**, and **`msm.copy()`**) for precise sub-system manipulation.

### 1. Extracting Sub-Systems

Let's begin by importing MolSysMT and loading our T4 Lysozyme demonstration system.

````python
import molsysmt as msm
from molsysmt import systems

# Load T4 Lysozyme as native MolSys
lysozyme = msm.convert(systems['T4 lysozyme L99A']['181l.bcif.gz'], to_form='molsysmt.MolSys')
````

`msm.extract()` creates a new independent molecular system containing only the selected atoms or components, leaving the original system untouched:

````python
# Extract only the protein component
protein_only = msm.extract(lysozyme, selection='molecule_type == "protein"')

# Inspect topology overview of extracted protein
msm.info(protein_only)
````

:::{hint}
:class: dropdown
**msm.extract()**: Creates a new independent molecular system from selected atoms, groups, or components. See API doc: {func}`molsysmt.basic.extract`.
:::

### 2. Removing Elements

If you want to keep everything in the system *except* specific unwanted elements (such as removing solvent water or counterions), use `msm.remove()`:

````python
# Remove water molecules and ions from the system
dry_system = msm.remove(lysozyme, selection='molecule_type == "water" or molecule_type == "ion"')

# Inspect topology overview of dry system
msm.info(dry_system)
````

### 3. Duplicating Systems

In Python, simple variable assignment (`sys_b = sys_a`) creates a shared reference rather than a new object. To create a deep, independent copy before applying mutations, use `msm.copy()`:

````python
# Create an independent copy of the system
lysozyme_copy = msm.copy(lysozyme)

# Verify identity using msm.compare()
print(f"Is the copy identical to the original? {msm.compare(lysozyme, lysozyme_copy)}")
````

# Module 17: Merging and Growing Systems

**Welcome back, Apprentice Master.** In [Module 16: Semantic Labeling](../00_Common_Core/16_Semantic_Labeling.ipynb), you mastered generating human-readable nomenclature strings using `msm.get_label()`. Now we transition to system composition: **Merging and Growing Systems**.

Complex molecular simulations frequently require assembling systems from separate components: fusing a receptor protein and a co-crystallized ligand, appending ions or water molecules, or joining sequential trajectory segments along a timeline. MolSysMT provides dedicated composition functions (**`msm.merge()`**, **`msm.add()`**, and **`msm.append_structures()`**) to construct unified molecular systems.

### 1. Merging Independent Systems

Let's begin by importing MolSysMT, loading a protein system (Villin Headpiece) and a small molecule (Benzamidine), and converting them to native `MolSys` objects.

````python
import molsysmt as msm
from molsysmt import systems

# Load protein and ligand structures
protein = msm.convert(systems['chicken villin HP35']['1vii.pdb'], to_form='molsysmt.MolSys')
ligand = msm.convert(systems['benzamidine']['benzamidine.pdb'], to_form='molsysmt.MolSys')
````

`msm.merge()` takes a list of independent molecular systems and fuses their topology and spatial coordinates into a single combined system:

````python
# Merge protein and ligand into a single complex
complex_sys = msm.merge([protein, ligand])

# Inspect topology overview of the merged complex
msm.info(complex_sys)
````

:::{hint}
:class: dropdown
**msm.merge()**: Fuses multiple independent molecular systems into a unified target system in memory. See API doc: {func}`molsysmt.basic.merge`.
:::

### 2. Adding Elements

While `msm.merge()` creates a new combined system from a list, **`msm.add()`** appends selected elements or molecules into an existing target modifiable system in place:

````python
# Print atom count before addition
print(f"Protein atom count before addition: {msm.get(protein, element='system', n_atoms=True)}")

# Add ligand elements directly into the protein system in place
msm.add(protein, ligand)

# Print updated atom count
print(f"Protein atom count after addition: {msm.get(protein, element='system', n_atoms=True)}")
````

### 3. Concatenating Trajectories

When a molecular dynamics simulation is executed in sequential segments (e.g. multiple DCD or XTC files), use **`msm.append_structures()`** to concatenate coordinate frames along the time/structure axis:

````python
# Load Villin Headpiece trajectory as native Structures object
traj = msm.convert(systems['chicken villin HP35']['traj_chicken_villin_HP35_solvated.dcd'], to_form='molsysmt.Structures')

# Concatenate trajectory with itself to simulate joining two sequential simulation runs
msm.append_structures(traj, traj)

n_structs_long = msm.get(traj, element='system', n_structures=True)
print(f"Concatenated trajectory total structure count: {n_structs_long}")
````

# Module 13: Iterating over Hierarchies

Extracting all data at once with `get()` is fast, but sometimes you need to process a system piece by piece. You might want to loop over every group to perform a custom calculation, or process a massive trajectory structure by structure without loading it all into RAM.

MolSysMT provides a powerful **`Iterator()`** for these cases.

````python
import molsysmt as msm
from molsysmt import systems

lysozyme = systems['T4 lysozyme L99A']['181l.bcif.gz']
````

### 1. Simple Biological Loops
You can iterate over any hierarchical level (`group`, `molecule`, `chain`, etc.). The iterator will yield the requested attributes for each unit.

````python
# Let's iterate over the first 5 groups (groups)
for name, atoms in msm.Iterator(lysozyme, element='group', selection=[0,1,2,3,4], 
                                group_name=True, atom_index=True):
    print(f"Processing group {name}... it has {len(atoms)} atoms.")
````

### 2. The Power of Chunking
For very large systems, you can process multiple units at a time using the `chunk` argument.

````python
# Process groups in chunks of 50
for names in msm.Iterator(lysozyme, element='group', chunk=50, group_name=True):
    print(f"Chunk received with {len(names)} groups.")
````

### 3. Iterating over Time (Structures)
This is the most common use of iterators: processing trajectories structure by structure.

````python
villin_traj = systems['chicken villin HP35']['traj_chicken_villin_HP35_solvated.dcd']

# Enumerate the structures yielded with their coordinates
for structure_index, coords in enumerate(msm.Iterator(villin_traj, coordinates=True)):
    if structure_index % 5 == 0:
        print(f"Processing structure {structure_index}... Coordinates shape: {coords.shape}")
````

# Module 11: Navigating Between Levels

**Welcome back, Apprentice Master.** In [Module 10: Modifying Molecular Attributes](../00_Common_Core/10_Modifying_Molecular_Attributes.ipynb), you learned how to update system properties and 3D spatial coordinates using `msm.set()`. Now we explore one of MolSysMT's most powerful relational features: **Navigating Between Levels**.

Molecular systems are hierarchical networks connecting atoms, groups, components, molecules, chains, entities, and whole systems (as established in Module 3). In traditional tools, mapping an atom index to its parent chain or finding all atoms belonging to a specific entity requires writing custom loops. In MolSysMT, **`msm.get()`** and **`msm.select()`** serve as a relational map allowing you to jump upwards or downwards across levels effortlessly.

### 1. Upward Hierarchical Mapping

Let's begin by importing MolSysMT and loading our T4 Lysozyme demonstration system.

````python
import molsysmt as msm
from molsysmt import systems

# Load T4 Lysozyme system
lysozyme = systems['T4 lysozyme L99A']['181l.bcif.gz']
````

To map upwards from child elements to their parent containers, pass the child selection and request parent attributes (such as `group_name`, `group_index`, or `chain_id`):

Let's find which groups and chains atom indices 10, 50, and 100 belong to:

````python
target_atoms = [10, 50, 100]

# Query group_name, group_index, and chain_id for specific atom indices
group_names, group_indices, chain_ids = msm.get(
    lysozyme,
    element='atom',
    selection=target_atoms,
    group_name=True,
    group_index=True,
    chain_id=True
)

for atom, g_name, g_idx, c_id in zip(target_atoms, group_names, group_indices, chain_ids):
    print(f"Atom {atom:>3} -> Group: {g_name} (index {g_idx}) | Chain: {c_id}")
````

:::{hint}
:class: dropdown
**Relational Navigation**: In `msm.get()`, specifying `element='atom'` while requesting `group_name` or `chain_id` automatically performs upward relational mapping. See API doc: {func}`molsysmt.basic.get`.
:::

### 2. Downward Hierarchical Mapping

You can also map downwards from a parent element (such as an entity or chain) to retrieve all of its constituent child elements (such as atom or group indices):

Let's find all atom indices belonging to the Benzene ligand entity:

````python
# Query atom indices composing the BENZENE entity
benzene_atom_indices = msm.get(lysozyme, element='entity', selection='entity_name == "BENZENE"', atom_index=True)

print(f"Atom indices composing BENZENE entity: {benzene_atom_indices}")
````

### 3. Cross-Level Summaries

You can combine element scoping to extract counts of child elements across parent containers (for instance, asking how many groups exist in each chain):

````python
# Extract chain indices and the number of groups contained in each chain
chain_indices, n_groups_per_chain = msm.get(lysozyme, element='chain', chain_index=True, n_groups=True)

for c_idx, n_g in zip(chain_indices, n_groups_per_chain):
    print(f"Chain index {c_idx} contains {n_g} groups.")
````

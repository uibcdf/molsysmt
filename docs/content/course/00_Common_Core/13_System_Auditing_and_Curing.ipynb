# Module 13: System Auditing and Curing

**Why does this matter?** A single missing atom or a broken covalent bond can cause a molecular dynamics simulation to crash or produce physical nonsense. Curing your system is the difference between a high-impact paper and a retracted one.

In this module, you will learn to detect gaps, missing atoms, and topological errors using the [**`build` submodule**](https://www.uibcdf.org/molsysmt).

````python
import molsysmt as msm
from molsysmt import systems
````

### 1. Identifying Non-Standard Groups
Many systems contain ligands or modified groups. MolSysMT can quickly identify which groups don't belong to the standard biological building blocks using `msm.build.get_non_standard_residues()`.

*Note: We use a PDB file here because it often contains a more explicit description of the sequence than some binary formats.*

````python
enzyme = systems['TcTIM']['1tcd.pdb']

# Which groups are not standard amino acids or nucleotides?
non_std = msm.build.get_non_standard_residues(enzyme)
print(f"Non-standard groups found: {non_std}")
````

### 2. Hunting for Missing Gaps and Atoms
A structure might have missing groups (gaps in the sequence) or missing heavy atoms (incomplete side chains). 

- `get_missing_residues()`: Finds gaps in the protein sequence.
- `get_missing_heavy_atoms()`: Identifies incomplete groups.

````python
# Are there gaps in the enzyme sequence?
gaps = msm.build.get_missing_residues(enzyme)
print(f"Missing groups (gaps) found: {gaps}")

# Let's check the Alzheimer Fibril (2BEG) for incomplete side-chains
alzheimer = 'pdb_id:2BEG'
missing_atoms = msm.build.get_missing_heavy_atoms(alzheimer)
print(f"Number of residues with missing atoms in Alzheimer system: {len(missing_atoms)}")
````

### 3. Auditing the Bonds
Standard PDB files often lack a complete list of covalent bonds for the protein. You can audit the connectivity using `msm.build.get_missing_bonds()` and even find special links like Disulfide bridges.

````python
# Does the raw Alzheimer PDB have all its bonds defined?
missing_bonds = msm.build.get_missing_bonds(alzheimer)
print(f"Found {len(missing_bonds)} missing covalent bonds in the PDB file.")

# Search for Disulfide Bridges
ss_bonds = msm.build.get_disulfide_bonds(alzheimer)
print(f"Found {len(ss_bonds)} S-S bonds.")
````

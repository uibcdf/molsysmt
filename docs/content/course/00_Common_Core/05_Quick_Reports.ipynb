# Module 5: Quick Reports & Summaries

Once you have loaded a system, the first thing you usually want is an inventory. What is inside? How many chains? Are there ligands? 

MolSysMT provides two main ways to answer these questions: `info()` (for human-readable tables) and `get()` (for raw data extraction).

```python
import molsysmt as msm
from molsysmt import systems

lysozyme = systems['T4 lysozyme L99A']['181l.h5msm']
```

### 1. The System Inventory with `info()`
By default, `info()` gives you a summary of the whole system. It returns a styled Pandas table that is very easy to read.

```python
msm.info(lysozyme)
```

### 2. Inspecting Specific Levels
You can change the `element` argument to get details about different hierarchical levels: `atom`, `group`, `molecule`, `chain`, or `entity`.

```python
# Let's see which biological entities are in the system
msm.info(lysozyme, element='entity')
```

```python
# Let's see the chains
msm.info(lysozyme, element='chain')
```

### 3. Combining Info with Selections
You can generate a report for only a part of the system. For example, let's see a table of all non-standard residues (ligands, ions, etc.).

```python
msm.info(lysozyme, element='group', selection='not molecule_type=="protein" and not molecule_type=="water"')
```

### 4. Extracting Raw Data with `get()`
While `info()` is for looking, `get()` is for computing. It returns lists or NumPy arrays that you can use in your code.

```python
# Get the names and IDs of all residues in chain 0
names, ids = msm.get(lysozyme, element='group', selection='chain_index==0', group_name=True, group_id=True)

print(f"First 5 residues: {list(zip(names, ids))[:5]}")
```

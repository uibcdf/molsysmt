# Module 16: Comparing Systems

In Computational Biology, you often need to verify if two models are consistent. Did you lose any atoms during a file conversion? Is your simulation system identical to the starting crystal structure? 

In MolSysMT, we use a single, universal tool for this: **`compare()`**.

```python
import molsysmt as msm
from molsysmt import systems

lysozyme = systems['T4 lysozyme L99A']['181l.bcif.gz']
```

### 1. Basic Comparison (Identity Check)
By default, `compare()` checks the fundamental topology of the systems. It ensures that the "biological blueprint" is the same.

```python
# Compare the system with a native object version of itself
lysozyme_molsys = msm.convert(lysozyme, to_form='molsysmt.MolSys')
print(f"Are they identical? {msm.compare(lysozyme, lysozyme_molsys)}")
```

### 2. Selective Comparison
You can focus the comparison on a specific type of data using the `attribute_type` argument. This is very useful when you expect coordinates to differ but topology to remain the same.

```python
# Compare only the topological part (names, IDs, connectivity)
print(f"Same topology? {msm.compare(lysozyme, lysozyme_molsys, attribute_type='topological')}")

# Compare only the structural part (coordinates, box)
print(f"Same pose? {msm.compare(lysozyme, lysozyme_molsys, attribute_type='structural')}")
```

### 3. Detailed Audit with Dictionaries
If `compare()` returns `False`, you usually want to know **why**. You can change the `output_type` to get a detailed report of every checked attribute.

```python
# Extract the protein and compare it with the full complex
protein_only = msm.extract(lysozyme_molsys, selection='molecule_type=="protein"')

# Get a dictionary report
report = msm.compare(lysozyme, protein_only, output_type='dictionary')

print("Difference report:")
for attr, result in report.items():
    if not result:
        print(f"- {attr}: MISMATCH")
```

### 4. Expecting Differences
You can pass specific attributes as keyword arguments to validate expected differences. This is a common pattern in unit tests.

```python
# We expect them to have a DIFFERENT number of atoms. 
# Passing n_atoms=False means 'validate that n_atoms is NOT the same'.
validation = msm.compare(lysozyme, protein_only, n_atoms=False)

print(f"Validation (Confirmed difference): {validation}")
```

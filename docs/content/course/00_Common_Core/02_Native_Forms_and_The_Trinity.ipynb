# Module 2: Native Forms and The Trinity

```{admonition} Glossary: Trinity
:class: info
The **Trinity** refers to the three pillars that define a molecular system in MolSysMT: **Topology** (connectivity), **Structures** (coordinates), and **Metadata** (context).
```

MolSysMT is agnostic, but it is not without a home. While we can work with PDBs or OpenMM objects, we have our own high-performance data structure: the **Native Form**.

In this module, you will learn about the **Trinity of MolSysMT** and the objects that make it the fastest framework for structural analysis.

````python
import molsysmt as msm
from molsysmt import systems
````

### 1. The Trinity: Topology, Structures, and Metadata

A molecular system in MolSysMT is composed of three independent pillars:
1.  **Topology:** The static covalent graph (names, bonds, sequences). It rarely changes.
2.  **Structures:** The dynamic data (coordinates, velocities, boxes, time). This is what changes in a simulation.
3.  **Metadata:** Contextual information (the form name, the precision, and smonitor signals).

The object **`molsysmt.MolSys`** orchestrates these three pillars.

````python
# Load a system and convert it to the native object
lysozyme = systems['T4 lysozyme L99A']['181l.bcif.gz']
molsys = msm.convert(lysozyme, to_form='molsysmt.MolSys')

print(f"Native object type: {type(molsys)}")
````

### 2. High-Performance Formats: H5MSM

The **H5MSM** format is the disk-based version of our native object. It is an HDF5 container that stores the Trinity with maximum efficiency, allowing near-instant loading of massive trajectories. New files use schema 0.4: stable hierarchy is stored separately from one or more chemical states, including nullable atom chemistry, components, and rich bonds. Partial files may contain structural arrays and a valid atom axis without materializing topology. Optional velocities, temperature, potential energy, and kinetic energy preserve their physical units, and total energy is available when both energy series exist. Existing 0.3 files remain readable and migrate to one reference chemical state; unknown future versions are rejected. Its ten direct native persistence routes have exhaustive, instance-aware reports: full-file projections declare omitted domains, partial files do not invent them, and strict mode rejects detected loss before writing. Coordinate-only XYZ, DCD, and XTC inputs can enter this structural branch without topology; atom selection is canonical, frame order is explicit, and XTC step identifiers are retained.

````python
# Saving a system is just a conversion to a file form
# msm.convert(molsys, to_form='my_system.h5msm')

h5msm_file = systems['T4 lysozyme L99A']['181l.h5msm']
print(f"Form of the file: {msm.get_form(h5msm_file)}")
````

### 3. Specialty Native Forms
Sometimes you only need a part of the Trinity. MolSysMT provides dictionaries for ultra-low latency access:
- **`molsysmt.TopologyDict`**: Only the covalent map.
- **`molsysmt.StructuresDict`**: Structural arrays and optional per-frame observables, including velocities, temperature, potential energy, and kinetic energy. Total energy is derived rather than duplicated. The MolSys, Topology, and Structures declarative 0.1 families preserve aligned subsets within their schemas and have exhaustive conversion reports; unsupported rich state, mechanics, or bioassembly data is reported explicitly.

PDB reads and writes have exhaustive, content-aware reports for native `MolSys`, `Topology`, and `Structures` targets. Native IDs remain labels while PDB atom serials are regenerated as unique local record identifiers. Valid MODEL IDs, occupancy, B factors, coordinates, the global box, formal charges, explicit connectivity (`CONECT`, `LINK`, and `SSBOND`), alternate sites, and `REMARK 350` bioassemblies are preserved within PDB precision and schema limits. Use `return_report=True` to inspect projection, insertion-code, ambiguity, or fixed-width limitations for the selected payload, and `strict=True` to reject detected loss before writing.

````python
topo_dict = msm.convert(molsys, to_form='molsysmt.TopologyDict')
print(f"Topology dictionary keys: {list(topo_dict.keys())[:5]}...")
````

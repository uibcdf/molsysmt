# Module 20: The Specialized Domains (The Atlas)

**Congratulations!** You have mastered the technical foundation of MolSysMT. You know how to load, select, extract, and modify molecular systems. 

But MolSysMT is more than a data manager; it is an ecosystem of specialized tools. Before you choose your scientific path, let's take a quick tour of the "Specialized Domains" (submodules) where the real science happens.

````python
import molsysmt as msm
from molsysmt import systems

lysozyme = systems['T4 lysozyme L99A']['181l.bcif.gz']
````

### 1. `msm.build`: The Workshop
Dedicated to repairing, mutating, and preparing systems for simulation.
```python
# Example: Solvate a system with 0.15M NaCl
cured_system = msm.build.solvate(lysozyme, salt_concentration='0.15 molar')
```

### 2. `msm.structure`: The Geometry Lab
Focuses on spatial analysis, RMSD, PCA, and conformational changes.
```python
# Example: Calculate the RMSD of a trajectory
rmsd = msm.structure.get_rmsd(trajectory, reference=topology)
```

### 3. `msm.topology`: The Biological Archive
Handles sequences, standard nomenclature, and secondary structure.
```python
# Example: Get the secondary structure using DSSP
ss = msm.topology.get_secondary_structure(lysozyme)
```

### 4. `msm.physchem`: The Scales
Calculates physical and chemical properties like Surface Area (SASA), volume, or mass.
```python
# Example: Get the Solvent Accessible Surface Area
sasa = msm.physchem.get_sasa(lysozyme)
```

### 5. `msm.hbonds`: The Interaction Network
Advanced tools for detecting and analyzing hydrogen bonds and salt bridges.
```python
# Example: Find all hydrogen bonds in a complex
hbonds = msm.hbonds.get_hbonds(lysozyme)
```

### 6. `msm.pbc` & `msm.molecular_mechanics`: The Physics Dept
Handles periodic boundary conditions and interacts with physics engines to get energies and forces.
```python
# Example: Get the potential energy of a system
energy = msm.molecular_mechanics.get_potential_energy(lysozyme)
```

### 7. `msm.third_party`: The Embassies
Target-aware bridges to the rest of the molecular ecosystem: OpenMM, MDAnalysis, NGLView, and more. The bridge preserves the documented semantics that each target model can represent; narrower targets may require an explicit conversion-fidelity check.

### 8. `msm.configure`: Process-wide execution policy
Configuration controls defaults such as CPU parallelism, GPU selection, precision, and chunking. Set the CPU session policy with `msm.configure.set_parallelization(...)`; supported functions can override `parallel` and `num_threads` for one call without mutating that session. Temporary `msm.configure.context(...)` session overrides remain process-global and must not overlap across threads.

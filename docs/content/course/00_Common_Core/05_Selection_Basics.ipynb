# Module 5: Selection Language Basics

To analyze or edit a system, you first need to be able to pick the parts you are interested in. In MolSysMT, we do this with the `select()` function and its powerful, human-readable language.

```{note}
For a full reference of the selection grammar, see the [**Selection Syntax section**](https://www.uibcdf.org/molsysmt) of the [**User Guide**](https://www.uibcdf.org/molsysmt).
```

````python
import molsysmt as msm
from molsysmt import systems
lysozyme = systems['T4 lysozyme L99A']['181l.bcif.gz']
````

### 1. The `select()` Function
The function `select()` always returns a **list of atom indices** (0-based).

````python
all_atoms = msm.select(lysozyme, selection='all')
print(f"The system has {len(all_atoms)} atoms.")
````

### 2. Biological Keywords
Pick the backbone or the solvent without knowing the specific names.

````python
protein_atoms = msm.select(lysozyme, selection='molecule_type=="protein"')
backbone_atoms = msm.select(lysozyme, selection='backbone')
print(f"Protein: {len(protein_atoms)} atoms, Backbone: {len(backbone_atoms)} atoms.")
````

### 3. Selecting by Name and ID
Precision picking using group names (`group_name`), atom names, or IDs.

````python
ca_atoms = msm.select(lysozyme, selection='atom_name=="CA"')
res_15 = msm.select(lysozyme, selection='group_id==15')
lys_atoms = msm.select(lysozyme, selection='group_name=="LYS"')
print(f"C-alphas: {len(ca_atoms)}, Residue 15: {len(res_15)} atoms.")
````

### 4. Logic and Sets
Combine criteria or check for belonging to a specific list.

````python
# Select C-alphas belonging to Lysines
lys_ca = msm.select(lysozyme, selection='(atom_name=="CA") and (group_name=="LYS")')

# Select using a specific set of atom names
main_heavy = msm.select(lysozyme, selection='atom_name in ["C", "N", "O", "CA"]')

print(f"Lysine C-alphas: {len(lys_ca)}, Heavy backbone atoms: {len(main_heavy)}.")
````

### 5. Proximity Selection (Spatial Intelligence)
Find atoms based on their physical distance to others.

````python
# All water molecules within 0.5 nm of the protein
shell_water = msm.select(lysozyme, selection='molecule_type=="water" within 0.5 nm of molecule_type=="protein"')
print(f"Hydration shell atoms: {len(shell_water)}.")
````

### 6. Hierarchical Grouping: `in ... of` 
Instead of a flat list, get a list of lists grouped by biological units.

````python
# Get all atoms of the protein, but grouped by group
grouped_atoms = msm.select(lysozyme, selection='all in groups of molecule_type=="protein"')

print(f"Residues found: {len(grouped_atoms)}")
print(f"Atoms in the first residue: {grouped_atoms[0]}")
````

### 7. Using Selections in View
Strings can be used to slice what the 3D viewer displays.

````python
msm.view(lysozyme, selection='group_name=="LYS" or (molecule_type=="water" within 0.3 nm of molecule_type=="protein")')
````

# Module 19: Surgical Extraction and Removal

**Why does this matter?** Real simulations are heavy. Analyzing a 100,000-atom system when you only care about a 20-atom binding pocket is a waste of time and energy. Knowing how to "cut" the system is the mark of an efficient researcher.

In this module, we will learn the high-level "cutting" verbs of MolSysMT: `extract()`, `remove()`, and `copy()`.

````python
import molsysmt as msm
from molsysmt import systems

lysozyme = msm.convert(
    systems['T4 lysozyme L99A']['181l.bcif.gz'],
    to_form='molsysmt.MolSys',
)
````

### 1. Extracting Components
The function `extract()` creates a **new molecular system** containing only the atoms you selected. The original system remains untouched. Atom selections use canonical source-topology order before the extracted system renumbers them from zero and keeps topology and coordinates aligned. Explicit structure-index collections preserve their requested order because structure order can carry meaning.

````python
# Extract only the protein part
protein = msm.extract(lysozyme, selection='molecule_type=="protein"')

msm.info(protein)
````

### 2. Removing Elements
If you want to keep everything **except** a certain selection (like removing all water molecules), use `remove()`.

````python
# Remove all water and ions from the original complex
dry_complex = msm.remove(lysozyme, selection='molecule_type=="water" or molecule_type=="ion"')

msm.info(dry_complex)
````

### 3. Safe Duplication with `copy()`
In Python, simple assignments (`sys_b = sys_a`) create references. To avoid side-effects, use `copy()`.

````python
# Create a true independent copy of the system
lysozyme_backup = msm.copy(lysozyme)

print(f"Are they identical? {msm.compare(lysozyme, lysozyme_backup)}")
````

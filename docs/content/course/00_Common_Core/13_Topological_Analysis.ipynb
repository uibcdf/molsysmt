Now we turn to non-spatial features: **Topological Analysis** with **`msm.topology`**.

In computational biophysics, a molecular system contains properties that are entirely independent of 3D spatial coordinates—such as covalent graph connectivity, connected molecular components, and sequence ordering. While spatial coordinates change continuously during a trajectory, the underlying chemical topology remains invariant.

```{admonition} Glossary: msm.topology module
:class: dropdown info
The **`msm.topology`** function module handles non-spatial, graph-theoretical, and sequence properties of molecular systems:
- **`get_covalent_blocks()`**: Partitions connected covalent components and analyzes sub-blocks under hypothetical bond cuts.
- **`get_bondgraph()`**: Generates the complete network topology as a `networkx.Graph` object.
- **`get_sequence_alignment()` & `get_sequence_identity()`**: Compares residue sequence topologies across systems.
```

### 1. Covalent Blocks

Let's begin by importing MolSysMT and loading our T4 Lysozyme demonstration system.

````python
import molsysmt as msm
from molsysmt import systems

# Load T4 Lysozyme as native MolSys
lysozyme = msm.convert(systems['T4 lysozyme L99A']['181l.bcif.gz'], to_form='molsysmt.MolSys')
````

A **covalent block** is a set of atoms connected continuously through covalent bonds. When called without parameters, `msm.topology.get_covalent_blocks()` partitions the system into independent connected components:

````python
# Partition system into connected covalent blocks
blocks = msm.topology.get_covalent_blocks(lysozyme)
n_components = msm.get(lysozyme, n_components=True)

print(f"Covalent blocks count: {len(blocks)}")
print(f"System components count: {n_components}")
````

You can also use `remove_bonds` to simulate hypothetical bond cuts. For example, cutting an N-CA backbone bond splits a connected protein component into two sub-blocks:

````python
# Select the N and CA atom indices of group index 20
n_atom = msm.select(lysozyme, selection='group_index == 20 and atom_name == "N"')[0]
ca_atom = msm.select(lysozyme, selection='group_index == 20 and atom_name == "CA"')[0]

# Partition blocks after hypothetically cutting the N-CA bond
blocks_after_cut = msm.topology.get_covalent_blocks(lysozyme, remove_bonds=[[n_atom, ca_atom]])

print(f"Blocks count before bond cut: {len(blocks)}")
print(f"Blocks count after N-CA cut : {len(blocks_after_cut)}")
````

:::{hint}
:class: dropdown
**msm.topology.get_covalent_blocks()**: Returns sets of atoms remaining connected when specified bonds are cut. See API doc: {func}`molsysmt.topology.get_covalent_blocks`.
:::

### 2. Topological Bond Graphs

To analyze a system using graph theory, `msm.topology.get_bondgraph()` extracts the complete covalent network as a **`networkx.Graph`** object, where nodes represent atoms and edges represent chemical bonds:

````python
# Extract network topology as a NetworkX Graph
graph = msm.topology.get_bondgraph(lysozyme)

print(f"NetworkX Graph created successfully!")
print(f"Total nodes (atoms): {graph.number_of_nodes()}")
print(f"Total edges (bonds): {graph.number_of_edges()}")
````

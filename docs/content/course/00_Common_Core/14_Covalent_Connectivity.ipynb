# Module 14: Covalent Connectivity

**Welcome back, Apprentice Master.** In [Module 13: System Auditing and Curing](../00_Common_Core/13_System_Auditing_and_Curing.ipynb), you learned how to audit structural completeness and detect missing atoms or gaps. Now we explore structural topology: **Covalent Connectivity**.

Molecular systems are bound together by covalent networks. Connectivity defines how atoms group into residues, independent molecules, or polymer chains. In MolSysMT, you can extract bonded atom pairs, calculate connected covalent blocks, select atoms based on topological bonds, infer missing bonds, or convert topologies directly into graph objects.

### 1. Extracting Covalent Bonds

Let's begin by importing MolSysMT and loading our T4 Lysozyme demonstration system.

````python
import molsysmt as msm
from molsysmt import systems

# Load T4 Lysozyme as native MolSys
lysozyme = msm.convert(systems['T4 lysozyme L99A']['181l.bcif.gz'], to_form='molsysmt.MolSys')
````

You can extract the full matrix of covalent bond pairs using `msm.get()` with `bonded_atom_pairs=True`:

````python
# Retrieve bonded atom index pairs
bonds = msm.get(lysozyme, element='system', bonded_atom_pairs=True)

print(f"Total covalent bonds in system: {len(bonds)}")
print(f"First 5 bonded atom pairs:\n{bonds[:5]}")
````

### 2. Topological Analysis

A **covalent block** is a set of atoms connected to each other through covalent bonds. `msm.topology.get_covalent_blocks()` partitions the system into independent connected components (such as protein chains, ligands, or individual water molecules):

````python
# Partition system into independent covalent blocks
blocks = msm.topology.get_covalent_blocks(lysozyme)

print(f"Independent covalent blocks found: {len(blocks)}")
print(f"Atom count of primary protein block: {len(blocks[0])}")
````

:::{hint}
:class: dropdown
**msm.topology.get_covalent_blocks()**: Computes covalently connected sets of atom indices across the molecular system. See API doc: {func}`molsysmt.topology.get_covalent_blocks`.
:::

### 3. Selection by Connectivity

MolSysMT's selection language supports topological operators such as `bonded to` or `covalently bonded to`:

````python
# Select all atoms covalently bonded to atom index 10
bonded_atoms = msm.select(lysozyme, selection='all bonded to atom_index == 10')
print(f"Atoms bonded to atom 10: {bonded_atoms}")
````

### 4. Reconstructing Missing Bonds

Unbonded PDB files downloaded from databases often lack explicit CONECT records. You can reconstruct missing bonds based on standard chemical templates using `msm.build.add_missing_bonds()`:

````python
# Load raw Villin Headpiece PDB
villin_raw = msm.convert('pdb_id:1VII', to_form='molsysmt.MolSys')
print(f"Bonds before reconstruction: {msm.get(villin_raw, element='system', n_bonds=True)}")

# Reconstruct missing covalent bonds
msm.build.add_missing_bonds(villin_raw)
print(f"Bonds after reconstruction: {msm.get(villin_raw, element='system', n_bonds=True)}")
````

### 5. NetworkX Graph Conversion

For advanced topological network calculations (such as shortest path algorithms or ring detection), convert the covalent topology directly into a **`networkx.Graph`** object:

````python
# Convert molecular system to a NetworkX Graph
graph = msm.convert(lysozyme, to_form='networkx.Graph')

print(f"NetworkX Graph created | Nodes (atoms): {graph.number_of_nodes()} | Edges (bonds): {graph.number_of_edges()}")
````

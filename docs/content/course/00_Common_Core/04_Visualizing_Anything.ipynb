# Module 4: Visualizing Anything

Molecular biology is visual. If you can't see your system, you are flying blind. In this module, you will learn to use the universal viewer of MolSysMT: `view()`.

MolSysMT is the keystone of the [**MolSysSuite**](https://www.uibcdf.org/). Its native visualization tool is [**MolSysViewer**](https://www.uibcdf.org/molsysviewer) (a powerful integration of Python and Mol*). While MolSysViewer is the recommended native choice, MolSysMT also provides a seamless bridge to third-party viewers like **NGLView**. The magic is that you don't need to learn different APIs; you just use `msm.view()`.

````python
import molsysmt as msm
from molsysmt import systems
````

### 1. The Cloud Viewer
You can visualize a system directly from its PDB ID. MolSysMT will fetch it and render it for you in one line.

````python
# Let's visualize the SARS-CoV-2 Main Protease (from our Antiviral Path)
msm.view('pdb_id:6LU7')
````

### 2. Visualizing Local Files
Whether it is an H5MSM, a PDB, or an XTC file on your disk, the command is the same.

````python
# Let's visualize our training system: T4 Lysozyme
lysozyme = systems['T4 lysozyme L99A']['181l.bcif.gz']
msm.view(lysozyme)
````

### 3. Selective Visualization
Often, systems are too crowded. You can use the `selection` argument inside `view()` to focus only on what matters.

```{note}
The selection language is a powerful core feature of MolSysMT. We will explore its full syntax in **Module 5**. For now, just know it allows you to filter atoms and molecules by name, type, or property.
```

````python
# View the Lysozyme but hide the water molecules and ions
msm.view(lysozyme, selection='molecule_type=="protein"')
````

### 4. Visualizing Trajectories
If your "Form" contains multiple structures (a trajectory), `view()` automatically provides a player with a slider.

Notice in the example below how we pass a **list** of items `[topology, trajectory]`. In MolSysMT, a molecular system can be defined by the combination of multiple data forms combined.

````python
# Visualize the unfolding of the Villin Headpiece (from the Alzheimer path examples)
villin_traj = systems['chicken villin HP35']['traj_chicken_villin_HP35_solvated.dcd']
villin_topo = systems['chicken villin HP35']['chicken_villin_HP35_solvated.h5msm']

msm.view([villin_topo, villin_traj])
````

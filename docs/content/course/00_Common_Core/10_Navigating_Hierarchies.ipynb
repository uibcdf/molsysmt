# Module 10: Navigating Hierarchies (Relational Algebra)

In MolSysMT, you don't need to manually keep track of which atom belongs to which residue. The framework acts like a **Relational Database**. You can ask: "Who is the parent of this atom?" or "Who are the children of this chain?"

We call this **Relational Algebra**, and it is handled by the `get()` function.

```python
import molsysmt as msm
from molsysmt import systems

lysozyme = systems['T4 lysozyme L99A']['181l.h5msm']
```

### 1. Mapping Upwards (Atom to Residue)
Imagine you have a list of atom indices (maybe from a contact analysis or a mouse click). You want to know which residues they belong to.

```python
# Let's pick atoms 10, 50, and 100
my_atoms = [10, 50, 100]

# We ask for the residue name (group_name) of those specific atoms
res_names, res_indices = msm.get(lysozyme, element='atom', selection=my_atoms, 
                                 group_name=True, group_index=True)

print(f"Atom 10 belongs to: {res_names[0]} (index {res_indices[0]})")
print(f"Atom 50 belongs to: {res_names[1]} (index {res_indices[1]})")
```

### 2. Mapping Downwards (Entity to Atoms)
You can also go the other way. If you have a biological entity (like the ligand), you can find all the atoms that compose it.

```python
# Find the index of the entity named 'BENZENE'
benzene_index = msm.select(lysozyme, selection='entity_name=="BENZENE"', to_syntax='MolSysMT') # This returns atoms

# But wait, we can also ask 'get' for the children atoms of an entity index
atom_indices = msm.get(lysozyme, element='entity', selection='name=="BENZENE"', atom_index=True)

print(f"The Benzene entity is composed of atoms: {atom_indices}")
```

### 3. Inverse Relational Logic
One of the most powerful tricks is finding what is **NOT** part of a relationship. For example, finding all atoms that do not belong to the main protein chain.

```python
# Select atoms where their parent chain index is NOT 0
non_chain_0 = msm.select(lysozyme, selection='not chain_index==0')

print(f"Found {len(non_chain_0)} atoms outside Chain 0 (usually water, ions, and ligands).")
```

### 4. Cross-Level Summaries
You can ask for properties of one level based on another. How many groups (residues) are in each chain?

```python
chain_indices, n_groups = msm.get(lysozyme, element='chain', chain_index=True, n_groups=True)

for idx, count in zip(chain_indices, n_groups):
    print(f"Chain {idx} has {count} residues.")
```

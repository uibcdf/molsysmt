# Module 18: Merging and Growing Systems

Biological complexity often requires building a system from parts. You might have a protein in one file and a ligand in another, or you might want to join two simulation trajectories. 

MolSysMT provides a set of verbs to expand your molecular world: `merge()`, `add()`, and `concatenate_structures()`.

```python
import molsysmt as msm
from molsysmt import systems
```

### 1. Merging Systems
The function `merge()` takes two independent systems and fuses them into a single new object. Lists passed through `selections` or `structure_indices` contain one entry per system. Use a NumPy array or range when the same index collection should be applied to every input.

```python
protein = systems['chicken villin HP35']['1vii.pdb']
ligand = systems['benzamidine']['benzamidine.pdb']

# Merge both into a single system
complex = msm.merge([protein, ligand], to_form='molsysmt.MolSys')

msm.info(complex)
```

### 2. Adding Selected Elements
`add()` grows a target system with atoms selected from another system and modifies the target by default. When both systems contain structures, their structure counts must match unless `structure_indices` selects compatible structures. Atom-aligned coordinates, velocities, B factors, and occupancies are extended only when present on both sides; a one-sided series is dropped with `StructuralAttributeDropWarning` instead of creating an incomplete atom axis. A rejected native `MolSys` addition leaves both its topology and structures unchanged.

### 3. Concatenating Trajectories
If you have a simulation split into several parts, `concatenate_structures()` joins them along the time axis. The topology can live in the first system only; later XTC, DCD, or XYZ parts need only preserve its atom count and ordering. Every stored structural series must cover the complete joined structure axis. The default intersection policy drops one-sided optional series with a warning, while `attribute_policy='strict'` rejects such a concatenation without modifying the inputs.

```python
traj = systems['chicken villin HP35']['traj_chicken_villin_HP35_solvated.dcd']

# Let's simulate having two parts by concatenating the same trajectory with itself
long_traj = msm.concatenate_structures(
    [traj, traj], to_form='molsysmt.Structures'
)

n_structures_orig = msm.get(traj, element='system', n_structures=True)
n_structures_long = msm.get(long_traj, element='system', n_structures=True)

print(f"Original structures: {n_structures_orig}, Concatenated: {n_structures_long}")
```

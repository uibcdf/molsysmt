When converting file formats, extracting sub-structures, or building simulation boxes, verifying topological and structural equality between systems is critical. Did a conversion drop atoms? ### 1. Basic Topology Comparison

Let's begin by importing MolSysMT and loading our T4 Lysozyme demonstration system.

````python
import molsysmt as msm
from molsysmt import systems

# Load T4 Lysozyme file
lysozyme_file = systems['T4 lysozyme L99A']['181l.bcif.gz']

# Convert to native MolSys object
lysozyme_molsys = msm.convert(lysozyme_file, to_form='molsysmt.MolSys')
````

By default, `msm.compare()` validates whether two molecular systems have identical core attributes, regardless of their underlying data forms:

````python
# Compare the file format directly with the native MolSys object
is_identical = msm.compare(lysozyme_file, lysozyme_molsys)
print(f"Are the file and in-memory object identical? {is_identical}")
````

:::{hint}
:class: dropdown
**msm.compare()**: Universal comparison engine for checking attribute equality, topological identity, or structural consistency across molecular systems. See API doc: {func}`molsysmt.basic.compare`.
:::

### 2. Targeted Attribute Comparison

You can target specific molecular attributes by passing keyword arguments (such as `n_atoms=True` or `coordinates=True`) to focus the comparison on particular properties:

````python
# Compare specific topological counts
same_counts = msm.compare(lysozyme_file, lysozyme_molsys, n_atoms=True, n_groups=True, n_chains=True)
print(f"Do atom, group, and chain counts match? {same_counts}")
````

### 3. Detailed Difference Auditing

When comparing two non-identical systems (such as a full system versus an extracted protein sub-system), set `output_type='dictionary'` to obtain a detailed audit of matching and mismatching attributes:

````python
# Extract protein-only sub-system
protein_only = msm.extract(lysozyme_molsys, selection='molecule_type == "protein"')

# Extract detailed dictionary comparison report
report = msm.compare(lysozyme_molsys, protein_only, output_type='dictionary')

print("Attribute mismatch audit:")
for attr, matches in report.items():
    if not matches:
        print(f"- {attr}: MISMATCH")
````

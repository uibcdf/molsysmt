# Path A - Module 24: The MolSysBuilder API

Sometimes, you need to create something that doesn't exist in any file. Maybe you want to add dummy atoms to guide a simulation, or build a custom molecule from scratch. 

MolSysMT provides the **MolSysBuilder**, a declarative API that allows you to construct systems step-by-step.

```python
import molsysmt as msm
from molsysmt import systems

# Create an empty builder in editable mode
molsys = msm.convert('pdb:2BEG', to_form='molsysmt.MolSys')
```

### 1. Entering Editable Mode
To use the builder on an existing system, we call the `editable()` method. This returns a builder object tied to our system.

```python
builder = msm.build.editable(molsys)
print(f"Builder ready for system with {msm.get(molsys, element='system', n_atoms=True)} atoms.")
```

### 2. Adding Atoms and Residues Manually
Let's add a "Dummy Atom" to mark the center of the fibril core. This is useful for many computational biology protocols.

```python
# Add a new entity, molecule, chain and group for our dummy atom
builder.add_atom(atom_name='DUM', atom_type='D', coordinates=[[0.0, 0.0, 0.0]], 
                 group_name='DUM', chain_id='X', molecule_name='Marker', entity_name='Scaffold')

# The system is updated in place through the builder
print(f"Atoms after adding dummy: {msm.get(molsys, element='system', n_atoms=True)}")
```

### 3. Building from Scratch
You can also start from a completely empty system. This is the ultimate level of control.

```python
empty_molsys = msm.convert(None, to_form='molsysmt.MolSys')
new_builder = msm.build.editable(empty_molsys)

# Add two atoms (like a diatomic molecule)
new_builder.add_atom(atom_name='H1', coordinates=[[0.0, 0.0, 0.0]])
new_builder.add_atom(atom_name='H2', coordinates=[[0.0, 0.0, 0.074]]) # 0.74 Angstroms

msm.info(empty_molsys)
```

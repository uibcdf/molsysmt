# Module 9: The Hierarchical Levels

A molecular system is not just a list of atoms. It has a complex biological and logical organization. MolSysMT recognizes 6 fundamental levels, and understanding them is crucial for advanced modeling.

---

### 1. The 6 Levels of MolSysMT

1.  **Atom:** The most basic particle.
2.  **Group:** Usually a residue (amino acid, nucleotide, etc.).
3.  **Component:** A set of groups connected by covalent bonds. If a protein has two chains not connected by a disulfide bridge, they are two components.
4.  **Molecule:** A biological unit. A protein dimer can be one molecule composed of two components (chains).
5.  **Chain:** A standard structural unit often defined in PDB files.
6.  **Entity:** The unique biological identity. If you have 1000 water molecules, they are 1000 molecules but only **1 entity** (Water).

```python
import molsysmt as msm
from molsysmt import systems

lysozyme = systems['T4 lysozyme L99A']['181l.h5msm']
```

### 2. Auditing the Hierarchy with `info()`
You have already used `info()`, but now let's use it to see the count of elements at each level.

```python
# System-level summary tells you how many of each there are
msm.info(lysozyme)
```

### 3. Biological Validation: `contains()` and `is_composed_of()`
These functions allow you to ask logical questions about your system.

```python
# Does this system contain any proteins?
print(f"Has protein? {msm.contains(lysozyme, 'molecule_type=="protein"')}")

# Is the system ONLY composed of water and ions? (This should be False)
print(f"Is it just solvent? {msm.is_composed_of(lysozyme, water=True, ion=True)}")

# Is it composed of protein, small molecules, water, and ions?
print(f"Is it the full complex? {msm.is_composed_of(lysozyme, protein=1, small_molecule=1, water=True, ion=True)}")
```

### 4. Exploring Entities
Entities are the biological "templates". Let's see the unique entities in the Lysozyme system.

```python
msm.info(lysozyme, element='entity')
```

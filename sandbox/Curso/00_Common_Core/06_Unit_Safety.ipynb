# Module 6: Unit Safety with PyUnitWizard

In Computational Biology, numbers without units are dangerous. Is that distance in Angstroms or Nanometers? Is that time in Picoseconds or Nanoseconds? Confusing them can ruin months of work.

MolSysMT uses a dedicated library called **PyUnitWizard** (imported as `puw`) to ensure that every physical property is treated as a **Quantity** (Value + Unit).

```python
import molsysmt as msm
from molsysmt import systems
from molsysmt import pyunitwizard as puw

lysozyme = systems['T4 lysozyme L99A']['181l.h5msm']
```

### 1. Retrieving Quantities
When you use `get()` to retrieve structural properties like coordinates or box dimensions, MolSysMT doesn't just give you a list of numbers; it gives you a Quantity object.

```python
# Let's get the coordinates of the first 5 atoms
coords = msm.get(lysozyme, selection='atom_index==[0,1,2,3,4]', coordinates=True)

print(f"Type of coords: {type(coords)}")
print(f"Coordinates with units:\n{coords}")
```

### 2. Extracting Value and Unit
You can separate the numeric value from its unit using `puw.get_value()` and `puw.get_unit()`.

```python
value = puw.get_value(coords)
unit = puw.get_unit(coords)

print(f"Numeric Value:\n{value}")
print(f"Unit: {unit}")
```

### 3. Converting Units
The power of PyUnitWizard is that it handles the math for you. You can convert any quantity to your preferred unit seamlessly.

```python
# Convert Nanometers to Angstroms
coords_A = puw.convert(coords, to_unit='angstroms')

print(f"In Angstroms:\n{coords_A}")
```

### 4. Creating your own Quantities
You can also create your own physical quantities to use in MolSysMT functions.

```python
my_distance = puw.quantity(1.5, 'nm')
print(f"My custom distance: {my_distance}")
```

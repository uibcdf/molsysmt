# Module 8: Working with Lists (The Piped Model)

In the real world of Computational Biology, your data is rarely in a single file. You might have a **PDB** file with the names of the atoms (topology) and a massive **XTC** or **DCD** file with the movement of those atoms (trajectory).

Traditionally, you would need to load both into a specific object or merge them. MolSysMT uses the **Piped Model**: you can just pass a list of files, and the framework will pipe the information from wherever it is available.

```python
import molsysmt as msm
from molsysmt import systems
```

### 1. The Separation of Powers
Let's take the **Villin Headpiece** system. We have its topology in one file and its trajectory in another.

```python
topology_file = systems['chicken villin HP35']['chicken_villin_HP35_solvated.h5msm']
trajectory_file = systems['chicken villin HP35']['traj_chicken_villin_HP35_solvated.dcd']

print(f"Topology source: {msm.get_form(topology_file)}")
print(f"Trajectory source: {msm.get_form(trajectory_file)}")
```

### 2. The Piped Input
Now, instead of passing one object, we pass a **list** `[topology, trajectory]`. MolSysMT will see this as a single virtual system.

```python
# We can visualize the combined system
msm.view([topology_file, trajectory_file])
```

### 3. Transparent Queries
When you ask for an attribute, MolSysMT will look for it in the list. If you ask for `atom_name`, it will take it from the topology file. If you ask for `coordinates`, it will take them from the trajectory file.

```python
virtual_system = [topology_file, trajectory_file]

# This name comes from the H5MSM file
name = msm.get(virtual_system, selection='atom_index==0', atom_name=True)

# These coordinates come from the DCD file
coords = msm.get(virtual_system, selection='atom_index==0', coordinates=True)

print(f"Atom name: {name}")
print(f"Coordinates shape: {coords.shape} (Time, Atoms, Space)")
```

### 4. Debugging the Pipe
Remember `where_is_attribute()`? It is incredibly useful here to verify which file is providing which piece of data.

```python
print(f"Source of names: {msm.where_is_attribute(virtual_system, 'atom_name')}")
print(f"Source of coordinates: {msm.where_is_attribute(virtual_system, 'coordinates')}")
```

# Module 3: Selection Language Basics

To analyze or edit a system, you first need to be able to pick the parts you are interested in. In MolSysMT, we do this with the `select()` function and its powerful, human-readable language.

In this module, we will use our training system: the **T4 Lysozyme**.

```python
import molsysmt as msm
from molsysmt import systems

lysozyme = systems['T4 lysozyme L99A']['181l.h5msm']
```

### 1. The `select()` Function
The function `select()` always returns a **list of atom indices** (0-based). If you don't provide a selection string, it defaults to `'all'`.

```python
all_atoms = msm.select(lysozyme, selection='all')
print(f"The system has {len(all_atoms)} atoms.")
print(f"First 5 indices: {all_atoms[:5]}")
```

### 2. Biological Keywords
MolSysMT understands biological concepts. You don't need to know the atom names to pick the backbone or the solvent.

```python
protein_atoms = msm.select(lysozyme, selection='molecule_type=="protein"')
water_atoms = msm.select(lysozyme, selection='molecule_type=="water"')
backbone_atoms = msm.select(lysozyme, selection='backbone')

print(f"Protein: {len(protein_atoms)} atoms")
print(f"Water: {len(water_atoms)} atoms")
print(f"Backbone: {len(backbone_atoms)} atoms")
```

### 3. Selecting by Name and ID
You can be very specific by using residue names (`group_name`), atom names, or residue IDs (`group_id`).

```python
# Select all Carbons Alpha (CA)
ca_atoms = msm.select(lysozyme, selection='atom_name=="CA"')

# Select all atoms in residue 15
res_15 = msm.select(lysozyme, selection='group_id==15')

# Select all Lysines
lys_atoms = msm.select(lysozyme, selection='group_name=="LYS"')

print(f"There are {len(ca_atoms)} C-alphas and {len(lys_atoms)} Lysine atoms.")
```

### 4. Logical Operators
You can combine criteria using `and`, `or`, and `not`.

```python
# Select C-alphas ONLY if they belong to a Lysine
lys_ca = msm.select(lysozyme, selection='(atom_name=="CA") and (group_name=="LYS")')

# Select everything EXCEPT water
no_water = msm.select(lysozyme, selection='not molecule_type=="water"')

print(f"Found {len(lys_ca)} Lysine C-alphas.")
```

### 5. Using Selections in View
Remember the previous module? You can pass these strings directly to `view()` to see your selection in 3D.

```python
msm.view(lysozyme, selection='group_name=="LYS"')
```

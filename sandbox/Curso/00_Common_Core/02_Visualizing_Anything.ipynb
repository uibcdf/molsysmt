# Module 2: Visualizing Anything

Molecular biology is visual. If you can't see your system, you are flying blind. In this module, you will learn to use the universal viewer of MolSysMT: `view()`.

MolSysMT doesn't have its own 3D engine from scratch; instead, it provides a seamless bridge to powerful viewers like **NGLView** or **MolSysViewer**. The magic is that you don't need to learn their specific APIs; you just use `msm.view()`.

```python
import molsysmt as msm
from molsysmt import systems
```

### 1. The Cloud Viewer
You can visualize a system directly from its PDB ID. MolSysMT will fetch it and render it for you in one line.

```python
# Let's visualize the SARS-CoV-2 Main Protease (from our Antiviral Path)
msm.view('pdb:6LU7')
```

### 2. Visualizing Local Files
Whether it is an H5MSM, a PDB, or an XTC file on your disk, the command is the same.

```python
# Let's visualize our training system: T4 Lysozyme
lysozyme = systems['T4 lysozyme L99A']['181l.h5msm']
msm.view(lysozyme)
```

### 3. Selective Visualization
Often, systems are too crowded. You can use the `selection` argument inside `view()` to focus only on what matters. (Don't worry, we will master the selection language in the next module!).

```python
# View the Lysozyme but hide the water molecules and ions
msm.view(lysozyme, selection='molecule_type=="protein"')
```

### 4. Visualizing Trajectories
If your "Form" contains multiple structures (a trajectory), `view()` automatically provides a player with a slider.

```python
# Visualize the unfolding of the Villin Headpiece (from the Alzheimer path examples)
villin_traj = systems['chicken villin HP35']['traj_chicken_villin_HP35_solvated.dcd']
villin_topo = systems['chicken villin HP35']['chicken_villin_HP35_solvated.h5msm']

msm.view([villin_topo, villin_traj])
```

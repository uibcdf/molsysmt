# Module 13: Covalent Connectivity (The Graph)

How does MolSysMT know that two atoms belong to the same molecule? The answer is in the **Covalent Graph**. By looking at which atoms are connected by chemical bonds, the framework can reconstruct the physical topology of the system.

```python
import molsysmt as msm
from molsysmt import systems

lysozyme = systems['T4 lysozyme L99A']['181l.h5msm']
```

### 1. The Bond List
You can retrieve the list of all chemical bonds in the system. Each bond is defined by a pair of atom indices.

```python
# Get all bonds
bonds = msm.get(lysozyme, element='bond', bonded_atom_pairs=True)

print(f"The system has {len(bonds)} covalent bonds.")
print(f"First 5 bonds (atom pairs):\n{bonds[:5]}")
```

### 2. The Bond Graph
The function `get_bondgraph()` returns a representation of the entire connectivity network. This is essentially a mathematical graph where nodes are atoms and edges are bonds.

```python
graph = msm.get_bondgraph(lysozyme)
print(f"Bond graph type: {type(graph)}")
```

### 3. Covalent Blocks
MolSysMT can partition the system into **Covalent Blocks**. A covalent block is a set of atoms where every atom can reach any other atom in the set through chemical bonds. 

In a protein-ligand system, the protein is one block, the ligand is another, and each water molecule is a tiny block of 3 atoms.

```python
# Get all independent covalent blocks
blocks = msm.get_covalent_blocks(lysozyme)

print(f"Found {len(blocks)} independent covalent blocks.")

# Let's see the size of the first 5 blocks
for i in range(5):
    print(f"Block {i} has {len(blocks[i])} atoms.")
```

### 4. Covalent Chains
Similarly, `get_covalent_chains()` identifies continuous chains of bonded atoms (like the backbone of a protein or a single-stranded DNA).

```python
covalent_chains = msm.get_covalent_chains(lysozyme)
print(f"There are {len(covalent_chains)} covalent chains.")
```

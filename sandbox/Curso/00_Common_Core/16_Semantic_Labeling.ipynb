# Module 16: Semantic Labeling

Computers love indices (0, 1, 2...), but humans love names. If you are writing a paper or a report, saying "Carbon Alpha of Lysine 15" is much better than "Atom 142".

MolSysMT has a semantic labeling engine that converts any selection into a professional nomenclature string.

```python
import molsysmt as msm
from molsysmt import systems

lysozyme = systems['T4 lysozyme L99A']['181l.h5msm']
```

### 1. The Default Label
The function `get_label()` provides a standard, unambiguous string for any atom or residue.

```python
# Get the label of the first atom
label = msm.get_label(lysozyme, selection='atom_index==0')
print(f"Default atom label: {label}")

# Get the label of a whole residue
res_label = msm.get_label(lysozyme, element='group', selection='group_index==10')
print(f"Default residue label: {res_label}")
```

### 2. Batch Labeling
You can generate labels for a whole list of indices at once. This is perfect for generating CSV tables.

```python
# Label all C-alphas of the first 5 residues
ca_labels = msm.get_label(lysozyme, selection='atom_name=="CA" and group_index==[0,1,2,3,4]')
print("C-alpha Labels:")
for l in ca_labels:
    print(f" - {l}")
```

### 3. Customizing Labels (Optional Deep Dive)
MolSysMT allows you to define your own string templates if you need a specific format (e.g., for a specific software input).

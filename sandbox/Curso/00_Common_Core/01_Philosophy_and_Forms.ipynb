# Module 1: The Philosophy of "Forms"

Welcome, Apprentice Master. Your journey starts here.

The most important concept in **MolSysMT** is the **Form**. In traditional Computational Biology, you are often a prisoner of your file format: if you have a PDB file, you use one library; if you have an OpenMM object, you use another. 

MolSysMT breaks these chains. It treats every molecular data structure as a **Form**. Whether it is a file on disk, an object in memory, or a string of text, MolSysMT sees them all as "Molecular Systems".

---

### 1. Importing the Tools
First, we bring the framework and its internal database into our environment.

```python
import molsysmt as msm
from molsysmt import systems
```

### 2. The Universal Sampler: 4 Systems, 4 Paths
To demonstrate that MolSysMT doesn't care about the size or the source, let's look at the four systems that define our specialized paths. Notice how we use the same functions for all of them.

```python
# 🔬 Path A: Alzheimer (Amyloid-Beta Fibril from PDB ID)
alzheimer_system = 'pdb:2BEG'

# ♻️ Path B: Enzyme Engineering (Triosephosphate Isomerase from internal database)
enzyme_system = systems['TcTIM']['1tcd.h5msm']

# 💊 Path C: Antiviral Hunter (Barnase-Barstar Complex)
antiviral_system = systems['Barnase-Barstar']['barnase_barstar.h5msm']

# ⚡ Path D: Nano-Mechanic (POPC Membrane)
membrane_system = systems['POPC membrane']['popc_membrane.dcd']

systems_list = [alzheimer_system, enzyme_system, antiviral_system, membrane_system]
names = ["Alzheimer Fibril", "Industrial Enzyme", "Antiviral Complex", "Cellular Membrane"]
```

### 3. Identifying the Forms
The function `get_form()` is the "eye" of MolSysMT. It tells us exactly what kind of object we are dealing with.

```python
for name, sys in zip(names, systems_list):
    form = msm.get_form(sys)
    print(f"{name} is currently in form: {form}")
```

### 4. The Power of Conversion
You are never stuck in a form. You can `convert()` any system into another supported form. Let's convert the Industrial Enzyme (currently an H5MSM file) into a native MolSys object.

```python
molsys_enzyme = msm.convert(enzyme_system, to_form='molsysmt.MolSys')

print(f"The Enzyme is now a: {msm.get_form(molsys_enzyme)}")
print(f"Is it a valid molecular system? {msm.is_a_molecular_system(molsys_enzyme)}")
```

### 5. Our Training Ground: T4 Lysozyme
While we will keep an eye on the four paths, the **Common Core** will primarily use the **T4 Lysozyme (PDB 181L)**. It is a robust and well-understood model, perfect for learning the basics.

```python
lysozyme = systems['T4 lysozyme L99A']['181l.h5msm']
print(f"Training system ready: {msm.get_form(lysozyme)}")
```

# Path C - Module 25: PDB Bioassemblies & AltLocs

High-resolution Mpro structures often have disordered regions in the binding pocket. In this module, you will learn to choose the correct crystallographic locations to avoid clashing with your inhibitor.

```python
import molsysmt as msm

# Load Mpro (6LU7) and solve AltLocs
molsys = msm.convert('pdb:6LU7', to_form='molsysmt.MolSys')
msm.build.solve_atoms_with_alternate_location(molsys)

print("Binding site cleaned from AltLoc ambiguities.")
```

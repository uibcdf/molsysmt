# Module 1: The Form-Agnostic Philosophy

Welcome to the first module of the **Introduction to MolSysMT** course. In this unit, we will explore the core philosophy of MolSysMT: **Form Agnosticism**. 

Traditionally, molecular analysis tools are tied to specific data structures (classes, objects, or file formats). If you have an OpenMM Topology, you use OpenMM tools; if you have an MDTraj Trajectory, you use MDTraj tools. MolSysMT acts as a **Molecular Form Manager**, allowing you to perform the same operations regardless of whether your input is a PDB file, a Python dictionary, or an object from another library.

### 1. Identifying the Form

The first step is knowing what we have. The function `get_form()` tells us the nature of our molecular system.

```python
import molsysmt as msm
from molsysmt import systems

# Let's take a PDB file from the internal database
pdb_file = systems['pentalanine']['pentalanine.pdb']

print(f"The form of the item is: {msm.get_form(pdb_file)}")
```

### 2. Validation

Before operating, we can check if an object is recognized as a valid molecular system by MolSysMT using `is_a_molecular_system()`.

```python
print(f"Is it a molecular system? {msm.is_a_molecular_system(pdb_file)}")
```

### 3. Universal Conversion

The power of MolSysMT lies in `convert()`. We can transform any form into another supported form seamlessly.

```python
# Convert PDB file path to a native MolSys object
molsys = msm.convert(pdb_file, to_form='molsysmt.MolSys')

print(f"New form: {msm.get_form(molsys)}")
```

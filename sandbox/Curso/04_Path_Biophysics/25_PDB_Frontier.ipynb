# Path D - Module 25: PDB Bioassemblies & AltLocs

Membrane crystals are notoriously messy. In this module, you will learn to use `make_bioassembly()` to generate the full functional tetramer of a channel from a single asymmetric chain.

```python
import molsysmt as msm
molsys = msm.convert('pdb:1K4C', to_form='molsysmt.MolSys')
full_tetramer = msm.build.make_bioassembly(molsys)

print(f"Tetramer built with {msm.get(full_tetramer, element='chain', n_chains=True)} chains.")
```

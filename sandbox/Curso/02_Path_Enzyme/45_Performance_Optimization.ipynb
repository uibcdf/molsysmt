# Path B - Module 45: Performance Optimization

MolSysMT uses **Numba (JIT)** to achieve C-like speeds in Python. However, the first time you call a function, there is a small delay because the code is being compiled. 

In this module, you will learn to use `warmup_numba()` to ensure your industrial pipeline is at maximum speed from the very first mutant.

```python
import molsysmt as msm
import time

print("Triggering Numba Warmup...")
start = time.time()
msm.warmup_numba()
end = time.time()
print(f"Warmup completed in {end - start:.2f} seconds. All kernels are now pre-compiled.")
```

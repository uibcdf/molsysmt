```python
import molsysmt as msm
```

```python
molsys = msm.convert('1TCD')
```

```python
msm.info(molsys)
```

```python
molsys = msm.remove(molsys, selection="molecule_type=='water'")
```

```python
msm.get(molsys, n_molecules=True, n_atoms=True)
```

```python
msm.select(molsys, selection="(atom_name in ['CA', 'CB'] and chain_index==0) within 3.5 angstroms of chain_index==1")
```

```python
monomer = msm.extract(molsys, selection='molecule_index==0')
```

```python
msm.build.has_hydrogens(monomer)
```

```python
monomer = msm.build.add_missing_hydrogens(monomer, pH=7.4)
```

```python
peptide = msm.build.build_peptide('AGTRL')
```

```python
msm.structure.translate(peptide, translation='[1,1,1] nm', in_place=True)
```

```python
new_molsys = msm.merge([monomer, peptide])
```

```python
msm.structure.get_contacts(new_molsys, selection='atom_name=="CA"', threshold='9 angstroms')
```

```python
msm.convert(new_molsys, 'new_system.pdb')
```
